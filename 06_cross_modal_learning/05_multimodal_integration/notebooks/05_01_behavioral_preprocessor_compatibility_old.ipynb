```python
# ============================================================
# CELL 79 — FINAL BEHAVIORAL PREPROCESSING INTEGRITY TEST
# ============================================================

from pathlib import Path
import subprocess
import textwrap

PROJECT_ROOT = Path(
    r"C:\Users\shasw\Documents\GitHub\GAN-Enhanced-Multimodal-ASD-Screening"
)

PYTHON_EXE = (
    PROJECT_ROOT
    / ".venv_behavioral_161"
    / "Scripts"
    / "python.exe"
)

PREPROCESSOR_PATH = (
    PROJECT_ROOT
    / "Behavioral_Model"
    / "02_preprocessing"
    / "behavioral_preprocessor.joblib"
)

PROCESSED_DATA_PATH = (
    PROJECT_ROOT
    / "Behavioral_Model"
    / "02_preprocessing"
    / "behavioral_train_processed.csv"
)

FEATURE_NAMES_PATH = (
    PROJECT_ROOT
    / "Behavioral_Model"
    / "02_preprocessing"
    / "behavioral_processed_feature_names.txt"
)

print("=" * 70)
print("FINAL BEHAVIORAL PREPROCESSING INTEGRITY TEST")
print("=" * 70)

# ------------------------------------------------------------
# Everything below runs inside Python 3.12 / sklearn 1.6.1
# ------------------------------------------------------------

test_code = textwrap.dedent(f"""
    import joblib
    import pandas as pd
    import numpy as np

    PREPROCESSOR_PATH = r"{PREPROCESSOR_PATH}"
    PROCESSED_DATA_PATH = r"{PROCESSED_DATA_PATH}"
    FEATURE_NAMES_PATH = r"{FEATURE_NAMES_PATH}"

    print("\\nLoading original preprocessor...")
    preprocessor = joblib.load(PREPROCESSOR_PATH)

    print("Preprocessor loaded successfully.")

    # --------------------------------------------------------
    # Load existing processed dataset
    # --------------------------------------------------------

    processed_df = pd.read_csv(PROCESSED_DATA_PATH)

    print("\\nExisting processed dataset shape:")
    print(processed_df.shape)

    # --------------------------------------------------------
    # Load the original 87 feature names
    # --------------------------------------------------------

    with open(FEATURE_NAMES_PATH, "r", encoding="utf-8") as f:
        feature_names = [
            line.strip()
            for line in f
            if line.strip()
        ]

    print("\\nStored processed feature count:")
    print(len(feature_names))

    # --------------------------------------------------------
    # Raw columns expected by the original preprocessor
    # --------------------------------------------------------

    raw_columns = [
        "A1_Score",
        "A2_Score",
        "A3_Score",
        "A4_Score",
        "A5_Score",
        "A6_Score",
        "A7_Score",
        "A8_Score",
        "A9_Score",
        "A10_Score",
        "age",
        "gender",
        "jundice",
        "austim",
        "used_app_before",
        "ethnicity",
        "contry_of_res",
        "relation"
    ]

    # --------------------------------------------------------
    # Check whether these raw columns are available in the
    # processed CSV.
    #
    # The processed CSV itself is NOT raw input, so we only
    # perform the structural checks here.
    # --------------------------------------------------------

    print("\\nOriginal preprocessor input count:")
    print(preprocessor.n_features_in_)

    print("\\nExpected original input count:")
    print(len(raw_columns))

    # --------------------------------------------------------
    # Verify stored processed feature names
    # --------------------------------------------------------

    processed_feature_columns = [
        column
        for column in processed_df.columns
        if column != "Class/ASD"
    ]

    print("\\nProcessed feature columns:")
    print(len(processed_feature_columns))

    print("\\nTarget column present:")
    print("Class/ASD" in processed_df.columns)

    # --------------------------------------------------------
    # Structural verification
    # --------------------------------------------------------

    checks = []

    checks.append(
        ("Preprocessor expects 18 inputs",
         preprocessor.n_features_in_ == 18)
    )

    checks.append(
        ("Stored feature list has 87 features",
         len(feature_names) == 87)
    )

    checks.append(
        ("Processed CSV has 87 feature columns",
         len(processed_feature_columns) == 87)
    )

    checks.append(
        ("Processed CSV contains Class/ASD",
         "Class/ASD" in processed_df.columns)
    )

    print("\\n" + "=" * 70)
    print("VERIFICATION RESULTS")
    print("=" * 70)

    all_passed = True

    for name, passed in checks:
        print(f"{{name}}: {{'PASSED' if passed else 'FAILED'}}")

        if not passed:
            all_passed = False

    print("\\n" + "=" * 70)

    if all_passed:
        print("GROUP 5 BEHAVIORAL PREPROCESSING CHECK: PASSED")
    else:
        print("GROUP 5 BEHAVIORAL PREPROCESSING CHECK: FAILED")

    print("=" * 70)
""")

result = subprocess.run(
    [
        str(PYTHON_EXE),
        "-c",
        test_code
    ],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print("\nSTDERR:")
    print(result.stderr)

print("\nReturn code:", result.returncode)
```
