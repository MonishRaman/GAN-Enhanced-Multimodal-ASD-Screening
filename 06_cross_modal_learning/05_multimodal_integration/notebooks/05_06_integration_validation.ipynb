```python
# ============================================================
# CELL 7 — LOAD 05.02 MODALITY INFERENCE OUTPUTS
# ============================================================

genetic_128 = pd.read_csv(
    REPORT_0502 / "05_02_genetic_128d_embeddings.csv"
)

behavioral_128 = pd.read_csv(
    REPORT_0502 / "05_02_behavioral_128d_embeddings.csv"
)

environmental_128 = pd.read_csv(
    REPORT_0502 / "05_02_environmental_128d_embeddings.csv"
)

print("Genetic 128-D shape      :", genetic_128.shape)
print("Behavioral 128-D shape  :", behavioral_128.shape)
print("Environmental 128-D shape:", environmental_128.shape)
```

```python
# ============================================================
# CELL 8 — 128-D REPRESENTATION VALIDATION
# ============================================================

representation_tables = {
    "Genetic": genetic_128,
    "Behavioral": behavioral_128,
    "Environmental": environmental_128
}

representation_dimensions = {}

for modality, df in representation_tables.items():

    numeric_columns = df.select_dtypes(
        include=[np.number]
    ).columns

    # Exclude obvious index/target columns
    embedding_columns = [
        column
        for column in numeric_columns
        if column not in [
            "Target",
            "Class",
            "ASD",
            "Sample_Index",
            "Original_Index"
        ]
    ]

    representation_dimensions[modality] = len(
        embedding_columns
    )

    print(
        f"{modality:<15} : "
        f"{len(embedding_columns)} embedding dimensions"
    )

if any(
    dimension != 128
    for dimension in representation_dimensions.values()
):
    raise AssertionError(
        "One or more modality representations are not 128-D."
    )

print("\n128-D REPRESENTATION VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 9 — LOAD 05.03 COMMON PROJECTION OUTPUTS
# ============================================================

genetic_32 = pd.read_csv(
    REPORT_0503 / "05_03_genetic_32d_projection.csv"
)

behavioral_32 = pd.read_csv(
    REPORT_0503 / "05_03_behavioral_32d_projection.csv"
)

environmental_32 = pd.read_csv(
    REPORT_0503 / "05_03_environmental_32d_projection.csv"
)

print("Genetic 32-D shape       :", genetic_32.shape)
print("Behavioral 32-D shape   :", behavioral_32.shape)
print("Environmental 32-D shape:", environmental_32.shape)
```

```python
# ============================================================
# CELL 10 — 32-D COMMON SPACE VALIDATION
# ============================================================

projection_tables = {
    "Genetic": genetic_32,
    "Behavioral": behavioral_32,
    "Environmental": environmental_32
}

projection_dimensions = {}

for modality, df in projection_tables.items():

    numeric_columns = df.select_dtypes(
        include=[np.number]
    ).columns

    projection_columns = [
        column
        for column in numeric_columns
        if column not in [
            "Target",
            "Class",
            "ASD",
            "Sample_Index",
            "Original_Index"
        ]
    ]

    projection_dimensions[modality] = len(
        projection_columns
    )

    print(
        f"{modality:<15} : "
        f"{len(projection_columns)} projection dimensions"
    )

if any(
    dimension != 32
    for dimension in projection_dimensions.values()
):
    raise AssertionError(
        "One or more modalities are not projected to 32-D."
    )

print("\n32-D COMMON SPACE VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 11 — INTEGRATION SAMPLE COUNT VALIDATION
# ============================================================

# IMPORTANT:
# The Phase 05.03 projection artifacts used by the integration
# pipeline may contain the integration/demo inference population
# rather than the complete original modality datasets.
#
# Therefore, this validation checks:
# 1. The 128-D and 32-D outputs have matching row counts
#    within each modality.
# 2. The row count is non-zero.
# 3. We do NOT incorrectly require the integration artifact
#    to contain the full original dataset size.

sample_count_validation = []

for modality in [
    "Genetic",
    "Behavioral",
    "Environmental"
]:

    representation_count = len(
        representation_tables[modality]
    )

    projection_count = len(
        projection_tables[modality]
    )

    print(
        f"\n{modality}"
    )

    print(
        f"128-D representation rows : "
        f"{representation_count}"
    )

    print(
        f"32-D projection rows      : "
        f"{projection_count}"
    )

    if representation_count == 0:
        raise AssertionError(
            f"{modality} 128-D representation is empty."
        )

    if projection_count == 0:
        raise AssertionError(
            f"{modality} 32-D projection is empty."
        )

    if representation_count != projection_count:
        raise AssertionError(
            f"{modality} row count changed between "
            f"05.02 and 05.03: "
            f"{representation_count} -> "
            f"{projection_count}"
        )

    sample_count_validation.append(
        {
            "modality": modality,
            "representation_128d_rows": representation_count,
            "projection_32d_rows": projection_count,
            "row_count_preserved": True
        }
    )

sample_count_validation_df = pd.DataFrame(
    sample_count_validation
)

print("\n")
display(sample_count_validation_df)

print(
    "\nSAMPLE COUNT PRESERVATION VALIDATION: PASSED"
)
```

```python
# ============================================================
# CELL 12 — NUMERICAL INTEGRITY VALIDATION
# ============================================================

for modality, df in projection_tables.items():

    numeric_values = df.select_dtypes(
        include=[np.number]
    ).to_numpy(
        dtype=float
    )

    if not np.isfinite(
        numeric_values
    ).all():

        raise AssertionError(
            f"{modality} contains NaN or Inf values."
        )

    print(
        f"{modality:<15}: "
        f"finite numerical values PASSED"
    )

print("\nNUMERICAL INTEGRITY VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 13 — LOAD 05.04 CROSS-MODAL INFERENCE
# ============================================================

cross_modal_predictions = pd.read_csv(
    REPORT_0504 / "05_04_cross_modal_predictions.csv"
)

cross_modal_metrics = pd.read_csv(
    REPORT_0504 / "05_04_modality_combination_metrics.csv"
)

with open(
    REPORT_0504 / "05_04_cross_modal_inference_report.json",
    "r",
    encoding="utf-8"
) as f:

    cross_modal_report = json.load(f)

print(
    "Cross-modal prediction rows:",
    len(cross_modal_predictions)
)

print(
    "Modality combination rows:",
    len(cross_modal_metrics)
)

print(
    "Report status:",
    cross_modal_report.get("status")
)
```

```python
# ============================================================
# CELL 14 — CROSS-MODAL INFERENCE VALIDATION
# ============================================================

expected_cross_modal_rows = 448

if len(cross_modal_predictions) != expected_cross_modal_rows:

    raise AssertionError(
        "05.04 prediction row count is incorrect."
    )

if len(cross_modal_metrics) != 3:

    raise AssertionError(
        "05.04 must contain three modality combinations."
    )

if cross_modal_report.get("status") != "PASSED":

    raise AssertionError(
        "05.04 report does not have PASSED status."
    )

required_prediction_columns = [
    "source_modality",
    "target_modality",
    "pair_label"
]

missing_columns = [
    column
    for column in required_prediction_columns
    if column not in cross_modal_predictions.columns
]

if missing_columns:

    raise AssertionError(
        f"Missing 05.04 columns: {missing_columns}"
    )

print("05.04 prediction count : PASSED")
print("05.04 modality count   : PASSED")
print("05.04 report status    : PASSED")
print("05.04 schema           : PASSED")

print("\n05.04 CROSS-MODAL VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 15 — 05.04 BASELINE METRIC VALIDATION
# ============================================================

expected_auc = 0.6626674107
expected_accuracy = 0.6116071429
expected_f1 = 0.6200873362

reported_auc = float(
    cross_modal_report["overall_metrics"]["roc_auc"]
)

reported_accuracy = float(
    cross_modal_report["overall_metrics"]["accuracy"]
)

reported_f1 = float(
    cross_modal_report["overall_metrics"]["f1"]
)

print("Reported ROC-AUC :", reported_auc)
print("Reported Accuracy:", reported_accuracy)
print("Reported F1      :", reported_f1)

if not np.isclose(
    reported_auc,
    expected_auc,
    atol=1e-6
):
    raise AssertionError(
        "05.04 ROC-AUC does not match the verified result."
    )

if not np.isclose(
    reported_accuracy,
    expected_accuracy,
    atol=1e-6
):
    raise AssertionError(
        "05.04 accuracy does not match the verified result."
    )

if not np.isclose(
    reported_f1,
    expected_f1,
    atol=1e-6
):
    raise AssertionError(
        "05.04 F1 does not match the verified result."
    )

print("\n05.04 BASELINE METRIC VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 16 — LOAD 05.05 MULTIMODAL FUSION OUTPUTS
# ============================================================

three_modal_fusion = pd.read_csv(
    REPORT_0505
    / "05_05_three_modal_fusion_results.csv"
)

flexible_fusion = pd.read_csv(
    REPORT_0505
    / "05_05_flexible_multimodal_results.csv"
)

with open(
    REPORT_0505
    / "05_05_multimodal_fusion_report.json",
    "r",
    encoding="utf-8"
) as f:

    fusion_report = json.load(f)

print(
    "Three-modal rows:",
    len(three_modal_fusion)
)

print(
    "Flexible rows:",
    len(flexible_fusion)
)

print(
    "05.05 report status:",
    fusion_report.get("status")
)
```

```python
# ============================================================
# CELL 17 — MULTIMODAL FUSION VALIDATION
# ============================================================

if len(three_modal_fusion) != 5:

    raise AssertionError(
        "Three-modal fusion must contain 5 demonstration rows."
    )

if len(flexible_fusion) != 35:

    raise AssertionError(
        "Flexible fusion must contain 35 rows."
    )

if fusion_report.get("status") != "PASSED":

    raise AssertionError(
        "05.05 report status is not PASSED."
    )

required_three_modal_columns = [
    "patient_id",
    "genetic_probability",
    "behavioral_probability",
    "environmental_probability",
    "final_probability",
    "final_prediction"
]

missing_columns = [
    column
    for column in required_three_modal_columns
    if column not in three_modal_fusion.columns
]

if missing_columns:

    raise AssertionError(
        f"Missing three-modal columns: {missing_columns}"
    )

print("05.05 three-modal rows : PASSED")
print("05.05 flexible rows    : PASSED")
print("05.05 report status    : PASSED")
print("05.05 schema           : PASSED")

print("\n05.05 MULTIMODAL FUSION VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 18 — FUSION PROBABILITY VALIDATION
# ============================================================

three_modal_probabilities = (
    three_modal_fusion[
        "final_probability"
    ]
    .to_numpy(dtype=float)
)

flexible_probabilities = (
    flexible_fusion[
        "final_probability"
    ]
    .to_numpy(dtype=float)
)

for name, values in [
    ("Three-modal", three_modal_probabilities),
    ("Flexible", flexible_probabilities)
]:

    if not np.isfinite(values).all():

        raise AssertionError(
            f"{name} fusion contains NaN or Inf."
        )

    if (
        (values < 0).any()
        or
        (values > 1).any()
    ):

        raise AssertionError(
            f"{name} fusion probability outside [0,1]."
        )

    print(
        f"{name:<15}: probability range PASSED"
    )

print("\nFUSION PROBABILITY VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 19 — INDEPENDENT FUSION FORMULA VALIDATION
# ============================================================

GENETIC_WEIGHT = 0.319370
BEHAVIORAL_WEIGHT = 0.400219
ENVIRONMENTAL_WEIGHT = 0.280411

independent_fusion = (
    GENETIC_WEIGHT
    * three_modal_fusion["genetic_probability"]
    +
    BEHAVIORAL_WEIGHT
    * three_modal_fusion["behavioral_probability"]
    +
    ENVIRONMENTAL_WEIGHT
    * three_modal_fusion["environmental_probability"]
)

stored_fusion = (
    three_modal_fusion[
        "final_probability"
    ]
)

maximum_difference = (
    independent_fusion
    - stored_fusion
).abs().max()

print(
    "Maximum independent formula difference:",
    f"{maximum_difference:.12f}"
)

if not np.isclose(
    maximum_difference,
    0.0,
    atol=1e-6
):

    raise AssertionError(
        "Stored fusion probabilities do not match "
        "the verified weighted fusion formula."
    )

print("\nINDEPENDENT FUSION FORMULA VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 20 — DECISION THRESHOLD VALIDATION
# ============================================================

DECISION_THRESHOLD = 0.50

recomputed_predictions = (
    three_modal_fusion[
        "final_probability"
    ]
    >= DECISION_THRESHOLD
).astype(int)

stored_predictions = (
    three_modal_fusion[
        "final_prediction_numeric"
    ]
    .astype(int)
)

prediction_difference = (
    recomputed_predictions
    !=
    stored_predictions
).sum()

print(
    "Prediction differences:",
    prediction_difference
)

if prediction_difference != 0:

    raise AssertionError(
        "Stored predictions do not match "
        "the 0.50 decision threshold."
    )

print("\nDECISION THRESHOLD VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 21 — FINAL DEMONSTRATION PATIENT VALIDATION
# ============================================================

expected_patients = {
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
}

actual_patients = set(
    three_modal_fusion[
        "patient_id"
    ]
)

print("Expected patients:", expected_patients)
print("Actual patients  :", actual_patients)

if actual_patients != expected_patients:

    raise AssertionError(
        "Three-modal demonstration patient IDs do not match."
    )

print("\nPATIENT ID VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 22 — FLEXIBLE INFERENCE VALIDATION
# ============================================================

expected_combinations = {
    "Genetic",
    "Behavioral",
    "Environmental",
    "Genetic + Behavioral",
    "Genetic + Environmental",
    "Behavioral + Environmental",
    "Genetic + Behavioral + Environmental"
}

actual_combinations = set(
    flexible_fusion[
        "available_modalities"
    ]
)

print("Expected combinations:")
for combination in sorted(
    expected_combinations
):
    print(" -", combination)

print("\nActual combinations:")
for combination in sorted(
    actual_combinations
):
    print(" -", combination)

if actual_combinations != expected_combinations:

    raise AssertionError(
        "Flexible fusion combinations are incomplete "
        "or incorrect."
    )

print("\nFLEXIBLE COMBINATION VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 23 — FLEXIBLE ROW COUNT VALIDATION
# ============================================================

number_of_patients = (
    flexible_fusion[
        "patient_id"
    ].nunique()
)

number_of_combinations = len(
    expected_combinations
)

expected_flexible_rows = (
    number_of_patients
    *
    number_of_combinations
)

actual_flexible_rows = len(
    flexible_fusion
)

print(
    "Patients:",
    number_of_patients
)

print(
    "Combinations per patient:",
    number_of_combinations
)

print(
    "Expected rows:",
    expected_flexible_rows
)

print(
    "Actual rows:",
    actual_flexible_rows
)

if expected_flexible_rows != actual_flexible_rows:

    raise AssertionError(
        "Flexible fusion row count is inconsistent."
    )

print("\nFLEXIBLE ROW COUNT VALIDATION: PASSED")
```

```python
# ============================================================
# CELL 24 — END-TO-END DIMENSION CHECK
# ============================================================

dimension_chain = {
    "Genetic": {
        "representation": representation_dimensions["Genetic"],
        "projection": projection_dimensions["Genetic"]
    },

    "Behavioral": {
        "representation": representation_dimensions["Behavioral"],
        "projection": projection_dimensions["Behavioral"]
    },

    "Environmental": {
        "representation": representation_dimensions["Environmental"],
        "projection": projection_dimensions["Environmental"]
    }
}

print("Phase 05 dimension chain:\n")

for modality, dimensions in dimension_chain.items():

    print(
        f"{modality:<15} "
        f"{dimensions['representation']}-D"
        " -> "
        f"{dimensions['projection']}-D"
    )

    if dimensions["representation"] != 128:
        raise AssertionError(
            f"{modality} representation is not 128-D."
        )

    if dimensions["projection"] != 32:
        raise AssertionError(
            f"{modality} projection is not 32-D."
        )

print("\nEND-TO-END DIMENSION VALIDATION: PASSED")
```
