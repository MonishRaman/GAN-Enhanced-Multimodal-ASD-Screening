```python
# ============================================================
# CELL 13 — CHECKPOINT MODALITY CLASSIFICATION
# ============================================================

print("=" * 70)
print("CHECKPOINT MODALITY CLASSIFICATION")
print("=" * 70)

modality_keywords = {
    "GENETIC": [
        "genetic",
        "gene",
        "genotype"
    ],

    "BEHAVIORAL": [
        "behavioral",
        "behavioural",
        "behavior",
        "behaviour"
    ],

    "ENVIRONMENTAL": [
        "environmental",
        "environment",
        "env"
    ],

    "CROSS_MODAL": [
        "cross_modal",
        "crossmodal",
        "baseline",
        "gan",
        "generator",
        "discriminator"
    ]
}

for index, relative_path in enumerate(
    model_checkpoint_files,
    start=1
):

    path_lower = relative_path.lower()

    detected = []

    for category, keywords in (
        modality_keywords.items()
    ):

        if any(
            keyword in path_lower
            for keyword in keywords
        ):

            detected.append(category)

    print(
        f"{index:02d}. {relative_path}"
    )

    if detected:

        print(
            "    Possible category:",
            ", ".join(detected)
        )

    else:

        print(
            "    Possible category: UNKNOWN"
        )

print(
    "\nFilename-based categorization complete."
)
```

```python
# ============================================================
# CELL 14 — INSPECT PYTORCH CHECKPOINT CONTENTS
# ============================================================

print("=" * 70)
print("PYTORCH CHECKPOINT CONTENT INSPECTION")
print("=" * 70)

pytorch_extensions = {
    ".pt",
    ".pth"
}

checkpoint_inspection = []

for relative_path in model_checkpoint_files:

    full_path = (
        PROJECT_ROOT /
        relative_path
    )

    if (
        not full_path.exists()
        or full_path.suffix.lower()
        not in pytorch_extensions
    ):
        continue

    print("\n" + "-" * 70)
    print(
        "FILE:",
        relative_path
    )
    print("-" * 70)

    try:

        import torch

        checkpoint = torch.load(
            full_path,
            map_location="cpu",
            weights_only=False
        )

        print(
            "Object type:",
            type(checkpoint)
        )

        if isinstance(
            checkpoint,
            dict
        ):

            print(
                "\nTop-level keys:"
            )

            for key in checkpoint.keys():

                value = checkpoint[key]

                if isinstance(
                    value,
                    dict
                ):

                    print(
                        f"    {key} "
                        f"→ dict "
                        f"({len(value)} entries)"
                    )

                elif torch.is_tensor(value):

                    print(
                        f"    {key} "
                        f"→ Tensor "
                        f"{tuple(value.shape)}"
                    )

                else:

                    print(
                        f"    {key} "
                        f"→ {type(value).__name__}"
                    )

            state_dict = None

            if "state_dict" in checkpoint:

                state_dict = checkpoint[
                    "state_dict"
                ]

            elif "model_state_dict" in checkpoint:

                state_dict = checkpoint[
                    "model_state_dict"
                ]

            elif "generator_state_dict" in checkpoint:

                state_dict = checkpoint[
                    "generator_state_dict"
                ]

            elif "discriminator_state_dict" in checkpoint:

                state_dict = checkpoint[
                    "discriminator_state_dict"
                ]

            if state_dict is not None:

                print(
                    "\nState-dict parameter shapes:"
                )

                for name, tensor in list(
                    state_dict.items()
                )[:30]:

                    if torch.is_tensor(tensor):

                        print(
                            f"    {name:45s}"
                            f" {tuple(tensor.shape)}"
                        )

                if len(state_dict) > 30:

                    print(
                        f"    ... "
                        f"{len(state_dict) - 30} "
                        "more parameters"
                    )

                checkpoint_inspection.append({
                    "path": relative_path,
                    "type": "state_dict",
                    "state_dict_keys": len(
                        state_dict
                    )
                })

            else:

                checkpoint_inspection.append({
                    "path": relative_path,
                    "type": str(
                        type(checkpoint)
                    ),
                    "state_dict_keys": None
                })

        else:

            checkpoint_inspection.append({
                "path": relative_path,
                "type": str(
                    type(checkpoint)
                ),
                "state_dict_keys": None
            })

    except Exception as error:

        print(
            "\nCould not inspect file:"
        )

        print(
            f"    {type(error).__name__}: "
            f"{error}"
        )

        checkpoint_inspection.append({
            "path": relative_path,
            "type": "inspection_failed",
            "error": str(error)
        })

print(
    "\n" + "=" * 70
)

print(
    "PyTorch checkpoint inspection complete."
)

print(
    f"Files inspected: "
    f"{len(checkpoint_inspection)}"
)
```

```python
# ============================================================
# CELL 15 — SEARCH SOURCE CODE FOR MODALITY ARCHITECTURES
# ============================================================

print("=" * 70)
print("SOURCE CODE ARCHITECTURE SEARCH")
print("=" * 70)

architecture_terms = [
    "GeneticModel",
    "Genetic",
    "BehavioralModel",
    "Behavioral",
    "EnvironmentalModel",
    "Environmental",
    "projection",
    "Linear(128",
    "Linear(32",
    "128, 32",
    "128 -> 32",
    "128-D",
    "32-D",
    "embedding"
]

architecture_matches = []

code_files = [
    path
    for path in CROSS_MODAL_ROOT.rglob("*")
    if (
        path.is_file()
        and path.suffix.lower()
        in {".py", ".ipynb"}
    )
]

print(
    f"\nSearching {len(code_files)} "
    "Python/notebook files...\n"
)

for path in code_files:

    try:

        text = path.read_text(
            encoding="utf-8",
            errors="ignore"
        )

    except Exception:

        continue

    matched_terms = []

    for term in architecture_terms:

        if term.lower() in text.lower():

            matched_terms.append(term)

    if matched_terms:

        relative_path = path.relative_to(
            PROJECT_ROOT
        )

        architecture_matches.append({
            "path": str(relative_path),
            "matched_terms": matched_terms
        })

        print(
            relative_path
        )

        print(
            "    matched:",
            ", ".join(matched_terms)
        )

print(
    "\nArchitecture source search complete."
)

print(
    "Matching files:",
    len(architecture_matches)
)
```

```python
# ============================================================
# CELL 25 — FIND 128-D → 32-D CHECKPOINTS
# ============================================================

print("=" * 70)
print("128-D → 32-D CHECKPOINT DISCOVERY")
print("=" * 70)

alignment_checkpoints = {}

for modality, alignment_root in alignment_roots.items():

    candidates = []

    for path in alignment_root.rglob("*"):

        if not path.is_file():
            continue

        name_lower = path.name.lower()

        if (
            path.suffix.lower()
            in {".pt", ".pth", ".ckpt"}
            and (
                "128" in name_lower
                or "32" in name_lower
                or "align" in name_lower
                or "project" in name_lower
            )
        ):

            candidates.append(path)

    alignment_checkpoints[
        modality
    ] = sorted(
        candidates,
        key=lambda p: str(p).lower()
    )

    print(
        f"\n{modality}: "
        f"{len(candidates)} candidate checkpoint(s)"
    )

    for path in alignment_checkpoints[
        modality
    ]:

        print(
            "   ",
            path.relative_to(PROJECT_ROOT)
        )

print(
    "\nAlignment checkpoint discovery complete."
)
```

```python
# ============================================================
# CELL 26 — FIND REPRESENTATION-LEARNING CHECKPOINTS
# ============================================================

print("=" * 70)
print("REPRESENTATION-LEARNING CHECKPOINT DISCOVERY")
print("=" * 70)

representation_checkpoints = {}

for modality, representation_root in representation_roots.items():

    candidates = []

    for path in representation_root.rglob("*"):

        if not path.is_file():
            continue

        if path.suffix.lower() in {
            ".pt",
            ".pth",
            ".ckpt",
            ".h5",
            ".keras",
            ".joblib",
            ".pkl"
        }:

            candidates.append(path)

    representation_checkpoints[
        modality
    ] = sorted(
        candidates,
        key=lambda p: str(p).lower()
    )

    print(
        f"\n{modality}: "
        f"{len(candidates)} checkpoint/model file(s)"
    )

    for path in representation_checkpoints[
        modality
    ]:

        size_mb = (
            path.stat().st_size
            /
            (1024 ** 2)
        )

        print(
            f"    {path.relative_to(PROJECT_ROOT)}"
            f"    ({size_mb:.3f} MB)"
        )

print(
    "\nRepresentation-learning checkpoint discovery complete."
)
```

```python
# ============================================================
# CELL 27 — INSPECT 128→32 CHECKPOINT CONTENTS
# ============================================================

import torch

print("=" * 70)
print("128-D → 32-D CHECKPOINT CONTENT INSPECTION")
print("=" * 70)

alignment_checkpoint_details = {}

for modality, candidates in alignment_checkpoints.items():

    print("\n" + "=" * 70)
    print(modality)
    print("=" * 70)

    alignment_checkpoint_details[
        modality
    ] = []

    for path in candidates:

        print(
            "\nFILE:",
            path.relative_to(PROJECT_ROOT)
        )

        try:

            checkpoint = torch.load(
                path,
                map_location="cpu",
                weights_only=False
            )

            print(
                "Object type:",
                type(checkpoint)
            )

            details = {
                "path": str(
                    path.relative_to(PROJECT_ROOT)
                ),
                "object_type":
                    str(type(checkpoint))
            }

            if isinstance(
                checkpoint,
                dict
            ):

                print(
                    "Top-level keys:"
                )

                for key, value in checkpoint.items():

                    if torch.is_tensor(value):

                        print(
                            f"    {key}"
                            f" → Tensor "
                            f"{tuple(value.shape)}"
                        )

                    elif isinstance(
                        value,
                        dict
                    ):

                        print(
                            f"    {key}"
                            f" → dict "
                            f"({len(value)} entries)"
                        )

                    else:

                        print(
                            f"    {key}"
                            f" → "
                            f"{type(value).__name__}"
                        )

                details[
                    "keys"
                ] = list(
                    checkpoint.keys()
                )

                possible_state_dict = None

                for state_key in [
                    "state_dict",
                    "model_state_dict",
                    "alignment_state_dict",
                    "projection_state_dict"
                ]:

                    if state_key in checkpoint:

                        possible_state_dict = (
                            checkpoint[state_key]
                        )

                        print(
                            "\nState dictionary found:",
                            state_key
                        )

                        break

                if possible_state_dict is not None:

                    print(
                        "Parameter shapes:"
                    )

                    for name, tensor in list(
                        possible_state_dict.items()
                    )[:40]:

                        if torch.is_tensor(
                            tensor
                        ):

                            print(
                                f"    {name:45s}"
                                f" {tuple(tensor.shape)}"
                            )

            alignment_checkpoint_details[
                modality
            ].append(details)

        except Exception as error:

            print(
                "Inspection failed:"
            )

            print(
                f"    {type(error).__name__}: "
                f"{error}"
            )

print(
    "\nAlignment checkpoint inspection complete."
)
```

```python
# ============================================================
# CELL 28 — INSPECT REPRESENTATION MODEL CHECKPOINTS
# ============================================================

print("=" * 70)
print("REPRESENTATION MODEL CHECKPOINT CONTENT INSPECTION")
print("=" * 70)

representation_checkpoint_details = {}

for modality, candidates in representation_checkpoints.items():

    print("\n" + "=" * 70)
    print(modality)
    print("=" * 70)

    representation_checkpoint_details[
        modality
    ] = []

    for path in candidates:

        print(
            "\nFILE:",
            path.relative_to(PROJECT_ROOT)
        )

        try:

            checkpoint = torch.load(
                path,
                map_location="cpu",
                weights_only=False
            )

            print(
                "Object type:",
                type(checkpoint)
            )

            details = {
                "path": str(
                    path.relative_to(PROJECT_ROOT)
                ),
                "object_type":
                    str(type(checkpoint))
            }

            if isinstance(
                checkpoint,
                dict
            ):

                print(
                    "Top-level keys:"
                )

                for key, value in checkpoint.items():

                    if torch.is_tensor(value):

                        print(
                            f"    {key}"
                            f" → Tensor "
                            f"{tuple(value.shape)}"
                        )

                    elif isinstance(
                        value,
                        dict
                    ):

                        print(
                            f"    {key}"
                            f" → dict "
                            f"({len(value)} entries)"
                        )

                    else:

                        print(
                            f"    {key}"
                            f" → "
                            f"{type(value).__name__}"
                        )

                details[
                    "keys"
                ] = list(
                    checkpoint.keys()
                )

                possible_state_dict = None

                for state_key in [
                    "state_dict",
                    "model_state_dict",
                    "encoder_state_dict"
                ]:

                    if state_key in checkpoint:

                        possible_state_dict = (
                            checkpoint[state_key]
                        )

                        print(
                            "\nState dictionary found:",
                            state_key
                        )

                        break

                if possible_state_dict is not None:

                    print(
                        "Parameter shapes:"
                    )

                    for name, tensor in list(
                        possible_state_dict.items()
                    )[:50]:

                        if torch.is_tensor(
                            tensor
                        ):

                            print(
                                f"    {name:45s}"
                                f" {tuple(tensor.shape)}"
                            )

            representation_checkpoint_details[
                modality
            ].append(details)

        except Exception as error:

            print(
                "Inspection failed:"
            )

            print(
                f"    {type(error).__name__}: "
                f"{error}"
            )

print(
    "\nRepresentation checkpoint inspection complete."
)
```

```python
# ============================================================
# CELL 34 — INSPECT 128-D → 32-D WEIGHT SHAPES
# ============================================================

print("=" * 70)
print("128-D → 32-D ALIGNMENT WEIGHT SHAPES")
print("=" * 70)

for modality, paths in alignment_checkpoints.items():

    print("\n" + "-" * 70)
    print(modality)
    print("-" * 70)

    for path in paths:

        print(
            "\nCheckpoint:"
        )

        print(
            path.relative_to(PROJECT_ROOT)
        )

        checkpoint = torch.load(
            path,
            map_location="cpu",
            weights_only=False
        )

        print(
            "\nObject type:",
            type(checkpoint)
        )

        if isinstance(checkpoint, dict):

            print(
                "\nTop-level keys:"
            )

            for key in checkpoint.keys():

                print(
                    "   ",
                    key
                )

            state_dict = None
            state_dict_name = None

            for possible_key in [
                "state_dict",
                "model_state_dict",
                "alignment_state_dict",
                "projection_state_dict"
            ]:

                if possible_key in checkpoint:

                    value = checkpoint[
                        possible_key
                    ]

                    if isinstance(
                        value,
                        dict
                    ):

                        state_dict = value
                        state_dict_name = possible_key
                        break

            if state_dict is None:

                # Sometimes the checkpoint itself
                # is the state dictionary.
                tensor_values = [
                    value
                    for value in checkpoint.values()
                    if torch.is_tensor(value)
                ]

                if len(tensor_values) > 0:

                    state_dict = checkpoint
                    state_dict_name = "checkpoint itself"

            if state_dict is not None:

                print(
                    "\nUsing state dictionary:",
                    state_dict_name
                )

                print(
                    "\nParameter shapes:"
                )

                for name, value in state_dict.items():

                    if torch.is_tensor(value):

                        print(
                            f"    {name:<45}"
                            f"{tuple(value.shape)}"
                        )

        else:

            print(
                "\nCheckpoint is not a dictionary."
            )

print(
    "\nAlignment weight-shape inspection complete."
)
```
