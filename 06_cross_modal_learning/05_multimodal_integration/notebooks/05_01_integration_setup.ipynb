```python
# ============================================================
# PHASE 05 — MULTIMODAL INTEGRATION
# 05.01 — INTEGRATION SETUP
# ============================================================

print("=" * 75)
print("PHASE 05 — MULTIMODAL INTEGRATION")
print("05.01 — INTEGRATION SETUP")
print("=" * 75)

print("""
Objective:
Identify and verify the previously trained modality-specific
representation models, 128→32 projection models, preprocessing
artifacts, and existing aligned representations required for
multimodal integration.

No model is trained, modified, or overwritten in this notebook.
""")
```

```python
# ============================================================
# CELL 2 — PROJECT PATH CONFIGURATION
# ============================================================

from pathlib import Path
import json
import hashlib
import re
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Users\shasw\Documents\GitHub\GAN-Enhanced-Multimodal-ASD-Screening"
)

PHASE05_ROOT = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "05_multimodal_integration"
)

NOTEBOOKS_DIR = PHASE05_ROOT / "notebooks"
MODELS_DIR = PHASE05_ROOT / "models"
REPORTS_DIR = PHASE05_ROOT / "reports"
FIGURES_DIR = PHASE05_ROOT / "figures"

SETUP_REPORT_DIR = REPORTS_DIR / "05_01_integration_setup"

print("=" * 75)
print("PROJECT PATH CONFIGURATION")
print("=" * 75)

print("\nProject root:")
print(PROJECT_ROOT)

print("\nPhase 05 root:")
print(PHASE05_ROOT)

print("\nSetup report directory:")
print(SETUP_REPORT_DIR)
```

```python
# ============================================================
# CELL 3 — CREATE PHASE 05 DIRECTORIES
# ============================================================

required_directories = [
    NOTEBOOKS_DIR,
    MODELS_DIR,
    REPORTS_DIR,
    FIGURES_DIR,
    SETUP_REPORT_DIR
]

for directory in required_directories:
    directory.mkdir(parents=True, exist_ok=True)

print("=" * 75)
print("PHASE 05 DIRECTORY VERIFICATION")
print("=" * 75)

for directory in required_directories:
    print(
        f"{'PASSED' if directory.exists() else 'FAILED'} : "
        f"{directory}"
    )

print("\nDirectory setup complete.")
```

```python
# ============================================================
# CELL 4 — VERIFY MODALITY DIRECTORIES
# ============================================================

MODALITY_ROOTS = {
    "Genetic": PROJECT_ROOT / "Genetic_Model",
    "Behavioral": PROJECT_ROOT / "Behavioral_Model",
    "Environmental": PROJECT_ROOT / "Environmental_Model"
}

print("=" * 75)
print("MODALITY DIRECTORY VERIFICATION")
print("=" * 75)

modality_status = {}

for modality, path in MODALITY_ROOTS.items():

    exists = path.exists()
    modality_status[modality] = exists

    print(
        f"{'PASSED' if exists else 'FAILED'} : "
        f"{modality} → {path}"
    )

if not all(modality_status.values()):
    raise FileNotFoundError(
        "One or more required modality directories are missing."
    )

print("\nAll three modality directories verified.")
```

```python
# ============================================================
# CELL 5 — REPRESENTATION MODEL INVENTORY
# ============================================================

REPRESENTATION_MODELS = {
    "Genetic": (
        PROJECT_ROOT
        / "Genetic_Model"
        / "03_representation_learning"
        / "genetic_transformer_128d.pt"
    ),

    "Behavioral": (
        PROJECT_ROOT
        / "Behavioral_Model"
        / "03_representation_learning"
        / "behavioral_encoder_128d.pt"
    ),

    "Environmental": (
        PROJECT_ROOT
        / "Environmental_Model"
        / "03_representation_learning"
        / "environmental_encoder_128d.pt"
    )
}

print("=" * 75)
print("128-D REPRESENTATION MODEL INVENTORY")
print("=" * 75)

representation_inventory = []

for modality, path in REPRESENTATION_MODELS.items():

    exists = path.exists()

    print(f"\n{modality}")
    print("Path:", path)
    print("Exists:", exists)

    if not exists:
        raise FileNotFoundError(
            f"Missing {modality} representation model:\n{path}"
        )

    representation_inventory.append({
        "modality": modality,
        "stage": "representation_learning",
        "output_dimension": 128,
        "model_path": str(path),
        "exists": True
    })

print("\nAll three 128-D representation models verified.")
```

```python
# ============================================================
# CELL 6 — COMMON 32-D PROJECTION MODEL INVENTORY
# ============================================================

PROJECTION_MODELS = {
    "Genetic": (
        PROJECT_ROOT
        / "Genetic_Model"
        / "04_cross_modal_alignment"
        / "genetic_128_to_32_alignment.pt"
    ),

    "Behavioral": (
        PROJECT_ROOT
        / "Behavioral_Model"
        / "04_cross_modal_alignment"
        / "behavioral_128_to_32_alignment.pt"
    ),

    "Environmental": (
        PROJECT_ROOT
        / "Environmental_Model"
        / "04_cross_modal_alignment"
        / "environmental_128_to_32_alignment.pt"
    )
}

print("=" * 75)
print("128-D → 32-D PROJECTION MODEL INVENTORY")
print("=" * 75)

projection_inventory = []

for modality, path in PROJECTION_MODELS.items():

    exists = path.exists()

    print(f"\n{modality}")
    print("Path:", path)
    print("Exists:", exists)

    if not exists:
        raise FileNotFoundError(
            f"Missing {modality} 128→32 projection model:\n{path}"
        )

    projection_inventory.append({
        "modality": modality,
        "stage": "cross_modal_alignment",
        "input_dimension": 128,
        "output_dimension": 32,
        "model_path": str(path),
        "exists": True
    })

print("\nAll three 128→32 projection models verified.")
```

```python
# ============================================================
# CELL 7 — EXISTING 32-D REPRESENTATION VERIFICATION
# ============================================================

ALIGNED_REPRESENTATIONS = {
    "Genetic": (
        PROJECT_ROOT
        / "Genetic_Model"
        / "04_cross_modal_alignment"
        / "genetic_32d_aligned_representation.csv"
    ),

    "Behavioral": (
        PROJECT_ROOT
        / "Behavioral_Model"
        / "04_cross_modal_alignment"
        / "behavioral_32d_aligned_representation.csv"
    ),

    "Environmental": (
        PROJECT_ROOT
        / "Environmental_Model"
        / "04_cross_modal_alignment"
        / "environmental_32d_aligned_representation.csv"
    )
}

print("=" * 75)
print("EXISTING 32-D REPRESENTATION VERIFICATION")
print("=" * 75)

representation_output_inventory = []

for modality, path in ALIGNED_REPRESENTATIONS.items():

    if not path.exists():
        raise FileNotFoundError(
            f"Missing aligned representation:\n{path}"
        )

    df = pd.read_csv(path)

    # Exclude identifier / target columns.
    metadata_columns = {
        "GSM_ID",
        "Sample_Index",
        "Target",
        "Class/ASD"
    }

    representation_columns = [
        column
        for column in df.columns
        if column not in metadata_columns
    ]

    print(f"\n{modality}")
    print("Shape:", df.shape)
    print("32-D representation columns:", len(representation_columns))

    if len(representation_columns) != 32:
        raise ValueError(
            f"{modality} does not contain exactly 32 representation columns."
        )

    representation_output_inventory.append({
        "modality": modality,
        "samples": len(df),
        "total_columns": len(df.columns),
        "representation_dimensions": 32,
        "path": str(path)
    })

print("\nAll existing 32-D representations verified.")
```

```python
# ============================================================
# CELL 8 — BEHAVIORAL PREPROCESSING SCHEMA
# ============================================================

BEHAVIORAL_PREPROCESSOR = (
    PROJECT_ROOT
    / "Behavioral_Model"
    / "02_preprocessing"
    / "behavioral_preprocessor.joblib"
)

BEHAVIORAL_FEATURE_NAMES = (
    PROJECT_ROOT
    / "Behavioral_Model"
    / "02_preprocessing"
    / "behavioral_processed_feature_names.txt"
)

print("=" * 75)
print("BEHAVIORAL PREPROCESSING SCHEMA")
print("=" * 75)

if not BEHAVIORAL_PREPROCESSOR.exists():
    raise FileNotFoundError(BEHAVIORAL_PREPROCESSOR)

if not BEHAVIORAL_FEATURE_NAMES.exists():
    raise FileNotFoundError(BEHAVIORAL_FEATURE_NAMES)

# ------------------------------------------------------------
# Read serialized sklearn version without deserializing the
# object. This avoids the sklearn 1.9 compatibility issue.
# ------------------------------------------------------------

with open(BEHAVIORAL_PREPROCESSOR, "rb") as f:
    raw_bytes = f.read()

serialized_text = raw_bytes.decode(
    "latin-1",
    errors="ignore"
)

version_match = re.search(
    r"_sklearn_version.{0,10}1\.6\.1",
    serialized_text
)

serialized_sklearn_version = (
    "1.6.1" if version_match else "Not detected"
)

# ------------------------------------------------------------
# Read processed feature names
# ------------------------------------------------------------

with open(
    BEHAVIORAL_FEATURE_NAMES,
    "r",
    encoding="utf-8"
) as f:
    processed_features = [
        line.strip()
        for line in f
        if line.strip()
    ]

behavioral_raw_features = [
    "A1_Score",
    "A2_Score",
    "A3_Score",
    "A4_Score",
    "A5_Score",
    "A6_Score",
    "A7_Score",
    "A8_Score",
    "A9_Score",
    "A10_Score",
    "age",
    "gender",
    "jundice",
    "austim",
    "used_app_before",
    "ethnicity",
    "contry_of_res",
    "relation"
]

print("\nOriginal raw input fields:", len(behavioral_raw_features))
print("Processed feature fields:", len(processed_features))
print("Serialized sklearn version:", serialized_sklearn_version)

print("\nRaw Behavioral inputs:")

for i, feature in enumerate(
    behavioral_raw_features,
    start=1
):
    print(f"{i:2d}. {feature}")

if len(behavioral_raw_features) != 18:
    raise ValueError("Behavioral raw schema should contain 18 fields.")

if len(processed_features) != 87:
    raise ValueError("Behavioral processed schema should contain 87 fields.")

print("\nBehavioral preprocessing schema verified.")
```

```python
# ============================================================
# CELL 9 — ENVIRONMENTAL PREPROCESSING VERIFICATION
# ============================================================

ENVIRONMENTAL_PREPROCESSOR = (
    PROJECT_ROOT
    / "Environmental_Model"
    / "02_preprocessing"
    / "environmental_preprocessor.joblib"
)

ENVIRONMENTAL_PROCESSED = (
    PROJECT_ROOT
    / "Environmental_Model"
    / "02_preprocessing"
    / "environmental_train_processed.csv"
)

print("=" * 75)
print("ENVIRONMENTAL PREPROCESSING VERIFICATION")
print("=" * 75)

print("\nPreprocessor:")
print(ENVIRONMENTAL_PREPROCESSOR)

print("Exists:", ENVIRONMENTAL_PREPROCESSOR.exists())

print("\nProcessed dataset:")
print(ENVIRONMENTAL_PROCESSED)

print("Exists:", ENVIRONMENTAL_PROCESSED.exists())

if not ENVIRONMENTAL_PREPROCESSOR.exists():
    raise FileNotFoundError(ENVIRONMENTAL_PREPROCESSOR)

if not ENVIRONMENTAL_PROCESSED.exists():
    raise FileNotFoundError(ENVIRONMENTAL_PROCESSED)

environmental_df = pd.read_csv(
    ENVIRONMENTAL_PROCESSED,
    nrows=5
)

print("\nProcessed preview shape:")
print(environmental_df.shape)

print("\nEnvironmental processed feature count:")
print(len(environmental_df.columns) - 1)

print("\nEnvironmental preprocessing artifacts verified.")
```

```python
# ============================================================
# CELL 10 — GENETIC PIPELINE ARTIFACT VERIFICATION
# ============================================================

GENETIC_RAW_DIR = (
    PROJECT_ROOT
    / "Genetic_Model"
    / "01_raw_data"
)

GENETIC_REPRESENTATION_DIR = (
    PROJECT_ROOT
    / "Genetic_Model"
    / "03_representation_learning"
)

GENETIC_ALIGNMENT_DIR = (
    PROJECT_ROOT
    / "Genetic_Model"
    / "04_cross_modal_alignment"
)

print("=" * 75)
print("GENETIC PIPELINE ARTIFACT VERIFICATION")
print("=" * 75)

directories = {
    "Raw data": GENETIC_RAW_DIR,
    "Representation learning": GENETIC_REPRESENTATION_DIR,
    "Cross-modal alignment": GENETIC_ALIGNMENT_DIR
}

for name, path in directories.items():

    print(f"\n{name}")
    print("Path:", path)
    print("Exists:", path.exists())

    if not path.exists():
        raise FileNotFoundError(path)

print("\nGenetic pipeline structure verified.")
```

```python
# ============================================================
# CELL 11 — MASTER MODALITY INTEGRATION INVENTORY
# ============================================================

master_inventory = pd.DataFrame([
    {
        "modality": "Genetic",
        "representation_input": "Modality-specific raw input",
        "representation_output": 128,
        "projection_input": 128,
        "projection_output": 32,
        "representation_model": str(REPRESENTATION_MODELS["Genetic"]),
        "projection_model": str(PROJECTION_MODELS["Genetic"]),
        "aligned_representation": str(ALIGNED_REPRESENTATIONS["Genetic"])
    },
    {
        "modality": "Behavioral",
        "representation_input": "18 raw fields → 87 processed features",
        "representation_output": 128,
        "projection_input": 128,
        "projection_output": 32,
        "representation_model": str(REPRESENTATION_MODELS["Behavioral"]),
        "projection_model": str(PROJECTION_MODELS["Behavioral"]),
        "aligned_representation": str(ALIGNED_REPRESENTATIONS["Behavioral"])
    },
    {
        "modality": "Environmental",
        "representation_input": "31 processed features",
        "representation_output": 128,
        "projection_input": 128,
        "projection_output": 32,
        "representation_model": str(REPRESENTATION_MODELS["Environmental"]),
        "projection_model": str(PROJECTION_MODELS["Environmental"]),
        "aligned_representation": str(ALIGNED_REPRESENTATIONS["Environmental"])
    }
])

print("=" * 75)
print("MASTER MODALITY INTEGRATION INVENTORY")
print("=" * 75)

display(master_inventory[
    [
        "modality",
        "representation_input",
        "representation_output",
        "projection_input",
        "projection_output"
    ]
])
```

```python
# ============================================================
# CELL 12 — PHASE 03 BASELINE CHECKPOINT VERIFICATION
# ============================================================

BASELINE_CHECKPOINT = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "03_cross_modal_learning"
    / "03_04_baseline_training"
    / "checkpoints"
    / "03_04_baseline_best_model.pt"
)

print("=" * 75)
print("PHASE 03 BASELINE CHECKPOINT")
print("=" * 75)

print("\nCheckpoint:")
print(BASELINE_CHECKPOINT)

print("\nExists:")
print(BASELINE_CHECKPOINT.exists())

if not BASELINE_CHECKPOINT.exists():
    raise FileNotFoundError(BASELINE_CHECKPOINT)

print("\nOfficial baseline information:")
print("Architecture: 128 → 128 → 64 → 1")
print("Parameters: 24,833")
print("Best epoch: 13")
print("Validation ROC-AUC: 0.6626674107")
print("Status: FROZEN / INFERENCE ONLY")

print("\nPhase 03 baseline checkpoint verified.")
```

```python
# ============================================================
# CELL 13 — PHASE 05 INTEGRATION ARCHITECTURE
# ============================================================

print("=" * 75)
print("PHASE 05 INTEGRATION ARCHITECTURE")
print("=" * 75)

print("""
USER / DEMO INPUT
       │
       ├── Genetic
       │      ↓
       │   Genetic Representation Model
       │      ↓
       │    128-D
       │      ↓
       │    32-D
       │
       ├── Behavioral
       │      ↓
       │   Original Preprocessing
       │      ↓
       │    87-D
       │      ↓
       │   Behavioral Representation Model
       │      ↓
       │    128-D
       │      ↓
       │    32-D
       │
       └── Environmental
              ↓
           Original Preprocessing
              ↓
             31-D
              ↓
           Environmental Representation Model
              ↓
             128-D
              ↓
              32-D

                ↓

        G32 / B32 / E32

                ↓

       Cross-Modal Inference
          G ↔ B
          G ↔ E
          B ↔ E

                ↓

        Frozen Phase 03
        Baseline Model

                ↓

       Relationship Outputs
""")
```

```python
# ============================================================
# CELL 14 — FINAL SETUP VERIFICATION + REPORT GENERATION
# ============================================================

print("=" * 75)
print("05.01 FINAL INTEGRATION SETUP VERIFICATION")
print("=" * 75)

checks = {
    "Phase 05 directories": all(
        directory.exists()
        for directory in required_directories
    ),

    "Three modality directories": all(
        modality_status.values()
    ),

    "Three 128-D representation models": all(
        path.exists()
        for path in REPRESENTATION_MODELS.values()
    ),

    "Three 128→32 projection models": all(
        path.exists()
        for path in PROJECTION_MODELS.values()
    ),

    "Three 32-D aligned representations": all(
        path.exists()
        for path in ALIGNED_REPRESENTATIONS.values()
    ),

    "Behavioral preprocessor": (
        BEHAVIORAL_PREPROCESSOR.exists()
    ),

    "Behavioral 87-feature schema": (
        len(processed_features) == 87
    ),

    "Environmental preprocessor": (
        ENVIRONMENTAL_PREPROCESSOR.exists()
    ),

    "Phase 03 baseline checkpoint": (
        BASELINE_CHECKPOINT.exists()
    )
}

all_passed = True

for name, status in checks.items():

    print(
        f"{'PASSED' if status else 'FAILED'} : {name}"
    )

    if not status:
        all_passed = False

if not all_passed:
    raise RuntimeError(
        "05.01 integration setup verification failed."
    )

# ------------------------------------------------------------
# Save inventories
# ------------------------------------------------------------

model_inventory_df = pd.DataFrame(
    representation_inventory + projection_inventory
)

model_inventory_path = (
    SETUP_REPORT_DIR
    / "05_01_model_inventory.csv"
)

model_inventory_df.to_csv(
    model_inventory_path,
    index=False
)

representation_inventory_df = pd.DataFrame(
    representation_output_inventory
)

representation_inventory_path = (
    SETUP_REPORT_DIR
    / "05_01_representation_inventory.csv"
)

representation_inventory_df.to_csv(
    representation_inventory_path,
    index=False
)

# ------------------------------------------------------------
# Save preprocessing report
# ------------------------------------------------------------

preprocessing_report = {
    "behavioral": {
        "preprocessor": str(BEHAVIORAL_PREPROCESSOR),
        "serialized_sklearn_version": serialized_sklearn_version,
        "raw_input_dimensions": 18,
        "processed_dimensions": 87,
        "note": (
            "Original preprocessor was verified separately "
            "using its compatible sklearn 1.6.1 environment."
        )
    },

    "environmental": {
        "preprocessor": str(ENVIRONMENTAL_PREPROCESSOR),
        "processed_feature_dimension": 31
    },

    "genetic": {
        "preprocessing_structure": (
            "Genetic pipeline uses its modality-specific "
            "raw-data/representation structure."
        )
    }
}

preprocessing_report_path = (
    SETUP_REPORT_DIR
    / "05_01_preprocessing_inventory.json"
)

with open(
    preprocessing_report_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        preprocessing_report,
        f,
        indent=4
    )

# ------------------------------------------------------------
# Save master report
# ------------------------------------------------------------

master_report = {
    "phase": "05_multimodal_integration",
    "stage": "05_01_integration_setup",

    "objective": (
        "Verify and document all existing modality-specific "
        "representation, common projection, preprocessing, "
        "and Phase 03 baseline artifacts required for "
        "multimodal integration."
    ),

    "modalities": [
        "Genetic",
        "Behavioral",
        "Environmental"
    ],

    "representation_dimension": 128,
    "common_projection": "128_to_32",
    "final_common_dimension": 32,

    "behavioral_raw_input_dimension": 18,
    "behavioral_processed_dimension": 87,
    "environmental_processed_dimension": 31,

    "phase03_baseline": {
        "architecture": "128-128-64-1",
        "parameters": 24833,
        "best_epoch": 13,
        "validation_roc_auc": 0.6626674107,
        "checkpoint": str(BASELINE_CHECKPOINT)
    },

    "verification": checks,

    "status": "PASSED"
}

master_report_path = (
    SETUP_REPORT_DIR
    / "05_01_integration_setup_report.json"
)

with open(
    master_report_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        master_report,
        f,
        indent=4
    )

print("\n" + "=" * 75)
print("05.01 COMPLETE")
print("=" * 75)

print("\nStatus:")
print("ALL REQUIRED INTEGRATION SETUP CHECKS PASSED")

print("\nReports saved to:")
print(SETUP_REPORT_DIR)

print("\nGenerated files:")
print("-", model_inventory_path.name)
print("-", representation_inventory_path.name)
print("-", preprocessing_report_path.name)
print("-", master_report_path.name)
```
