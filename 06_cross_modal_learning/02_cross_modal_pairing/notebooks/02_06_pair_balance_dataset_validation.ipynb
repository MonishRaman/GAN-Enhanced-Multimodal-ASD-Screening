```python
import numpy as np
import pandas as pd

from pathlib import Path
import hashlib

print("✓ Libraries imported successfully")
```

```python
# =============================================================================
# PROJECT PATH RESOLUTION
# =============================================================================

PROJECT_ROOT = Path.cwd()

# Move upward until project root is found
while PROJECT_ROOT.name != "GAN-Enhanced-Multimodal-ASD-Screening":
    PROJECT_ROOT = PROJECT_ROOT.parent

PAIR_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "02_cross_modal_pairing"
    / "pairs"
)

REPORT_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "02_cross_modal_pairing"
    / "reports"
)

REPORT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 100)
print("PROJECT PATH CONFIGURATION")
print("=" * 100)

print(f"\nProject Root:\n{PROJECT_ROOT}")
print(f"\nPair Directory:\n{PAIR_DIR}")
print(f"\nReport Directory:\n{REPORT_DIR}")

print("\n✓ Paths configured successfully")
```

```python
# =============================================================================
# PAIR ARTIFACT PATHS
# =============================================================================

POSITIVE_PAIR_PATH = (
    PAIR_DIR
    / "02_04_positive_cross_modal_embedding_pairs.npz"
)

POSITIVE_METADATA_PATH = (
    PAIR_DIR
    / "02_04_positive_cross_modal_pair_metadata.csv"
)

NEGATIVE_PAIR_PATH = (
    PAIR_DIR
    / "02_05_negative_cross_modal_embedding_pairs.npz"
)

NEGATIVE_METADATA_PATH = (
    PAIR_DIR
    / "02_05_negative_cross_modal_pair_metadata.csv"
)

print("=" * 100)
print("PAIR ARTIFACT AVAILABILITY")
print("=" * 100)

artifacts = {
    "Positive Embedding Pairs": POSITIVE_PAIR_PATH,
    "Positive Metadata": POSITIVE_METADATA_PATH,
    "Negative Embedding Pairs": NEGATIVE_PAIR_PATH,
    "Negative Metadata": NEGATIVE_METADATA_PATH
}

for name, path in artifacts.items():
    status = "FOUND" if path.exists() else "MISSING"
    print(f"{name:<30}: {status}")
    print(f"  {path}")

print("\n✓ Artifact path verification completed")
```

```python
# =============================================================================
# LOAD POSITIVE AND NEGATIVE PAIRS
# =============================================================================

print("=" * 100)
print("LOADING CROSS-MODAL PAIR DATASETS")
print("=" * 100)

# -------------------------------------------------------------
# Positive pairs
# -------------------------------------------------------------

positive_data = np.load(POSITIVE_PAIR_PATH)

positive_source = positive_data["source_embeddings"]
positive_target = positive_data["target_embeddings"]

positive_metadata = pd.read_csv(POSITIVE_METADATA_PATH)


# -------------------------------------------------------------
# Negative pairs
# -------------------------------------------------------------

negative_data = np.load(NEGATIVE_PAIR_PATH)

negative_source = negative_data["source_embeddings"]
negative_target = negative_data["target_embeddings"]

negative_metadata = pd.read_csv(NEGATIVE_METADATA_PATH)


print("\nPOSITIVE PAIRS")
print("-" * 60)
print(f"Source embeddings : {positive_source.shape}")
print(f"Target embeddings : {positive_target.shape}")
print(f"Metadata rows     : {len(positive_metadata)}")

print("\nNEGATIVE PAIRS")
print("-" * 60)
print(f"Source embeddings : {negative_source.shape}")
print(f"Target embeddings : {negative_target.shape}")
print(f"Metadata rows     : {len(negative_metadata)}")

print("\n✓ Pair datasets loaded successfully")
```

```python
# =============================================================================
# METADATA STRUCTURE INSPECTION
# =============================================================================

print("=" * 100)
print("POSITIVE PAIR METADATA")
print("=" * 100)

print("\nColumns:")
print(list(positive_metadata.columns))

print("\nFirst 5 rows:")
display(positive_metadata.head())


print("\n" + "=" * 100)
print("NEGATIVE PAIR METADATA")
print("=" * 100)

print("\nColumns:")
print(list(negative_metadata.columns))

print("\nFirst 5 rows:")
display(negative_metadata.head())
```

```python
# =============================================================================
# STRUCTURAL INTEGRITY VALIDATION
# =============================================================================

print("=" * 100)
print("STRUCTURAL INTEGRITY VALIDATION")
print("=" * 100)

validation_results = []

datasets = {
    "Positive": {
        "source": positive_source,
        "target": positive_target,
        "metadata": positive_metadata
    },
    "Negative": {
        "source": negative_source,
        "target": negative_target,
        "metadata": negative_metadata
    }
}

for pair_type, data in datasets.items():

    source = data["source"]
    target = data["target"]
    metadata = data["metadata"]

    source_count = source.shape[0]
    target_count = target.shape[0]
    metadata_count = len(metadata)

    source_dimension = source.shape[1]
    target_dimension = target.shape[1]

    count_match = (
        source_count ==
        target_count ==
        metadata_count
    )

    dimension_match = (
        source_dimension == 32
        and target_dimension == 32
    )

    overall_status = count_match and dimension_match

    validation_results.append({
        "Pair_Type": pair_type,
        "Source_Count": source_count,
        "Target_Count": target_count,
        "Metadata_Count": metadata_count,
        "Source_Dimension": source_dimension,
        "Target_Dimension": target_dimension,
        "Count_Match": count_match,
        "32D_Verified": dimension_match,
        "Status": "PASS" if overall_status else "FAIL"
    })

validation_df = pd.DataFrame(validation_results)

display(validation_df)

print("\n✓ Structural integrity validation completed")
```

```python
# =============================================================================
# POSITIVE VS NEGATIVE BALANCE VALIDATION
# =============================================================================

print("=" * 100)
print("POSITIVE VS NEGATIVE PAIR BALANCE")
print("=" * 100)

positive_count = len(positive_source)
negative_count = len(negative_source)

total_pairs = positive_count + negative_count

balance_difference = abs(
    positive_count - negative_count
)

balance_ratio = (
    min(positive_count, negative_count)
    / max(positive_count, negative_count)
)

pair_balance_summary = pd.DataFrame([
    {
        "Positive_Pairs": positive_count,
        "Negative_Pairs": negative_count,
        "Total_Pairs": total_pairs,
        "Count_Difference": balance_difference,
        "Balance_Ratio": balance_ratio,
        "Perfectly_Balanced": positive_count == negative_count,
        "Status": (
            "PASS"
            if positive_count == negative_count
            else "FAIL"
        )
    }
])

display(pair_balance_summary)

print("\nInterpretation:")

if positive_count == negative_count:
    print(
        "✓ Positive and negative learning signals are perfectly balanced."
    )
else:
    print(
        "⚠ Positive and negative pair counts are not balanced."
    )
```

```python
# =============================================================================
# COMBINE METADATA FOR VALIDATION ANALYSIS
# =============================================================================

positive_metadata_validation = positive_metadata.copy()
negative_metadata_validation = negative_metadata.copy()

# Add pair type if not already available
positive_metadata_validation["validation_pair_type"] = "Positive"
negative_metadata_validation["validation_pair_type"] = "Negative"

combined_metadata = pd.concat(
    [
        positive_metadata_validation,
        negative_metadata_validation
    ],
    ignore_index=True
)

print("=" * 100)
print("COMBINED DATASET FOR VALIDATION")
print("=" * 100)

print(f"Positive metadata rows : {len(positive_metadata)}")
print(f"Negative metadata rows : {len(negative_metadata)}")
print(f"Combined metadata rows : {len(combined_metadata)}")

print("\nPair Type Distribution:")
display(
    combined_metadata[
        "validation_pair_type"
    ].value_counts()
)
```

```python
# =============================================================================
# NUMERICAL INTEGRITY VALIDATION
# =============================================================================

print("=" * 100)
print("NUMERICAL INTEGRITY VALIDATION")
print("=" * 100)

numerical_results = []

embedding_sets = {
    "Positive_Source": positive_source,
    "Positive_Target": positive_target,
    "Negative_Source": negative_source,
    "Negative_Target": negative_target
}

for name, embeddings in embedding_sets.items():

    nan_count = np.isnan(embeddings).sum()
    inf_count = np.isinf(embeddings).sum()

    status = (
        "PASS"
        if nan_count == 0 and inf_count == 0
        else "FAIL"
    )

    numerical_results.append({
        "Embedding_Set": name,
        "Shape": str(embeddings.shape),
        "NaN_Count": int(nan_count),
        "Inf_Count": int(inf_count),
        "Status": status
    })

numerical_validation_df = pd.DataFrame(
    numerical_results
)

display(numerical_validation_df)

print("\n✓ Numerical integrity validation completed")
```

```python
# =============================================================================
# DUPLICATE PAIR DETECTION
# =============================================================================

print("=" * 100)
print("DUPLICATE PAIR DETECTION")
print("=" * 100)


def create_pair_hash(source_embedding, target_embedding):
    """
    Create a reproducible hash representing one complete pair.
    """
    
    combined = np.concatenate(
        [source_embedding, target_embedding]
    )
    
    return hashlib.sha256(
        combined.tobytes()
    ).hexdigest()


duplicate_results = []

for pair_type, data in datasets.items():

    source = data["source"]
    target = data["target"]

    pair_hashes = [
        create_pair_hash(s, t)
        for s, t in zip(source, target)
    ]

    total_pairs = len(pair_hashes)
    unique_pairs = len(set(pair_hashes))

    duplicate_pairs = (
        total_pairs - unique_pairs
    )

    duplicate_results.append({
        "Pair_Type": pair_type,
        "Total_Pairs": total_pairs,
        "Unique_Pairs": unique_pairs,
        "Duplicate_Pairs": duplicate_pairs,
        "Duplicate_Percentage": round(
            (duplicate_pairs / total_pairs) * 100,
            4
        ),
        "Status": (
            "PASS"
            if duplicate_pairs == 0
            else "REVIEW"
        )
    })

duplicate_validation_df = pd.DataFrame(
    duplicate_results
)

display(duplicate_validation_df)

print("\n✓ Duplicate pair analysis completed")
```

```python
# =============================================================================
# IDENTIFY MODALITY-RELATED METADATA COLUMNS
# =============================================================================

print("=" * 100)
print("METADATA COLUMN IDENTIFICATION")
print("=" * 100)

print("\nAvailable columns:")
for column in combined_metadata.columns:
    print(f" - {column}")

modality_columns = [
    column
    for column in combined_metadata.columns
    if "modality" in column.lower()
]

class_columns = [
    column
    for column in combined_metadata.columns
    if "class" in column.lower()
    or "label" in column.lower()
]

print("\nDetected modality-related columns:")
print(modality_columns)

print("\nDetected class/label-related columns:")
print(class_columns)
```
