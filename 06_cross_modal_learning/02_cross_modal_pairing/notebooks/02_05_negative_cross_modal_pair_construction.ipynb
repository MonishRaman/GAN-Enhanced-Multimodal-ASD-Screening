```python
# =====================================================================================
# STEP 02.5 — NEGATIVE CROSS-MODAL PAIR CONSTRUCTION
# =====================================================================================
#
# Objective:
# Construct balanced different-class cross-modal embedding relationships
# for knowledge-level cross-modal learning.
#
# Pair Definition:
#   ASD ↔ NON-ASD
#   NON-ASD ↔ ASD
#
# Important Methodological Principle:
# These pairs DO NOT represent different measurements from the same patient.
#
# They represent controlled semantic relationships between independently
# learned representations from different modalities.
#
# Patient-level matching: NOT USED
# Knowledge-level supervision: USED
#
# =====================================================================================

print("=" * 100)
print("STEP 02.5 — NEGATIVE CROSS-MODAL PAIR CONSTRUCTION")
print("=" * 100)

print("""
Analysis Type:
Knowledge-Level Cross-Modal Learning

Pair Type:
Negative

Semantic Relationship:
Different-Class Cross-Modal Relationship

Patient-Level Matching:
NOT USED

Primary Objective:
Construct controlled different-class relationships so that future
cross-modal learning can distinguish representations that should
not naturally align semantically.
""")

print("=" * 100)
```

```python
# =====================================================================================
# CELL 2 — IMPORT LIBRARIES AND CONFIGURATION
# =====================================================================================

from pathlib import Path
import numpy as np
import pandas as pd

# Reproducibility
RANDOM_SEED = 42
rng = np.random.default_rng(RANDOM_SEED)

print("=" * 100)
print("ENVIRONMENT INITIALIZATION")
print("=" * 100)

print(f"Random Seed: {RANDOM_SEED}")
print("NumPy imported successfully")
print("Pandas imported successfully")

print("\n✓ ENVIRONMENT READY")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 3 — LOAD FROZEN 32D MODAL REPRESENTATIONS
# =====================================================================================

print("=" * 100)
print("LOADING FROZEN 32D EMBEDDINGS")
print("=" * 100)

# -------------------------------------------------------------
# Resolve Project Root
# -------------------------------------------------------------

PROJECT_ROOT = Path.cwd()

while not (PROJECT_ROOT / "05_multimodal_fusion").exists():

    if PROJECT_ROOT.parent == PROJECT_ROOT:
        raise FileNotFoundError(
            "Could not locate project root containing '05_multimodal_fusion'."
        )

    PROJECT_ROOT = PROJECT_ROOT.parent


print("\nProject Root:")
print(PROJECT_ROOT)


# -------------------------------------------------------------
# Frozen Embedding Directory
# -------------------------------------------------------------

FUSION_DIR = PROJECT_ROOT / "05_multimodal_fusion"


# -------------------------------------------------------------
# Artifact Paths
# -------------------------------------------------------------

GENETIC_EMBEDDING_PATH = FUSION_DIR / "genetic_32d_normalized.npy"
GENETIC_TARGET_PATH = FUSION_DIR / "genetic_targets.npy"

BEHAVIORAL_EMBEDDING_PATH = FUSION_DIR / "behavioral_32d_normalized.npy"
BEHAVIORAL_TARGET_PATH = FUSION_DIR / "behavioral_targets.npy"

ENVIRONMENTAL_EMBEDDING_PATH = FUSION_DIR / "environmental_32d_normalized.npy"
ENVIRONMENTAL_TARGET_PATH = FUSION_DIR / "environmental_targets.npy"


# -------------------------------------------------------------
# Load Data
# -------------------------------------------------------------

genetic_embeddings = np.load(GENETIC_EMBEDDING_PATH)
genetic_labels = np.load(GENETIC_TARGET_PATH)

behavioral_embeddings = np.load(BEHAVIORAL_EMBEDDING_PATH)
behavioral_labels = np.load(BEHAVIORAL_TARGET_PATH)

environmental_embeddings = np.load(ENVIRONMENTAL_EMBEDDING_PATH)
environmental_labels = np.load(ENVIRONMENTAL_TARGET_PATH)


# -------------------------------------------------------------
# Store in Dictionary
# -------------------------------------------------------------

modalities = {
    "Genetic": {
        "embeddings": genetic_embeddings,
        "labels": genetic_labels
    },
    "Behavioral": {
        "embeddings": behavioral_embeddings,
        "labels": behavioral_labels
    },
    "Environmental": {
        "embeddings": environmental_embeddings,
        "labels": environmental_labels
    }
}


# -------------------------------------------------------------
# Display Summary
# -------------------------------------------------------------

for modality, data in modalities.items():

    embeddings = data["embeddings"]
    labels = data["labels"]

    print(f"\n{modality.upper()}")
    print("-" * 60)
    print(f"Embedding Shape : {embeddings.shape}")
    print(f"Label Shape     : {labels.shape}")
    print(f"Class 0 Count   : {(labels == 0).sum()}")
    print(f"Class 1 Count   : {(labels == 1).sum()}")


print("\n" + "=" * 100)
print("✓ FROZEN EMBEDDINGS LOADED SUCCESSFULLY")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 4 — DEFINE CROSS-MODAL NEGATIVE PAIRING RULES
# =====================================================================================

print("=" * 100)
print("NEGATIVE PAIRING RULES")
print("=" * 100)

# -------------------------------------------------------------
# Cross-Modal Relationships
# -------------------------------------------------------------

MODALITY_PAIRS = [
    ("Genetic", "Behavioral"),
    ("Genetic", "Environmental"),
    ("Behavioral", "Environmental")
]


# -------------------------------------------------------------
# Negative Class Relationships
# -------------------------------------------------------------

NEGATIVE_CLASS_PAIRS = [
    (0, 1),  # NON-ASD → ASD
    (1, 0)   # ASD → NON-ASD
]


print("\nMODALITY RELATIONSHIPS:")

for source_modality, target_modality in MODALITY_PAIRS:
    print(f"  {source_modality} ↔ {target_modality}")


print("\nNEGATIVE CLASS RELATIONSHIPS:")

for source_class, target_class in NEGATIVE_CLASS_PAIRS:

    source_name = "ASD" if source_class == 1 else "NON-ASD"
    target_name = "ASD" if target_class == 1 else "NON-ASD"

    print(
        f"  {source_name} → {target_name}"
    )


print("\nScientific Meaning:")
print(
    "Negative pairs connect cross-modal representations belonging "
    "to different diagnostic classes."
)

print("\n✓ NEGATIVE PAIRING RULES DEFINED")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 5 — DETERMINE BALANCED NEGATIVE PAIR BUDGET
# =====================================================================================

print("=" * 100)
print("BALANCED NEGATIVE PAIR BUDGET CALCULATION")
print("=" * 100)

# -------------------------------------------------------------
# Count Samples Per Class Per Modality
# -------------------------------------------------------------

class_counts = {}

for modality, data in modalities.items():

    labels = data["labels"]

    class_counts[modality] = {
        0: int(np.sum(labels == 0)),
        1: int(np.sum(labels == 1))
    }


# -------------------------------------------------------------
# Display Counts
# -------------------------------------------------------------

print("\nCLASS AVAILABILITY")

for modality, counts in class_counts.items():

    print(f"\n{modality}")
    print(f"  NON-ASD (0): {counts[0]}")
    print(f"  ASD (1)    : {counts[1]}")


# -------------------------------------------------------------
# Determine Global Balanced Budget
# -------------------------------------------------------------

all_class_counts = []

for modality in class_counts:

    all_class_counts.append(class_counts[modality][0])
    all_class_counts.append(class_counts[modality][1])


PAIR_BUDGET = min(all_class_counts)


print("\n" + "-" * 100)
print("BALANCED PAIR BUDGET")
print("-" * 100)

print(f"\nMinimum class availability across all modalities: {PAIR_BUDGET}")

print(
    "\nThis value is selected to ensure that every modality and "
    "every class can contribute equally."
)

print("\n✓ BALANCED PAIR BUDGET ESTABLISHED")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 6 — NEGATIVE PAIR CONSTRUCTION FUNCTION
# =====================================================================================

def construct_negative_pairs(
    source_embeddings,
    source_labels,
    target_embeddings,
    target_labels,
    source_modality,
    target_modality,
    pair_budget,
    rng
):
    """
    Construct balanced negative cross-modal pairs.

    Negative Pair Definition:
    Different-class representations are paired.

    Example:
    ASD → NON-ASD
    NON-ASD → ASD

    Returns:
    --------
    source_pairs
    target_pairs
    metadata
    """

    source_pair_list = []
    target_pair_list = []
    metadata = []

    negative_class_relationships = [
        (0, 1),
        (1, 0)
    ]

    for source_class, target_class in negative_class_relationships:

        # ---------------------------------------------------------
        # Find Available Samples
        # ---------------------------------------------------------

        source_indices = np.where(
            source_labels == source_class
        )[0]

        target_indices = np.where(
            target_labels == target_class
        )[0]

        # ---------------------------------------------------------
        # Validate Availability
        # ---------------------------------------------------------

        if len(source_indices) < pair_budget:
            raise ValueError(
                f"{source_modality} class {source_class} "
                f"does not contain enough samples."
            )

        if len(target_indices) < pair_budget:
            raise ValueError(
                f"{target_modality} class {target_class} "
                f"does not contain enough samples."
            )

        # ---------------------------------------------------------
        # Random Balanced Sampling
        # ---------------------------------------------------------

        selected_source_indices = rng.choice(
            source_indices,
            size=pair_budget,
            replace=False
        )

        selected_target_indices = rng.choice(
            target_indices,
            size=pair_budget,
            replace=False
        )

        # ---------------------------------------------------------
        # Construct Pair Records
        # ---------------------------------------------------------

        for source_idx, target_idx in zip(
            selected_source_indices,
            selected_target_indices
        ):

            source_pair_list.append(
                source_embeddings[source_idx]
            )

            target_pair_list.append(
                target_embeddings[target_idx]
            )

            metadata.append({
                "source_modality": source_modality,
                "target_modality": target_modality,
                "source_index": int(source_idx),
                "target_index": int(target_idx),
                "source_class": int(source_class),
                "target_class": int(target_class),
                "pair_type": "negative",
                "semantic_relationship": "different_class"
            })

    return (
        np.array(source_pair_list),
        np.array(target_pair_list),
        pd.DataFrame(metadata)
    )


print("=" * 100)
print("✓ NEGATIVE PAIR CONSTRUCTION FUNCTION READY")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 7 — CONSTRUCT ALL NEGATIVE CROSS-MODAL PAIRS
# =====================================================================================

print("=" * 100)
print("CONSTRUCTING NEGATIVE CROSS-MODAL PAIRS")
print("=" * 100)

all_source_pairs = []
all_target_pairs = []
all_metadata = []

for source_modality, target_modality in MODALITY_PAIRS:

    print(
        f"\nConstructing: "
        f"{source_modality} ↔ {target_modality}"
    )

    source_embeddings = modalities[source_modality]["embeddings"]
    source_labels = modalities[source_modality]["labels"]

    target_embeddings = modalities[target_modality]["embeddings"]
    target_labels = modalities[target_modality]["labels"]

    (
        source_pairs,
        target_pairs,
        metadata
    ) = construct_negative_pairs(
        source_embeddings=source_embeddings,
        source_labels=source_labels,
        target_embeddings=target_embeddings,
        target_labels=target_labels,
        source_modality=source_modality,
        target_modality=target_modality,
        pair_budget=PAIR_BUDGET,
        rng=rng
    )

    all_source_pairs.append(source_pairs)
    all_target_pairs.append(target_pairs)
    all_metadata.append(metadata)

    print(f"Pairs constructed: {len(metadata)}")


# -------------------------------------------------------------
# Combine All Relationships
# -------------------------------------------------------------

negative_source_pairs = np.vstack(all_source_pairs)

negative_target_pairs = np.vstack(all_target_pairs)

negative_pair_metadata = pd.concat(
    all_metadata,
    ignore_index=True
)


print("\n" + "=" * 100)
print("NEGATIVE PAIR CONSTRUCTION SUMMARY")
print("=" * 100)

print(f"\nTotal Negative Pairs: {len(negative_pair_metadata)}")

print(
    f"Source Embedding Shape: "
    f"{negative_source_pairs.shape}"
)

print(
    f"Target Embedding Shape: "
    f"{negative_target_pairs.shape}"
)

print("\n✓ ALL NEGATIVE PAIRS CONSTRUCTED")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 8 — VALIDATE NEGATIVE PAIR CONSTRUCTION
# =====================================================================================

print("=" * 100)
print("NEGATIVE PAIR VALIDATION")
print("=" * 100)

validation_results = []

# -------------------------------------------------------------
# Verify Every Pair is Different-Class
# -------------------------------------------------------------

negative_pair_metadata["is_different_class"] = (
    negative_pair_metadata["source_class"]
    !=
    negative_pair_metadata["target_class"]
)


# -------------------------------------------------------------
# Verify Embedding Dimensions
# -------------------------------------------------------------

source_dimension_valid = (
    negative_source_pairs.shape[1] == 32
)

target_dimension_valid = (
    negative_target_pairs.shape[1] == 32
)


# -------------------------------------------------------------
# Verify Pair Count
# -------------------------------------------------------------

expected_pairs = (
    len(MODALITY_PAIRS)
    * len(NEGATIVE_CLASS_PAIRS)
    * PAIR_BUDGET
)

actual_pairs = len(negative_pair_metadata)


validation_results.append({
    "Validation": "All pairs are different-class",
    "Expected": True,
    "Actual": bool(
        negative_pair_metadata["is_different_class"].all()
    ),
    "Status": (
        "PASS"
        if negative_pair_metadata["is_different_class"].all()
        else "FAIL"
    )
})

validation_results.append({
    "Validation": "Source embeddings are 32D",
    "Expected": True,
    "Actual": source_dimension_valid,
    "Status": "PASS" if source_dimension_valid else "FAIL"
})

validation_results.append({
    "Validation": "Target embeddings are 32D",
    "Expected": True,
    "Actual": target_dimension_valid,
    "Status": "PASS" if target_dimension_valid else "FAIL"
})

validation_results.append({
    "Validation": "Expected pair count",
    "Expected": expected_pairs,
    "Actual": actual_pairs,
    "Status": (
        "PASS"
        if expected_pairs == actual_pairs
        else "FAIL"
    )
})


validation_df = pd.DataFrame(validation_results)

print()
display(validation_df)

print("\n" + "=" * 100)

if (validation_df["Status"] == "PASS").all():
    print("✓ ALL NEGATIVE PAIR VALIDATIONS PASSED")
else:
    print("⚠ SOME VALIDATION CHECKS FAILED")

print("=" * 100)
```

```python
# =====================================================================================
# CELL 9 — NEGATIVE PAIR BALANCE ANALYSIS
# =====================================================================================

print("=" * 100)
print("NEGATIVE PAIR BALANCE ANALYSIS")
print("=" * 100)

# -------------------------------------------------------------
# Class Relationship Distribution
# -------------------------------------------------------------

class_distribution = (
    negative_pair_metadata
    .groupby(
        ["source_class", "target_class"]
    )
    .size()
    .reset_index(name="pair_count")
)


# -------------------------------------------------------------
# Modality Relationship Distribution
# -------------------------------------------------------------

modality_distribution = (
    negative_pair_metadata
    .groupby(
        ["source_modality", "target_modality"]
    )
    .size()
    .reset_index(name="pair_count")
)


print("\nCLASS RELATIONSHIP DISTRIBUTION")
print("-" * 100)

display(class_distribution)


print("\nMODALITY RELATIONSHIP DISTRIBUTION")
print("-" * 100)

display(modality_distribution)


# -------------------------------------------------------------
# Expected Balance
# -------------------------------------------------------------

print("\nEXPECTED STRUCTURE")

print(
    f"Each negative class direction: "
    f"{PAIR_BUDGET} pairs"
)

print(
    f"Each modality relationship: "
    f"{PAIR_BUDGET * 2} pairs"
)

print(
    f"Total expected pairs: "
    f"{expected_pairs}"
)

print("\n✓ BALANCE ANALYSIS COMPLETED")
print("=" * 100)
```

```python
# =====================================================================================
# CELL 10 — SAVE NEGATIVE CROSS-MODAL PAIR ARTIFACTS
# =====================================================================================

print("=" * 100)
print("SAVING NEGATIVE CROSS-MODAL PAIR ARTIFACTS")
print("=" * 100)

# -------------------------------------------------------------
# Output Directories
# -------------------------------------------------------------

PAIR_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "02_cross_modal_pairing"
    / "pairs"
)

REPORT_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "02_cross_modal_pairing"
    / "reports"
)

PAIR_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)


# -------------------------------------------------------------
# Save Embedding Pairs
# -------------------------------------------------------------

embedding_pair_path = (
    PAIR_DIR
    / "02_05_negative_cross_modal_embedding_pairs.npz"
)

np.savez_compressed(
    embedding_pair_path,
    source_embeddings=negative_source_pairs,
    target_embeddings=negative_target_pairs
)

print(f"\n✓ Embedding pairs saved:\n{embedding_pair_path}")


# -------------------------------------------------------------
# Save Pair Metadata
# -------------------------------------------------------------

metadata_path = (
    PAIR_DIR
    / "02_05_negative_cross_modal_pair_metadata.csv"
)

negative_pair_metadata.to_csv(
    metadata_path,
    index=False
)

print(f"\n✓ Pair metadata saved:\n{metadata_path}")


# -------------------------------------------------------------
# Save Validation Report
# -------------------------------------------------------------

validation_path = (
    REPORT_DIR
    / "02_05_negative_pair_validation.csv"
)

validation_df.to_csv(
    validation_path,
    index=False
)

print(f"\n✓ Validation report saved:\n{validation_path}")


# -------------------------------------------------------------
# Save Class Distribution
# -------------------------------------------------------------

class_distribution_path = (
    REPORT_DIR
    / "02_05_negative_pair_class_distribution.csv"
)

class_distribution.to_csv(
    class_distribution_path,
    index=False
)

print(f"\n✓ Class distribution saved:\n{class_distribution_path}")


# -------------------------------------------------------------
# Save Modality Distribution
# -------------------------------------------------------------

modality_distribution_path = (
    REPORT_DIR
    / "02_05_negative_pair_modality_distribution.csv"
)

modality_distribution.to_csv(
    modality_distribution_path,
    index=False
)

print(f"\n✓ Modality distribution saved:\n{modality_distribution_path}")

print("\n" + "=" * 100)
print("✓ ALL NEGATIVE PAIR ARTIFACTS SAVED")
print("=" * 100)
```
