```python
# ============================================================
# STEP 00.3 — PROJECT STRUCTURE DISCOVERY
# ============================================================

from pathlib import Path

MAX_DEPTH = 3

EXCLUDE_DIRS = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    ".venv",
    "venv",
    "node_modules"
}

print("=" * 100)
print("STEP 00.3 — PROJECT STRUCTURE DISCOVERY")
print("=" * 100)


def print_tree(path, prefix="", depth=0):
    """Print project tree up to MAX_DEPTH."""
    
    if depth > MAX_DEPTH:
        return

    try:
        items = sorted(
            [
                item for item in path.iterdir()
                if item.name not in EXCLUDE_DIRS
            ],
            key=lambda x: (not x.is_dir(), x.name.lower())
        )
    except PermissionError:
        return

    for index, item in enumerate(items):
        is_last = index == len(items) - 1

        connector = "└── " if is_last else "├── "
        icon = "📁" if item.is_dir() else "📄"

        print(prefix + connector + icon + item.name)

        if item.is_dir() and depth < MAX_DEPTH:
            extension = "    " if is_last else "│   "
            print_tree(
                item,
                prefix + extension,
                depth + 1
            )


print(f"\nPROJECT ROOT:\n{PROJECT_ROOT}\n")
print(f"STRUCTURE (MAX DEPTH = {MAX_DEPTH})\n")

print(f"📁 {PROJECT_ROOT.name}")
print_tree(PROJECT_ROOT)

print("\n" + "=" * 100)
print("PROJECT STRUCTURE DISCOVERY COMPLETED")
print("=" * 100)
```

```python
# ============================================================
# STEP 00.4 — MODALITY ARTIFACT INVENTORY
# Purpose:
# Discover and catalog existing Genetic, Behavioral,
# Environmental, Alignment, and Fusion artifacts.
# NO files are modified.
# ============================================================

from pathlib import Path
from collections import defaultdict
import pandas as pd

print("=" * 100)
print("STEP 00.4 — MODALITY ARTIFACT INVENTORY")
print("=" * 100)

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

EXCLUDE_DIRS = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    ".venv",
    "venv",
    "node_modules"
}

# File extensions worth auditing
RELEVANT_EXTENSIONS = {
    ".py", ".ipynb",
    ".pt", ".pth", ".ckpt", ".h5", ".keras",
    ".pkl", ".joblib",
    ".npy", ".npz",
    ".csv", ".json",
    ".txt", ".md"
}

# Keywords used for artifact classification
CATEGORY_KEYWORDS = {
    "Genetic": [
        "genetic", "genetics", "gene", "snp",
        "transformer", "genomics"
    ],
    "Behavioral": [
        "behavior", "behaviour", "behavioral"
    ],
    "Environmental": [
        "environment", "environmental"
    ],
    "Alignment": [
        "alignment", "align", "128_to_32",
        "128d", "32d", "embedding"
    ],
    "Fusion": [
        "fusion", "multimodal", "multi_modal",
        "weighted", "evidence"
    ]
}

# ------------------------------------------------------------
# Scan Project Files
# ------------------------------------------------------------

inventory = []

for file_path in PROJECT_ROOT.rglob("*"):

    # Skip directories
    if not file_path.is_file():
        continue

    # Skip excluded directory contents
    if any(part in EXCLUDE_DIRS for part in file_path.parts):
        continue

    # Only relevant file types
    if file_path.suffix.lower() not in RELEVANT_EXTENSIONS:
        continue

    relative_path = file_path.relative_to(PROJECT_ROOT)
    searchable_text = str(relative_path).lower()

    matched_categories = []

    for category, keywords in CATEGORY_KEYWORDS.items():
        if any(keyword in searchable_text for keyword in keywords):
            matched_categories.append(category)

    if not matched_categories:
        matched_categories = ["Other"]

    inventory.append({
        "File Name": file_path.name,
        "Relative Path": str(relative_path),
        "Extension": file_path.suffix.lower(),
        "Size (KB)": round(file_path.stat().st_size / 1024, 2),
        "Categories": ", ".join(matched_categories)
    })


# ------------------------------------------------------------
# Create DataFrame
# ------------------------------------------------------------

inventory_df = pd.DataFrame(inventory)

if inventory_df.empty:
    print("\n⚠️ No relevant artifacts found.")
else:
    inventory_df = inventory_df.sort_values(
        by=["Categories", "Relative Path"]
    ).reset_index(drop=True)

    print(f"\nTotal relevant files discovered: {len(inventory_df)}")

    print("\n" + "=" * 100)
    print("CATEGORY SUMMARY")
    print("=" * 100)

    category_counts = defaultdict(int)

    for categories in inventory_df["Categories"]:
        for category in categories.split(", "):
            category_counts[category] += 1

    for category, count in sorted(category_counts.items()):
        print(f"{category:<20}: {count} files")


# ------------------------------------------------------------
# Detailed Modality-Wise Output
# ------------------------------------------------------------

for category in [
    "Genetic",
    "Behavioral",
    "Environmental",
    "Alignment",
    "Fusion"
]:

    print("\n" + "=" * 100)
    print(f"{category.upper()} ARTIFACTS")
    print("=" * 100)

    if inventory_df.empty:
        print("No files found.")
        continue

    category_files = inventory_df[
        inventory_df["Categories"].str.contains(
            category,
            case=False,
            na=False
        )
    ]

    if category_files.empty:
        print("No matching artifacts found.")
    else:
        for _, row in category_files.iterrows():
            print(
                f"\n📄 {row['File Name']}"
                f"\n   Path       : {row['Relative Path']}"
                f"\n   Type       : {row['Extension']}"
                f"\n   Size (KB)  : {row['Size (KB)']}"
            )


# ------------------------------------------------------------
# Save Inventory Report
# ------------------------------------------------------------

REPORT_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "00_setup_and_baseline_audit"
    / "reports"
)

REPORT_DIR.mkdir(parents=True, exist_ok=True)

inventory_csv = REPORT_DIR / "00_04_modality_artifact_inventory.csv"

inventory_df.to_csv(
    inventory_csv,
    index=False
)

print("\n" + "=" * 100)
print("INVENTORY REPORT SAVED")
print("=" * 100)

print(f"\nSaved CSV:")
print(inventory_csv)

print("\n" + "=" * 100)
print("STEP 00.4 — MODALITY ARTIFACT INVENTORY COMPLETED")
print("=" * 100)
```

```python
# ============================================================
# STEP 00.5 — MODEL & CHECKPOINT AUDIT
# ============================================================
# Purpose:
# Identify all saved model/checkpoint artifacts and create
# a reproducible audit inventory.
#
# IMPORTANT:
# This cell only READS files.
# Nothing is modified or deleted.
# ============================================================

from pathlib import Path
import hashlib
from datetime import datetime
import pandas as pd

print("=" * 100)
print("STEP 00.5 — MODEL & CHECKPOINT AUDIT")
print("=" * 100)

# ------------------------------------------------------------
# Model / artifact extensions
# ------------------------------------------------------------

MODEL_EXTENSIONS = {
    ".pt",
    ".pth",
    ".ckpt",
    ".keras",
    ".h5",
    ".pkl",
    ".joblib"
}

EXCLUDE_DIRS = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    ".venv",
    "venv",
    "node_modules"
}


# ------------------------------------------------------------
# SHA256 Function
# ------------------------------------------------------------

def calculate_sha256(file_path, chunk_size=1024 * 1024):
    """
    Calculate SHA256 hash without loading the entire file
    into memory at once.
    """
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


# ------------------------------------------------------------
# Artifact Role Identification
# ------------------------------------------------------------

def identify_role(path_string):

    path_lower = path_string.lower()

    if "preprocessor" in path_lower:
        return "Preprocessing artifact"

    if "scaler" in path_lower:
        return "Feature / representation scaler"

    if "128_to_32" in path_lower or "alignment" in path_lower:
        return "128D → 32D alignment model"

    if "expert" in path_lower:
        return "Modality expert classifier"

    if "fusion" in path_lower:
        return "Multimodal fusion model"

    if "conditional_gan" in path_lower or "gan" in path_lower:
        return "Generative / augmentation model"

    if "transformer" in path_lower or "encoder" in path_lower:
        return "Representation encoder"

    if "fold_" in path_lower:
        return "Cross-validation checkpoint"

    return "Other model artifact"


# ------------------------------------------------------------
# Scan Model Artifacts
# ------------------------------------------------------------

checkpoint_records = []

for file_path in PROJECT_ROOT.rglob("*"):

    if not file_path.is_file():
        continue

    if any(part in EXCLUDE_DIRS for part in file_path.parts):
        continue

    if file_path.suffix.lower() not in MODEL_EXTENSIONS:
        continue

    relative_path = file_path.relative_to(PROJECT_ROOT)

    stat = file_path.stat()

    checkpoint_records.append({
        "File Name": file_path.name,
        "Relative Path": str(relative_path),
        "Extension": file_path.suffix.lower(),
        "Role": identify_role(str(relative_path)),
        "Size (MB)": round(stat.st_size / (1024 * 1024), 4),
        "Last Modified": datetime.fromtimestamp(
            stat.st_mtime
        ).strftime("%Y-%m-%d %H:%M:%S"),
        "SHA256": calculate_sha256(file_path)
    })


# ------------------------------------------------------------
# Create DataFrame
# ------------------------------------------------------------

checkpoint_df = pd.DataFrame(checkpoint_records)

checkpoint_df = checkpoint_df.sort_values(
    by=["Role", "Relative Path"]
).reset_index(drop=True)


# ------------------------------------------------------------
# Print Short Summary
# ------------------------------------------------------------

print(f"\nTotal model/checkpoint artifacts found: {len(checkpoint_df)}")

print("\n" + "-" * 100)
print("MODEL ARTIFACT COUNT BY ROLE")
print("-" * 100)

role_counts = checkpoint_df["Role"].value_counts()

for role, count in role_counts.items():
    print(f"{role:<40}: {count}")


print("\n" + "-" * 100)
print("KEY PRODUCTION CANDIDATE ARTIFACTS")
print("-" * 100)

production_keywords = [
    "Genetic_Model",
    "Behavioral_Model",
    "Environmental_Model",
    "05_multimodal_fusion"
]

production_candidates = checkpoint_df[
    checkpoint_df["Relative Path"].str.contains(
        "|".join(production_keywords),
        case=False,
        regex=True,
        na=False
    )
]

for _, row in production_candidates.iterrows():
    print(f"\n[{row['Role']}]")
    print(f"  File : {row['File Name']}")
    print(f"  Path : {row['Relative Path']}")
    print(f"  Size : {row['Size (MB)']} MB")


# ------------------------------------------------------------
# Save Full Audit
# ------------------------------------------------------------

REPORT_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "00_setup_and_baseline_audit"
    / "reports"
)

REPORT_DIR.mkdir(parents=True, exist_ok=True)

checkpoint_csv = REPORT_DIR / "00_05_model_checkpoint_audit.csv"

checkpoint_df.to_csv(
    checkpoint_csv,
    index=False
)

print("\n" + "=" * 100)
print("CHECKPOINT AUDIT SAVED")
print("=" * 100)

print(f"\nSaved report:\n{checkpoint_csv}")

print("\n" + "=" * 100)
print("STEP 00.5 — MODEL & CHECKPOINT AUDIT COMPLETED")
print("=" * 100)
```

```python
# ============================================================
# STEP 00 FINAL — BASELINE READINESS CHECK
# Combined shortcut to close the baseline audit
# ============================================================

import numpy as np
from pathlib import Path
from datetime import datetime
import pandas as pd

print("=" * 100)
print("STEP 00 FINAL — BASELINE READINESS CHECK")
print("=" * 100)

FUSION_DIR = PROJECT_ROOT / "05_multimodal_fusion"

artifacts = {
    "Genetic": {
        "embedding": FUSION_DIR / "genetic_32d_normalized.npy",
        "targets": FUSION_DIR / "genetic_targets.npy",
        "alignment": PROJECT_ROOT / "Genetic_Model/04_cross_modal_alignment/genetic_128_to_32_alignment.pt",
        "expert": FUSION_DIR / "M3_1_genetic_expert.pt"
    },
    
    "Behavioral": {
        "embedding": FUSION_DIR / "behavioral_32d_normalized.npy",
        "targets": FUSION_DIR / "behavioral_targets.npy",
        "alignment": PROJECT_ROOT / "Behavioral_Model/04_cross_modal_alignment/behavioral_128_to_32_alignment.pt",
        "expert": FUSION_DIR / "M3_1_behavioral_expert.pt"
    },
    
    "Environmental": {
        "embedding": FUSION_DIR / "environmental_32d_normalized.npy",
        "targets": FUSION_DIR / "environmental_targets.npy",
        "alignment": PROJECT_ROOT / "Environmental_Model/04_cross_modal_alignment/environmental_128_to_32_alignment.pt",
        "expert": FUSION_DIR / "M3_1_environmental_expert.pt"
    }
}

results = []

print("\nBASELINE ARTIFACT VERIFICATION\n")

for modality, paths in artifacts.items():
    
    print("-" * 100)
    print(f"{modality.upper()}")

    embedding_exists = paths["embedding"].exists()
    targets_exists = paths["targets"].exists()
    alignment_exists = paths["alignment"].exists()
    expert_exists = paths["expert"].exists()

    row = {
        "Modality": modality,
        "Embedding Exists": embedding_exists,
        "Targets Exist": targets_exists,
        "Alignment Model Exists": alignment_exists,
        "Expert Model Exists": expert_exists,
        "Embedding Shape": None,
        "Target Shape": None,
        "Is 32D": False,
        "Has NaN": None,
        "Has Inf": None,
        "Status": "FAILED"
    }

    if embedding_exists:
        X = np.load(paths["embedding"])

        row["Embedding Shape"] = str(X.shape)
        row["Is 32D"] = len(X.shape) == 2 and X.shape[1] == 32
        row["Has NaN"] = bool(np.isnan(X).any())
        row["Has Inf"] = bool(np.isinf(X).any())

        print(f"Embedding shape : {X.shape}")
        print(f"32D verified    : {row['Is 32D']}")
        print(f"Contains NaN    : {row['Has NaN']}")
        print(f"Contains Inf    : {row['Has Inf']}")
    else:
        print("Embedding       : NOT FOUND")

    if targets_exists:
        y = np.load(paths["targets"])
        row["Target Shape"] = str(y.shape)

        print(f"Target shape    : {y.shape}")
        print(f"Classes         : {np.unique(y)}")
    else:
        print("Targets         : NOT FOUND")

    print(f"Alignment model : {'FOUND' if alignment_exists else 'NOT FOUND'}")
    print(f"Expert model    : {'FOUND' if expert_exists else 'NOT FOUND'}")

    # Overall modality readiness
    ready = (
        embedding_exists
        and targets_exists
        and alignment_exists
        and expert_exists
        and row["Is 32D"]
        and not row["Has NaN"]
        and not row["Has Inf"]
    )

    row["Status"] = "READY" if ready else "REVIEW REQUIRED"

    print(f"\nSTATUS: {row['Status']}")

    results.append(row)


# ------------------------------------------------------------
# Final Summary
# ------------------------------------------------------------

summary_df = pd.DataFrame(results)

print("\n" + "=" * 100)
print("BASELINE READINESS SUMMARY")
print("=" * 100)

print(summary_df.to_string(index=False))

overall_ready = all(
    summary_df["Status"] == "READY"
)

print("\n" + "=" * 100)

if overall_ready:
    print("✓ BASELINE AUDIT COMPLETE")
    print("✓ ALL THREE MODALITIES ARE READY FOR EMBEDDING ANALYSIS")
    print("✓ STEP 01 CAN BEGIN")
else:
    print("⚠ REVIEW REQUIRED BEFORE STARTING STEP 01")

print("=" * 100)


# ------------------------------------------------------------
# Save Final Step 00 Report
# ------------------------------------------------------------

REPORT_DIR = (
    PROJECT_ROOT
    / "06_cross_modal_learning"
    / "00_setup_and_baseline_audit"
    / "reports"
)

REPORT_DIR.mkdir(parents=True, exist_ok=True)

report_path = REPORT_DIR / "00_final_baseline_readiness_check.csv"

summary_df.to_csv(report_path, index=False)

print(f"\nFinal readiness report saved to:\n{report_path}")
print(f"\nTimestamp: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
```
