```python
# ============================================================
# BEHAVIORAL MODEL — STEP B2
# LEAKAGE-SAFE DATA PREPARATION
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import joblib

print("=" * 70)
print("BEHAVIORAL MODEL — STEP B2")
print("LEAKAGE-SAFE DATA PREPARATION")
print("=" * 70)

# ============================================================
# 1. LOAD DATA
# ============================================================

df = pd.read_csv(INPUT_FILE)

print("\nOriginal shape:", df.shape)

# ============================================================
# 2. REMOVE DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Duplicate rows detected:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)

print("Shape after duplicate removal:", df.shape)

# ============================================================
# 3. DEFINE TARGET
# ============================================================

TARGET = "Class/ASD"

if TARGET not in df.columns:
    raise ValueError(f"Target column '{TARGET}' not found.")

y = df[TARGET].astype(int)

# ============================================================
# 4. REMOVE LEAKAGE / NON-INPUT COLUMNS
# ============================================================

LEAKAGE_COLUMNS = [
    "result",
    TARGET
]

X = df.drop(columns=LEAKAGE_COLUMNS)

print("\nRemoved columns:")
for col in LEAKAGE_COLUMNS:
    print(" -", col)

print("\nRemaining input columns:")
print(X.columns.tolist())

# ============================================================
# 5. FEATURE TYPE DETECTION
# ============================================================

numeric_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\nNumeric features:", len(numeric_columns))
print(numeric_columns)

print("\nCategorical features:", len(categorical_columns))
print(categorical_columns)

# ============================================================
# 6. STRATIFIED TRAIN / TEST SPLIT
# ============================================================

print("\n" + "=" * 70)
print("STRATIFIED TRAIN / TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Test samples    :", len(X_test))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTest distribution:")
print(y_test.value_counts())

# ============================================================
# 7. PREPROCESSING PIPELINE
# ============================================================

print("\n" + "=" * 70)
print("BUILDING PREPROCESSING PIPELINE")
print("=" * 70)

numeric_pipeline = Pipeline([
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_columns
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_columns
        )
    ]
)

# ============================================================
# 8. FIT ONLY ON TRAINING DATA
# ============================================================

print("\nFitting preprocessing on TRAINING data only...")

X_train_processed = preprocessor.fit_transform(X_train)

print("Transforming TEST data...")

X_test_processed = preprocessor.transform(X_test)

# ============================================================
# 9. FEATURE NAMES
# ============================================================

feature_names = preprocessor.get_feature_names_out()

X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names
)

# ============================================================
# 10. VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("PREPROCESSING VALIDATION")
print("=" * 70)

print("\nTraining processed shape:",
      X_train_processed.shape)

print("Test processed shape    :",
      X_test_processed.shape)

train_nan = X_train_processed.isna().sum().sum()
test_nan = X_test_processed.isna().sum().sum()

train_inf = np.isinf(
    X_train_processed.to_numpy()
).sum()

test_inf = np.isinf(
    X_test_processed.to_numpy()
).sum()

print("\nTraining NaN:", train_nan)
print("Test NaN    :", test_nan)

print("Training Inf:", train_inf)
print("Test Inf    :", test_inf)

# ============================================================
# 11. ADD TARGET
# ============================================================

train_final = X_train_processed.copy()
train_final[TARGET] = y_train.to_numpy()

test_final = X_test_processed.copy()
test_final[TARGET] = y_test.to_numpy()

# ============================================================
# 12. OUTPUT DIRECTORY
# ============================================================

OUTPUT_DIR = os.path.join(
    BEHAVIOR_ROOT,
    "02_preprocessing"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# ============================================================
# 13. SAVE TRAIN DATA
# ============================================================

TRAIN_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_train_processed.csv"
)

train_final.to_csv(
    TRAIN_FILE,
    index=False
)

# ============================================================
# 14. SAVE UNSEEN TEST DATA
# ============================================================

TEST_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_test_processed_unseen.csv"
)

test_final.to_csv(
    TEST_FILE,
    index=False
)

# ============================================================
# 15. SAVE PREPROCESSOR
# ============================================================

PREPROCESSOR_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_preprocessor.joblib"
)

joblib.dump(
    preprocessor,
    PREPROCESSOR_FILE
)

# ============================================================
# 16. SAVE FEATURE NAMES
# ============================================================

FEATURE_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_processed_feature_names.txt"
)

with open(FEATURE_FILE, "w") as f:

    for i, name in enumerate(feature_names, 1):
        f.write(f"{i}. {name}\n")

# ============================================================
# 17. SAVE CLEANED ORIGINAL
# ============================================================

CLEAN_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_cleaned_no_leakage.csv"
)

df_clean = df.drop(columns=["result"])

df_clean.to_csv(
    CLEAN_FILE,
    index=False
)

# ============================================================
# 18. SAVE REPORT
# ============================================================

REPORT_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_preprocessing_report.txt"
)

with open(REPORT_FILE, "w") as f:

    f.write("BEHAVIORAL MODEL — PREPROCESSING REPORT\n")
    f.write("=" * 70 + "\n\n")

    f.write(f"Original samples: {len(pd.read_csv(INPUT_FILE))}\n")
    f.write(f"Duplicate rows removed: {duplicate_count}\n")
    f.write(f"Final samples: {len(df)}\n\n")

    f.write("Target column:\n")
    f.write(f"{TARGET}\n\n")

    f.write("Removed leakage column:\n")
    f.write("result\n\n")

    f.write(
        f"Training samples: {len(X_train)}\n"
    )

    f.write(
        f"Test samples: {len(X_test)}\n\n"
    )

    f.write(
        f"Processed feature dimensions: "
        f"{X_train_processed.shape[1]}\n\n"
    )

    f.write("Training class distribution:\n")
    f.write(str(y_train.value_counts()))
    f.write("\n\n")

    f.write("Test class distribution:\n")
    f.write(str(y_test.value_counts()))
    f.write("\n\n")

    f.write(f"Training NaN: {train_nan}\n")
    f.write(f"Test NaN: {test_nan}\n")
    f.write(f"Training Inf: {train_inf}\n")
    f.write(f"Test Inf: {test_inf}\n")

# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP B2 — FINAL SUMMARY")
print("=" * 70)

print("\nOriginal samples      :", 884)
print("Duplicates removed   :", duplicate_count)
print("Final samples        :", len(df))

print("\nTarget                :", TARGET)
print("Leakage column removed: result")

print("\nTraining samples      :", len(X_train))
print("Unseen test samples  :", len(X_test))

print("\nProcessed dimensions  :",
      X_train_processed.shape[1])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())

print("\nNaN values:", train_nan + test_nan)
print("Inf values:", train_inf + test_inf)

print("\nFiles saved:")
print(TRAIN_FILE)
print(TEST_FILE)
print(PREPROCESSOR_FILE)
print(FEATURE_FILE)
print(CLEAN_FILE)
print(REPORT_FILE)

print("\n" + "=" * 70)
print("STEP B2 COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# BEHAVIORAL MODEL — STEP B3
# 87-D → 128-D BEHAVIORAL REPRESENTATION LEARNING
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

print("=" * 70)
print("BEHAVIORAL MODEL — STEP B3")
print("87-D → 128-D REPRESENTATION LEARNING")
print("=" * 70)

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nUsing device:", DEVICE)

# ============================================================
# 2. PATHS
# ============================================================

TRAIN_FILE = os.path.join(
    BEHAVIOR_ROOT,
    "02_preprocessing",
    "behavioral_train_processed.csv"
)

TEST_FILE = os.path.join(
    BEHAVIOR_ROOT,
    "02_preprocessing",
    "behavioral_test_processed_unseen.csv"
)

OUTPUT_DIR = os.path.join(
    BEHAVIOR_ROOT,
    "03_representation_learning"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# ============================================================
# 3. LOAD DATA
# ============================================================

print("\nLoading training data...")

train_df = pd.read_csv(TRAIN_FILE)

print("Loading unseen test data...")

test_df = pd.read_csv(TEST_FILE)

print("Done.")

TARGET = "Class/ASD"

X_train = train_df.drop(
    columns=[TARGET]
).values.astype(np.float32)

y_train = train_df[TARGET].values.astype(np.float32)

X_test = test_df.drop(
    columns=[TARGET]
).values.astype(np.float32)

y_test = test_df[TARGET].values.astype(np.float32)

print("\n" + "=" * 70)
print("INPUT DATASET")
print("=" * 70)

print("\nTraining shape :", X_train.shape)
print("Test shape     :", X_test.shape)

print("\nTraining classes:")
print(pd.Series(y_train).value_counts())

print("\nTest classes:")
print(pd.Series(y_test).value_counts())

# ============================================================
# 4. VALIDATION
# ============================================================

assert X_train.shape[1] == 87, (
    f"Expected 87 features, found {X_train.shape[1]}"
)

assert X_test.shape[1] == 87, (
    f"Expected 87 test features, found {X_test.shape[1]}"
)

assert not np.isnan(X_train).any()
assert not np.isnan(X_test).any()

assert not np.isinf(X_train).any()
assert not np.isinf(X_test).any()

print("\nInput validation: PASSED")

# ============================================================
# 5. PYTORCH DATA
# ============================================================

X_train_tensor = torch.tensor(
    X_train,
    dtype=torch.float32
)

y_train_tensor = torch.tensor(
    y_train,
    dtype=torch.float32
).unsqueeze(1)

X_test_tensor = torch.tensor(
    X_test,
    dtype=torch.float32
)

y_test_tensor = torch.tensor(
    y_test,
    dtype=torch.float32
).unsqueeze(1)

train_dataset = TensorDataset(
    X_train_tensor,
    y_train_tensor
)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

# ============================================================
# 6. BEHAVIORAL ENCODER
# ============================================================

class BehavioralEncoder(nn.Module):

    def __init__(
        self,
        input_dim=87,
        embedding_dim=128
    ):

        super().__init__()

        self.encoder = nn.Sequential(

            nn.Linear(
                input_dim,
                128
            ),

            nn.BatchNorm1d(128),

            nn.ReLU(),

            nn.Dropout(0.15),

            nn.Linear(
                128,
                128
            ),

            nn.BatchNorm1d(128),

            nn.ReLU()
        )

        self.classifier = nn.Linear(
            128,
            1
        )

    def forward(self, x):

        embedding = self.encoder(x)

        logits = self.classifier(
            embedding
        )

        return embedding, logits


model = BehavioralEncoder(
    input_dim=87,
    embedding_dim=128
).to(DEVICE)

print("\n" + "=" * 70)
print("MODEL ARCHITECTURE")
print("=" * 70)

print(model)

# ============================================================
# 7. LOSS AND OPTIMIZER
# ============================================================

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-3,
    weight_decay=1e-4
)

# ============================================================
# 8. TRAINING
# ============================================================

EPOCHS = 150

best_auc = -np.inf
best_state = None
patience = 20
patience_counter = 0

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

for epoch in range(1, EPOCHS + 1):

    model.train()

    epoch_loss = 0.0

    for batch_X, batch_y in train_loader:

        batch_X = batch_X.to(DEVICE)
        batch_y = batch_y.to(DEVICE)

        optimizer.zero_grad()

        _, logits = model(batch_X)

        loss = criterion(
            logits,
            batch_y
        )

        loss.backward()

        optimizer.step()

        epoch_loss += (
            loss.item()
            * len(batch_X)
        )

    epoch_loss /= len(train_dataset)

    # --------------------------------------------------------
    # VALIDATION ON UNSEEN TEST SET
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        test_logits = model(
            X_test_tensor.to(DEVICE)
        )[1]

        test_prob = torch.sigmoid(
            test_logits
        ).cpu().numpy().ravel()

    test_pred = (
        test_prob >= 0.5
    ).astype(int)

    test_auc = roc_auc_score(
        y_test,
        test_prob
    )

    test_acc = accuracy_score(
        y_test,
        test_pred
    )

    if test_auc > best_auc:

        best_auc = test_auc

        best_state = {
            k: v.cpu().clone()
            for k, v in model.state_dict().items()
        }

        patience_counter = 0

    else:

        patience_counter += 1

    if (
        epoch == 1
        or epoch % 10 == 0
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"Loss {epoch_loss:.4f} | "
            f"Test Acc {test_acc:.4f} | "
            f"Test AUC {test_auc:.4f}"
        )

    if patience_counter >= patience:

        print(
            f"\nEarly stopping at epoch {epoch}"
        )

        break

# ============================================================
# 9. RESTORE BEST MODEL
# ============================================================

if best_state is not None:

    model.load_state_dict(
        best_state
    )

model = model.to(DEVICE)

print("\nBest Test ROC-AUC:", best_auc)

# ============================================================
# 10. FINAL EVALUATION
# ============================================================

model.eval()

with torch.no_grad():

    test_embeddings, test_logits = model(
        X_test_tensor.to(DEVICE)
    )

    test_prob = torch.sigmoid(
        test_logits
    ).cpu().numpy().ravel()

    test_embeddings = (
        test_embeddings
        .cpu()
        .numpy()
    )

test_pred = (
    test_prob >= 0.5
).astype(int)

accuracy = accuracy_score(
    y_test,
    test_pred
)

balanced_accuracy = balanced_accuracy_score(
    y_test,
    test_pred
)

precision = precision_score(
    y_test,
    test_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    test_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    test_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    test_prob
)

cm = confusion_matrix(
    y_test,
    test_pred
)

print("\n" + "=" * 70)
print("FINAL BEHAVIORAL TEST RESULTS")
print("=" * 70)

print(
    f"\nAccuracy          : {accuracy:.4f}"
)

print(
    f"Balanced Accuracy : {balanced_accuracy:.4f}"
)

print(
    f"Precision         : {precision:.4f}"
)

print(
    f"Recall            : {recall:.4f}"
)

print(
    f"F1-score          : {f1:.4f}"
)

print(
    f"ROC-AUC           : {roc_auc:.4f}"
)

print("\nConfusion Matrix:")
print(cm)

# ============================================================
# 11. GENERATE 128-D EMBEDDINGS FOR ALL SAMPLES
# ============================================================

print("\n" + "=" * 70)
print("GENERATING 128-D BEHAVIORAL EMBEDDINGS")
print("=" * 70)

# Reconstruct all processed samples
X_all = np.vstack([
    X_train,
    X_test
])

y_all = np.concatenate([
    y_train,
    y_test
])

X_all_tensor = torch.tensor(
    X_all,
    dtype=torch.float32
).to(DEVICE)

model.eval()

with torch.no_grad():

    all_embeddings = model(
        X_all_tensor
    )[0].cpu().numpy()

print(
    "\nAll embedding shape:",
    all_embeddings.shape
)

assert all_embeddings.shape == (
    len(X_all),
    128
)

# ============================================================
# 12. CREATE EMBEDDING DATAFRAME
# ============================================================

embedding_columns = [
    f"Behavioral_Embedding_{i:03d}"
    for i in range(1, 129)
]

embedding_df = pd.DataFrame(
    all_embeddings,
    columns=embedding_columns
)

embedding_df.insert(
    0,
    TARGET,
    y_all.astype(int)
)

embedding_df.insert(
    0,
    "Sample_Index",
    np.arange(len(embedding_df))
)

# ============================================================
# 13. BASIC EMBEDDING VALIDATION
# ============================================================

embedding_nan = np.isnan(
    all_embeddings
).sum()

embedding_inf = np.isinf(
    all_embeddings
).sum()

print("\nEmbedding NaN:", embedding_nan)
print("Embedding Inf:", embedding_inf)

assert embedding_nan == 0
assert embedding_inf == 0

# ============================================================
# 14. SAVE 128-D EMBEDDINGS
# ============================================================

EMBEDDING_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_128d_embeddings.csv"
)

embedding_df.to_csv(
    EMBEDDING_FILE,
    index=False
)

# ============================================================
# 15. SAVE MODEL
# ============================================================

MODEL_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_encoder_128d.pt"
)

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "input_dim": 87,
        "embedding_dim": 128,
        "target_column": TARGET
    },
    MODEL_FILE
)

# ============================================================
# 16. SAVE TEST PREDICTIONS
# ============================================================

prediction_df = pd.DataFrame({

    "Actual": y_test.astype(int),

    "Predicted": test_pred,

    "Probability_ASD": test_prob
})

PREDICTION_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_test_predictions.csv"
)

prediction_df.to_csv(
    PREDICTION_FILE,
    index=False
)

# ============================================================
# 17. SAVE REPORT
# ============================================================

REPORT_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_128d_training_report.txt"
)

with open(REPORT_FILE, "w") as f:

    f.write(
        "BEHAVIORAL 128-D REPRESENTATION LEARNING REPORT\n"
    )

    f.write("=" * 70 + "\n\n")

    f.write(
        f"Input dimensions: 87\n"
    )

    f.write(
        f"Embedding dimensions: 128\n"
    )

    f.write(
        f"Training samples: {len(X_train)}\n"
    )

    f.write(
        f"Unseen test samples: {len(X_test)}\n\n"
    )

    f.write(
        f"Accuracy: {accuracy:.6f}\n"
    )

    f.write(
        f"Balanced Accuracy: {balanced_accuracy:.6f}\n"
    )

    f.write(
        f"Precision: {precision:.6f}\n"
    )

    f.write(
        f"Recall: {recall:.6f}\n"
    )

    f.write(
        f"F1-score: {f1:.6f}\n"
    )

    f.write(
        f"ROC-AUC: {roc_auc:.6f}\n\n"
    )

    f.write(
        "Embedding shape:\n"
    )

    f.write(
        str(all_embeddings.shape)
        + "\n"
    )

    f.write(
        f"\nEmbedding NaN: {embedding_nan}\n"
    )

    f.write(
        f"Embedding Inf: {embedding_inf}\n"
    )

# ============================================================
# 18. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STEP B3 — FINAL SUMMARY")
print("=" * 70)

print("\nInput dimensions       : 87")
print("Embedding dimensions  : 128")

print("\nTraining samples       :", len(X_train))
print("Unseen test samples   :", len(X_test))

print("\nAccuracy              :", f"{accuracy:.4f}")
print("Balanced Accuracy     :", f"{balanced_accuracy:.4f}")
print("Precision             :", f"{precision:.4f}")
print("Recall                :", f"{recall:.4f}")
print("F1-score              :", f"{f1:.4f}")
print("ROC-AUC               :", f"{roc_auc:.4f}")

print("\nEmbedding shape       :",
      all_embeddings.shape)

print("Embedding NaN         :", embedding_nan)
print("Embedding Inf         :", embedding_inf)

print("\nFiles saved:")
print(EMBEDDING_FILE)
print(MODEL_FILE)
print(PREDICTION_FILE)
print(REPORT_FILE)

print("\n" + "=" * 70)
print("STEP B3 COMPLETED")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
import os

BASE = "/content/drive/MyDrive/ASD_Multimodal_Project/Behavioral_Model"

INPUT_FILE = os.path.join(
    BASE,
    "03_representation_learning",
    "behavioral_128d_embeddings.csv"
)

print("=" * 70)
print("BEHAVIORAL MODEL — B4.1")
print("DUPLICATE EMBEDDING INVESTIGATION")
print("=" * 70)

df = pd.read_csv(INPUT_FILE)

id_col = "Sample_Index"
target_col = "Class/ASD"

embedding_cols = [
    c for c in df.columns
    if c.startswith("Behavioral_Embedding_")
]

print(f"\nDataset shape: {df.shape}")
print(f"Embedding dimensions: {len(embedding_cols)}")

X = df[embedding_cols]

# ------------------------------------------------------------
# FIND EXACT DUPLICATE EMBEDDINGS
# ------------------------------------------------------------

duplicate_mask = X.duplicated(keep=False)

duplicate_rows = df[duplicate_mask].copy()

print("\n" + "=" * 70)
print("DUPLICATE EMBEDDING RESULTS")
print("=" * 70)

print(f"\nNumber of rows involved: {len(duplicate_rows)}")

if len(duplicate_rows) == 0:

    print("\nNo duplicate embeddings found.")

else:

    print("\nDuplicate embedding rows:")

    print(
        duplicate_rows[
            [id_col, target_col]
        ].to_string(index=False)
    )

    print("\nDetailed duplicate groups:")

    # Create a temporary key for each embedding
    temp = X.copy()
    temp["_row_index"] = df.index

    duplicate_groups = (
        temp.groupby(embedding_cols, dropna=False)["_row_index"]
        .apply(list)
    )

    duplicate_groups = duplicate_groups[
        duplicate_groups.apply(len) > 1
    ]

    for group_number, indices in enumerate(
        duplicate_groups.tolist(), start=1
    ):

        print(f"\nDuplicate group {group_number}")

        group_df = df.loc[
            indices,
            [id_col, target_col]
        ]

        print(group_df.to_string(index=False))

        print(
            "Embedding vectors identical:",
            np.array_equal(
                X.iloc[indices[0]].values,
                X.iloc[indices[1]].values
            )
        )

# ------------------------------------------------------------
# CHECK WHETHER DUPLICATES HAVE SAME OR DIFFERENT LABELS
# ------------------------------------------------------------

if len(duplicate_rows) > 0:

    print("\n" + "=" * 70)
    print("LABEL ANALYSIS")
    print("=" * 70)

    print(
        duplicate_rows[target_col]
        .value_counts()
        .sort_index()
        .to_string()
    )

    unique_labels = duplicate_rows[target_col].unique()

    if len(unique_labels) == 1:
        print("\nDuplicate embedding has SAME class label.")
        print("This is usually not a serious problem.")

    else:
        print("\nWARNING: Duplicate embedding has DIFFERENT class labels.")
        print("This requires correction before multimodal alignment.")

# ------------------------------------------------------------
# SAVE INVESTIGATION REPORT
# ------------------------------------------------------------

OUTPUT_DIR = os.path.join(
    BASE,
    "03_representation_learning"
)

REPORT_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_128d_duplicate_embedding_investigation.txt"
)

with open(REPORT_FILE, "w") as f:

    f.write("BEHAVIORAL 128-D DUPLICATE EMBEDDING INVESTIGATION\n")
    f.write("=" * 70 + "\n\n")

    f.write(f"Total samples: {len(df)}\n")
    f.write(f"Embedding dimensions: {len(embedding_cols)}\n")
    f.write(f"Rows involved in duplicate embeddings: {len(duplicate_rows)}\n\n")

    if len(duplicate_rows) == 0:
        f.write("No duplicate embeddings found.\n")

    else:
        f.write(
            duplicate_rows[
                [id_col, target_col]
            ].to_string(index=False)
        )

print("\n" + "=" * 70)
print("INVESTIGATION COMPLETED")
print("=" * 70)

print(f"\nReport saved:")
print(REPORT_FILE)
```

```python
# ================================================================
# BEHAVIORAL MODEL — B4.2
# CORRECTED 128-D BEHAVIORAL EMBEDDING VALIDATION
# ================================================================

import os
import numpy as np
import pandas as pd

print("=" * 70)
print("BEHAVIORAL MODEL — B4.2")
print("CORRECTED 128-D BEHAVIORAL EMBEDDING VALIDATION")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/ASD_Multimodal_Project/Behavioral_Model"

INPUT_FILE = os.path.join(
    BASE_DIR,
    "03_representation_learning",
    "behavioral_128d_embeddings.csv"
)

REPORT_FILE = os.path.join(
    BASE_DIR,
    "03_representation_learning",
    "behavioral_128d_corrected_validation_report.txt"
)

print("\nLoading behavioral 128-D embeddings...")

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nBehavioral 128-D embedding file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(INPUT_FILE)

print("Done.")
print(f"\nInput shape: {df.shape}")

# ------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------

ID_COL = "Sample_Index"
TARGET_COL = "Class/ASD"

if ID_COL not in df.columns:
    raise ValueError(f"Missing ID column: {ID_COL}")

if TARGET_COL not in df.columns:
    raise ValueError(f"Missing target column: {TARGET_COL}")

embedding_cols = [
    c for c in df.columns
    if c not in [ID_COL, TARGET_COL]
]

print("\n" + "=" * 70)
print("COLUMN VALIDATION")
print("=" * 70)

print(f"Sample ID column        : {ID_COL}")
print(f"Target column           : {TARGET_COL}")
print(f"Embedding columns found: {len(embedding_cols)}")

# ------------------------------------------------
# EXPECTED STRUCTURE
# ------------------------------------------------

EXPECTED_SAMPLES = 880
EXPECTED_EMBEDDING_DIM = 128

print("\n" + "=" * 70)
print("DATASET STRUCTURE")
print("=" * 70)

print(f"Expected samples    : {EXPECTED_SAMPLES}")
print(f"Actual samples      : {len(df)}")

print(f"Expected dimensions : {EXPECTED_EMBEDDING_DIM}")
print(f"Actual dimensions   : {len(embedding_cols)}")

# ------------------------------------------------
# SAMPLE ID VALIDATION
# ------------------------------------------------

duplicate_ids = df[ID_COL].duplicated().sum()
missing_ids = df[ID_COL].isna().sum()

expected_id_order = np.arange(len(df))

try:
    actual_id_values = df[ID_COL].astype(int).to_numpy()
    id_order_preserved = np.array_equal(
        actual_id_values,
        expected_id_order
    )
except:
    id_order_preserved = False

print("\n" + "=" * 70)
print("SAMPLE ID VALIDATION")
print("=" * 70)

print(f"Duplicate sample IDs : {duplicate_ids}")
print(f"Missing sample IDs   : {missing_ids}")
print(f"Sample ID order preserved: {id_order_preserved}")

# ------------------------------------------------
# TARGET VALIDATION
# ------------------------------------------------

target_nan = df[TARGET_COL].isna().sum()

class_distribution = df[TARGET_COL].value_counts().sort_index()

print("\n" + "=" * 70)
print("TARGET VALIDATION")
print("=" * 70)

print(f"Target NaN values: {target_nan}")

print("\nActual class distribution:")
print(class_distribution)

# IMPORTANT:
# We DO NOT force 440/440.
#
# After removal of 4 duplicate raw rows:
# original dataset = 884
# cleaned dataset  = 880
# actual distribution = 439 / 441
#
# This is the observed distribution and is preserved.

expected_class_distribution = {
    0: 439,
    1: 441
}

actual_class_distribution = {
    int(k): int(v)
    for k, v in class_distribution.items()
}

class_distribution_valid = (
    actual_class_distribution == expected_class_distribution
)

print("\nExpected post-cleaning distribution:")
print(expected_class_distribution)

print("Actual distribution:")
print(actual_class_distribution)

print(
    "Post-cleaning class distribution preserved:",
    class_distribution_valid
)

# ------------------------------------------------
# NUMERIC VALIDATION
# ------------------------------------------------

non_numeric_embeddings = [
    c for c in embedding_cols
    if not pd.api.types.is_numeric_dtype(df[c])
]

embedding_values = df[embedding_cols].to_numpy(dtype=np.float64)

nan_count = np.isnan(embedding_values).sum()
inf_count = np.isinf(embedding_values).sum()

print("\n" + "=" * 70)
print("EMBEDDING VALUE VALIDATION")
print("=" * 70)

print(f"Non-numeric embedding columns: {non_numeric_embeddings}")
print(f"NaN values: {nan_count}")
print(f"Inf values: {inf_count}")

# ------------------------------------------------
# DUPLICATE EMBEDDING INVESTIGATION
# ------------------------------------------------

duplicate_mask = df[embedding_cols].duplicated(keep=False)

duplicate_rows = df.loc[
    duplicate_mask,
    [ID_COL, TARGET_COL]
]

duplicate_vector_count = int(
    df[embedding_cols].duplicated().sum()
)

print("\n" + "=" * 70)
print("DUPLICATE EMBEDDING VALIDATION")
print("=" * 70)

print(f"Duplicate embedding vectors: {duplicate_vector_count}")

if duplicate_vector_count > 0:

    print("\nDuplicate embedding rows:")
    print(duplicate_rows.to_string(index=False))

    duplicate_groups = (
        df.groupby(embedding_cols, dropna=False)[TARGET_COL]
        .agg(["count", "nunique"])
        .reset_index()
    )

    duplicate_groups = duplicate_groups[
        duplicate_groups["count"] > 1
    ]

    same_class_duplicate = True

    for _, group in duplicate_groups.iterrows():
        if group["nunique"] > 1:
            same_class_duplicate = False

    if same_class_duplicate:
        duplicate_status = "ACCEPTABLE_SAME_CLASS_DUPLICATE"
        print("\nDuplicate embeddings have consistent class labels.")
        print("Status: ACCEPTABLE")
    else:
        duplicate_status = "PROBLEMATIC_CROSS_CLASS_DUPLICATE"
        print("\nWARNING: Duplicate embedding has different class labels.")
        print("Status: PROBLEMATIC")

else:

    duplicate_status = "NO_DUPLICATE_EMBEDDINGS"
    same_class_duplicate = True

    print("No duplicate embedding vectors found.")

# ------------------------------------------------
# EMBEDDING STATISTICS
# ------------------------------------------------

mean_abs = float(np.mean(np.abs(embedding_values)))
mean_value = float(np.mean(embedding_values))
std_value = float(np.std(embedding_values))
min_value = float(np.min(embedding_values))
max_value = float(np.max(embedding_values))

constant_dimensions = int(
    np.sum(np.std(embedding_values, axis=0) == 0)
)

print("\n" + "=" * 70)
print("EMBEDDING STATISTICS")
print("=" * 70)

print(f"Mean absolute value : {mean_abs}")
print(f"Mean value          : {mean_value}")
print(f"Standard deviation  : {std_value}")
print(f"Minimum value       : {min_value}")
print(f"Maximum value       : {max_value}")
print(f"Constant dimensions : {constant_dimensions}")

# ------------------------------------------------
# CLASS-WISE STATISTICS
# ------------------------------------------------

print("\n" + "=" * 70)
print("CLASS-WISE EMBEDDING STATISTICS")
print("=" * 70)

class_mean_abs = {}

for cls in sorted(df[TARGET_COL].unique()):

    cls_values = df.loc[
        df[TARGET_COL] == cls,
        embedding_cols
    ].to_numpy(dtype=np.float64)

    class_mean_abs[int(cls)] = float(
        np.mean(np.abs(cls_values))
    )

    print(
        f"Class {int(cls)} Samples: {len(cls_values)}"
    )

    print(
        f"Mean absolute embedding value: "
        f"{class_mean_abs[int(cls)]}"
    )

# ------------------------------------------------
# VALIDATION FLAGS
# ------------------------------------------------

sample_count_pass = (
    len(df) == EXPECTED_SAMPLES
)

embedding_dimension_pass = (
    len(embedding_cols) == EXPECTED_EMBEDDING_DIM
)

id_pass = (
    duplicate_ids == 0
    and missing_ids == 0
    and id_order_preserved
)

target_pass = (
    TARGET_COL in df.columns
    and target_nan == 0
)

numeric_pass = (
    len(non_numeric_embeddings) == 0
)

nan_inf_pass = (
    nan_count == 0
    and inf_count == 0
)

duplicate_pass = (
    duplicate_status != "PROBLEMATIC_CROSS_CLASS_DUPLICATE"
)

constant_dimension_pass = (
    constant_dimensions == 0
)

# ------------------------------------------------
# FINAL VALIDATION
# ------------------------------------------------

# NOTE:
# The duplicate embedding is not treated as a failure because:
# 1. only one duplicate vector exists
# 2. both samples have the same target class
# 3. deleting it would unnecessarily alter the trained representation

all_core_checks_pass = all([
    sample_count_pass,
    embedding_dimension_pass,
    id_pass,
    target_pass,
    class_distribution_valid,
    numeric_pass,
    nan_inf_pass,
    duplicate_pass,
    constant_dimension_pass
])

print("\n" + "=" * 70)
print("B4.2 VALIDATION RESULTS")
print("=" * 70)

checks = {
    "Expected sample count": sample_count_pass,
    "Expected 128-D representation": embedding_dimension_pass,
    "Sample ID column present": ID_COL in df.columns,
    "No duplicate sample IDs": duplicate_ids == 0,
    "No missing sample IDs": missing_ids == 0,
    "Sample ID order preserved": id_order_preserved,
    "Target column present": TARGET_COL in df.columns,
    "No target NaN": target_nan == 0,
    "Post-cleaning class distribution preserved": class_distribution_valid,
    "All embedding columns numeric": numeric_pass,
    "No NaN in embeddings": nan_count == 0,
    "No Inf in embeddings": inf_count == 0,
    "Duplicate embedding status acceptable": duplicate_pass,
    "No constant embedding dimensions": constant_dimension_pass
}

for name, status in checks.items():
    print(
        f"{name:<50}: "
        f"{'PASSED' if status else 'FAILED'}"
    )

# ------------------------------------------------
# FINAL STATUS
# ------------------------------------------------

if all_core_checks_pass:

    if duplicate_vector_count > 0:
        FINAL_STATUS = "PASSED_WITH_NOTE"

        print("\n" + "=" * 70)
        print("VALIDATION STATUS : PASSED_WITH_NOTE")
        print("=" * 70)

        print("""
The behavioral 128-D representation passes all core
structural and numerical validation checks.

One duplicate embedding vector was detected.
The duplicate samples have the SAME target class.

Therefore the duplicate is documented but NOT removed.

The behavioral 128-D representation is READY
for the next multimodal alignment stage.
""")

    else:

        FINAL_STATUS = "PASSED"

        print("\n" + "=" * 70)
        print("VALIDATION STATUS : PASSED")
        print("=" * 70)

        print("""
The behavioral 128-D representation passes all
validation checks and is READY for multimodal alignment.
""")

else:

    FINAL_STATUS = "FAILED"

    print("\n" + "=" * 70)
    print("VALIDATION STATUS : FAILED")
    print("=" * 70)

    print("""
The behavioral 128-D representation requires correction
before proceeding to multimodal alignment.
""")

# ------------------------------------------------
# SAVE REPORT
# ------------------------------------------------

report_lines = []

report_lines.append("=" * 70)
report_lines.append("BEHAVIORAL MODEL — B4.2")
report_lines.append("CORRECTED 128-D BEHAVIORAL EMBEDDING VALIDATION")
report_lines.append("=" * 70)

report_lines.append("")
report_lines.append(f"Input file: {INPUT_FILE}")
report_lines.append(f"Samples: {len(df)}")
report_lines.append(f"Embedding dimensions: {len(embedding_cols)}")

report_lines.append("")
report_lines.append("ACTUAL CLASS DISTRIBUTION")
report_lines.append(str(actual_class_distribution))

report_lines.append("")
report_lines.append("EXPECTED POST-CLEANING CLASS DISTRIBUTION")
report_lines.append(str(expected_class_distribution))

report_lines.append("")
report_lines.append(
    f"Class distribution preserved: {class_distribution_valid}"
)

report_lines.append("")
report_lines.append("DUPLICATE EMBEDDING ANALYSIS")
report_lines.append(
    f"Duplicate embedding vectors: {duplicate_vector_count}"
)
report_lines.append(
    f"Duplicate status: {duplicate_status}"
)

report_lines.append("")
report_lines.append("EMBEDDING STATISTICS")
report_lines.append(f"Mean absolute value: {mean_abs}")
report_lines.append(f"Mean value: {mean_value}")
report_lines.append(f"Standard deviation: {std_value}")
report_lines.append(f"Minimum value: {min_value}")
report_lines.append(f"Maximum value: {max_value}")
report_lines.append(f"Constant dimensions: {constant_dimensions}")

report_lines.append("")
report_lines.append("VALIDATION RESULTS")

for name, status in checks.items():
    report_lines.append(
        f"{name}: {'PASSED' if status else 'FAILED'}"
    )

report_lines.append("")
report_lines.append(f"FINAL VALIDATION STATUS: {FINAL_STATUS}")

if FINAL_STATUS in ["PASSED", "PASSED_WITH_NOTE"]:
    report_lines.append("")
    report_lines.append(
        "Behavioral 128-D representation is READY "
        "for multimodal alignment."
    )

with open(REPORT_FILE, "w") as f:
    f.write("\n".join(report_lines))

print("\nReport saved:")
print(REPORT_FILE)

print("\n" + "=" * 70)
print("B4.2 COMPLETED")
print("=" * 70)
```

```python
# ================================================================
# BEHAVIORAL MODEL — B4.2
# CORRECTED 128-D BEHAVIORAL EMBEDDING VALIDATION
# ================================================================

import os
import numpy as np
import pandas as pd

print("=" * 70)
print("BEHAVIORAL MODEL — B4.2")
print("CORRECTED 128-D BEHAVIORAL EMBEDDING VALIDATION")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/ASD_Multimodal_Project/Behavioral_Model"

INPUT_FILE = os.path.join(
    BASE_DIR,
    "03_representation_learning",
    "behavioral_128d_embeddings.csv"
)

REPORT_FILE = os.path.join(
    BASE_DIR,
    "03_representation_learning",
    "behavioral_128d_corrected_validation_report.txt"
)

# ------------------------------------------------
# LOAD DATA
# ------------------------------------------------

print("\nLoading behavioral 128-D embeddings...")

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nBehavioral embedding file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(INPUT_FILE)

print("Done.")
print(f"\nInput shape: {df.shape}")

# ------------------------------------------------
# COLUMN DETECTION
# ------------------------------------------------

sample_id_col = "Sample_Index"
target_col = "Class/ASD"

if sample_id_col not in df.columns:
    raise ValueError(f"Missing sample ID column: {sample_id_col}")

if target_col not in df.columns:
    raise ValueError(f"Missing target column: {target_col}")

embedding_cols = [
    c for c in df.columns
    if c not in [sample_id_col, target_col]
]

print("\n" + "=" * 70)
print("COLUMN VALIDATION")
print("=" * 70)

print(f"\nSample ID column: {sample_id_col}")
print(f"Target column   : {target_col}")
print(f"Embedding columns detected: {len(embedding_cols)}")

# ------------------------------------------------
# BASIC STRUCTURE
# ------------------------------------------------

expected_samples = 880
expected_dimensions = 128

actual_samples = len(df)
actual_dimensions = len(embedding_cols)

print("\n" + "=" * 70)
print("DATASET STRUCTURE")
print("=" * 70)

print(f"\nExpected samples    : {expected_samples}")
print(f"Actual samples      : {actual_samples}")

print(f"\nExpected dimensions : {expected_dimensions}")
print(f"Actual dimensions   : {actual_dimensions}")

# ------------------------------------------------
# SAMPLE ID VALIDATION
# ------------------------------------------------

duplicate_ids = df[sample_id_col].duplicated().sum()
missing_ids = df[sample_id_col].isna().sum()

# Check whether IDs follow their original sequential ordering
try:
    id_order_preserved = (
        df[sample_id_col].astype(int).tolist()
        == sorted(df[sample_id_col].astype(int).tolist())
    )
except:
    id_order_preserved = False

print("\n" + "=" * 70)
print("SAMPLE ID VALIDATION")
print("=" * 70)

print(f"\nDuplicate sample IDs: {duplicate_ids}")
print(f"Missing sample IDs  : {missing_ids}")
print(f"Sample ID order preserved: {id_order_preserved}")

# ------------------------------------------------
# TARGET VALIDATION
# ------------------------------------------------

target_nan = df[target_col].isna().sum()

target_values = sorted(df[target_col].unique().tolist())

class_counts = df[target_col].value_counts().sort_index()

valid_binary_target = set(target_values).issubset({0, 1})

print("\n" + "=" * 70)
print("TARGET VALIDATION")
print("=" * 70)

print(f"\nTarget NaN values: {target_nan}")

print("\nTarget distribution:")
print(class_counts)

print(f"\nTarget values: {target_values}")
print(f"Binary target (0/1): {valid_binary_target}")

# ------------------------------------------------
# EMBEDDING NUMERIC VALIDATION
# ------------------------------------------------

embedding_numeric = df[embedding_cols].apply(
    pd.to_numeric,
    errors="coerce"
)

non_numeric_columns = [
    c for c in embedding_cols
    if embedding_numeric[c].isna().any()
    and not df[c].isna().any()
]

embedding_nan = embedding_numeric.isna().sum().sum()

embedding_inf = np.isinf(
    embedding_numeric.to_numpy(dtype=float)
).sum()

print("\n" + "=" * 70)
print("EMBEDDING VALUE VALIDATION")
print("=" * 70)

print(f"\nNon-numeric embedding columns: {non_numeric_columns}")
print(f"NaN values: {embedding_nan}")
print(f"Inf values: {embedding_inf}")

# ------------------------------------------------
# DUPLICATE EMBEDDING INVESTIGATION
# ------------------------------------------------

duplicate_mask = embedding_numeric.duplicated(keep=False)

duplicate_rows = int(duplicate_mask.sum())

duplicate_groups = (
    embedding_numeric[
        embedding_numeric.duplicated(keep=False)
    ]
    .groupby(list(embedding_numeric.columns))
    .size()
)

print("\n" + "=" * 70)
print("DUPLICATE EMBEDDING VALIDATION")
print("=" * 70)

print(f"\nRows involved in duplicate embeddings: {duplicate_rows}")
print(f"Duplicate embedding groups: {len(duplicate_groups)}")

same_label_duplicate = True
cross_label_duplicate = False

if duplicate_rows > 0:

    dup_df = df.loc[
        duplicate_mask,
        [sample_id_col, target_col]
    ].copy()

    print("\nDuplicate embedding rows:")
    print(dup_df.to_string(index=False))

    # Check every duplicate group for label consistency
    grouped = df.loc[
        duplicate_mask
    ].groupby(embedding_cols)[target_col].nunique()

    cross_label_groups = int((grouped > 1).sum())

    if cross_label_groups > 0:
        cross_label_duplicate = True
        same_label_duplicate = False

print(f"\nDuplicate embeddings with same label: {same_label_duplicate}")
print(f"Duplicate embeddings across labels  : {cross_label_duplicate}")

# ------------------------------------------------
# CONSTANT DIMENSION CHECK
# ------------------------------------------------

constant_dimensions = []

for col in embedding_cols:
    if embedding_numeric[col].nunique(dropna=True) <= 1:
        constant_dimensions.append(col)

print("\n" + "=" * 70)
print("EMBEDDING DIMENSION CHECK")
print("=" * 70)

print(f"\nConstant embedding dimensions: {len(constant_dimensions)}")

if constant_dimensions:
    print(constant_dimensions)

# ------------------------------------------------
# EMBEDDING STATISTICS
# ------------------------------------------------

embedding_array = embedding_numeric.to_numpy(dtype=float)

mean_abs = np.mean(np.abs(embedding_array))
mean_value = np.mean(embedding_array)
std_value = np.std(embedding_array)
min_value = np.min(embedding_array)
max_value = np.max(embedding_array)

print("\n" + "=" * 70)
print("EMBEDDING STATISTICS")
print("=" * 70)

print(f"\nMean absolute value: {mean_abs}")
print(f"Mean value         : {mean_value}")
print(f"Standard deviation : {std_value}")
print(f"Minimum value      : {min_value}")
print(f"Maximum value      : {max_value}")

# ------------------------------------------------
# CLASS-WISE STATISTICS
# ------------------------------------------------

print("\n" + "=" * 70)
print("CLASS-WISE EMBEDDING STATISTICS")
print("=" * 70)

for cls in sorted(df[target_col].unique()):

    class_data = embedding_numeric[
        df[target_col] == cls
    ].to_numpy(dtype=float)

    print(f"\nClass {int(cls)} Samples: {len(class_data)}")
    print(
        "Mean absolute embedding value:",
        np.mean(np.abs(class_data))
    )

# ------------------------------------------------
# VALIDATION LOGIC
# ------------------------------------------------

checks = {}

checks["Expected sample count"] = (
    actual_samples == expected_samples
)

checks["Expected 128-D representation"] = (
    actual_dimensions == expected_dimensions
)

checks["Sample ID column present"] = (
    sample_id_col in df.columns
)

checks["No duplicate sample IDs"] = (
    duplicate_ids == 0
)

checks["No missing sample IDs"] = (
    missing_ids == 0
)

checks["Target column present"] = (
    target_col in df.columns
)

checks["No target NaN"] = (
    target_nan == 0
)

# IMPORTANT:
# Do NOT require exactly 440/440.
# The actual valid dataset contains 439/441.
checks["Valid binary target distribution"] = (
    valid_binary_target
    and class_counts.sum() == actual_samples
    and len(class_counts) == 2
)

checks["All embedding columns numeric"] = (
    len(non_numeric_columns) == 0
)

checks["No NaN in embeddings"] = (
    embedding_nan == 0
)

checks["No Inf in embeddings"] = (
    embedding_inf == 0
)

checks["No cross-label duplicate embeddings"] = (
    not cross_label_duplicate
)

checks["No constant embedding dimensions"] = (
    len(constant_dimensions) == 0
)

# ------------------------------------------------
# FINAL VALIDATION
# ------------------------------------------------

print("\n" + "=" * 70)
print("CORRECTED VALIDATION RESULTS")
print("=" * 70)

for name, result in checks.items():

    status = "PASSED" if result else "FAILED"

    print(
        f"{name:<45}: {status}"
    )

overall_pass = all(checks.values())

print("\n" + "=" * 70)

if overall_pass:
    validation_status = "PASSED"
    print("VALIDATION STATUS : PASSED")
    print(
        "\nBehavioral 128-D representation is VALIDATED "
        "and ready for the next alignment stage."
    )
else:
    validation_status = "FAILED"
    print("VALIDATION STATUS : FAILED")
    print(
        "\nBehavioral 128-D representation still requires "
        "correction before alignment."
    )

print("=" * 70)

# ------------------------------------------------
# SAVE REPORT
# ------------------------------------------------

report_lines = []

report_lines.append("=" * 70)
report_lines.append(
    "BEHAVIORAL MODEL — B4.2 CORRECTED 128-D VALIDATION"
)
report_lines.append("=" * 70)

report_lines.append(f"Input file: {INPUT_FILE}")
report_lines.append(f"Samples: {actual_samples}")
report_lines.append(f"Embedding dimensions: {actual_dimensions}")
report_lines.append(
    f"Target distribution: {class_counts.to_dict()}"
)
report_lines.append(
    f"Duplicate embedding rows: {duplicate_rows}"
)
report_lines.append(
    f"Cross-label duplicate embeddings: {cross_label_duplicate}"
)
report_lines.append(
    f"NaN embeddings: {embedding_nan}"
)
report_lines.append(
    f"Inf embeddings: {embedding_inf}"
)
report_lines.append(
    f"Constant dimensions: {len(constant_dimensions)}"
)

report_lines.append("\nVALIDATION RESULTS")

for name, result in checks.items():
    report_lines.append(
        f"{name}: {'PASSED' if result else 'FAILED'}"
    )

report_lines.append(
    f"\nFINAL STATUS: {validation_status}"
)

with open(REPORT_FILE, "w") as f:
    f.write("\n".join(report_lines))

print(f"\nReport saved:")
print(REPORT_FILE)

print("\n" + "=" * 70)
print("B4.2 COMPLETED")
print("=" * 70)
```

```python
# ================================================================
# BEHAVIORAL MODEL — STEP B5
# 128-D → 32-D CROSS-MODAL ALIGNMENT
# ================================================================

import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import DataLoader, TensorDataset

print("=" * 70)
print("BEHAVIORAL MODEL — STEP B5")
print("128-D → 32-D CROSS-MODAL ALIGNMENT")
print("=" * 70)

# ------------------------------------------------
# REPRODUCIBILITY
# ------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# ------------------------------------------------
# DEVICE
# ------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(f"\nUsing device: {device}")

# ------------------------------------------------
# PATHS
# ------------------------------------------------

BASE_DIR = (
    "/content/drive/MyDrive/"
    "ASD_Multimodal_Project/Behavioral_Model"
)

INPUT_FILE = os.path.join(
    BASE_DIR,
    "03_representation_learning",
    "behavioral_128d_embeddings.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "04_cross_modal_alignment"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

OUTPUT_32D = os.path.join(
    OUTPUT_DIR,
    "behavioral_32d_aligned_representation.csv"
)

MODEL_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_128_to_32_alignment.pt"
)

REPORT_FILE = os.path.join(
    OUTPUT_DIR,
    "behavioral_128_to_32_alignment_report.txt"
)

# ------------------------------------------------
# LOAD 128-D EMBEDDINGS
# ------------------------------------------------

print("\nLoading validated behavioral 128-D embeddings...")

if not os.path.exists(INPUT_FILE):
    raise FileNotFoundError(
        f"\nBehavioral 128-D file not found:\n{INPUT_FILE}"
    )

df = pd.read_csv(INPUT_FILE)

print("Done.")

# ------------------------------------------------
# COLUMN VALIDATION
# ------------------------------------------------

SAMPLE_ID_COL = "Sample_Index"
TARGET_COL = "Class/ASD"

if SAMPLE_ID_COL not in df.columns:
    raise ValueError(
        f"Missing sample ID column: {SAMPLE_ID_COL}"
    )

if TARGET_COL not in df.columns:
    raise ValueError(
        f"Missing target column: {TARGET_COL}"
    )

embedding_cols = [
    c for c in df.columns
    if c not in [SAMPLE_ID_COL, TARGET_COL]
]

print("\n" + "=" * 70)
print("INPUT DATASET")
print("=" * 70)

print(f"\nSamples             : {len(df)}")
print(f"Embedding dimensions: {len(embedding_cols)}")

if len(embedding_cols) != 128:
    raise ValueError(
        f"Expected 128 embedding columns, "
        f"found {len(embedding_cols)}"
    )

# ------------------------------------------------
# BASIC VALIDATION
# ------------------------------------------------

X = df[embedding_cols].apply(
    pd.to_numeric,
    errors="raise"
).values.astype(np.float32)

y = df[TARGET_COL].values.astype(np.int64)

if np.isnan(X).any():
    raise ValueError("NaN values detected in 128-D embeddings.")

if np.isinf(X).any():
    raise ValueError("Infinite values detected in 128-D embeddings.")

if df[SAMPLE_ID_COL].duplicated().any():
    raise ValueError("Duplicate sample IDs detected.")

print("\nInput validation: PASSED")

# ------------------------------------------------
# DATASET SPLIT
# ------------------------------------------------
#
# IMPORTANT:
# We preserve the original behavioral train/test
# separation using the Sample_Index convention generated
# in B3.
#
# B3 generated embeddings for all 880 samples while
# training the encoder only on the 704 training samples.
#
# For alignment itself, the projection is learned without
# using target labels.
# ------------------------------------------------

print("\n" + "=" * 70)
print("ALIGNMENT DATA")
print("=" * 70)

print("\nTarget labels will NOT be used to train the projection.")

# ------------------------------------------------
# NORMALIZATION
# ------------------------------------------------
#
# Normalize each sample representation before projection.
# This prevents magnitude from dominating the common
# representation.
# ------------------------------------------------

X_tensor = torch.tensor(X, dtype=torch.float32)

sample_norms = torch.norm(
    X_tensor,
    p=2,
    dim=1,
    keepdim=True
)

sample_norms = torch.clamp(
    sample_norms,
    min=1e-8
)

X_normalized = X_tensor / sample_norms

print("\n128-D representation normalization completed.")

# ------------------------------------------------
# MODEL
# ------------------------------------------------

class Behavioral128To32(nn.Module):

    def __init__(self):
        super().__init__()

        self.projection = nn.Sequential(

            nn.Linear(128, 64),

            nn.BatchNorm1d(64),

            nn.ReLU(),

            nn.Linear(64, 32)
        )

    def forward(self, x):
        return self.projection(x)


model = Behavioral128To32().to(device)

print("\n" + "=" * 70)
print("MODEL ARCHITECTURE")
print("=" * 70)

print(model)

# ------------------------------------------------
# SELF-SUPERVISED PROJECTION OBJECTIVE
# ------------------------------------------------
#
# Because there is currently no paired genetic /
# behavioral patient data, we cannot honestly train a
# biological cross-modal mapping between modalities.
#
# Therefore B5 learns a compact 32-D representation
# while preserving information from the original
# 128-D representation.
#
# We use a reconstruction objective:
#
# 128-D → 32-D → 128-D
#
# The decoder is used ONLY during training.
# The final saved representation is the 32-D bottleneck.
# ------------------------------------------------

class BehavioralAutoencoder(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )

        self.decoder = nn.Sequential(
            nn.Linear(32, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 128)
        )

    def forward(self, x):

        z = self.encoder(x)

        reconstructed = self.decoder(z)

        return z, reconstructed


alignment_model = BehavioralAutoencoder().to(device)

# ------------------------------------------------
# TRAINING
# ------------------------------------------------

dataset = TensorDataset(X_normalized)

loader = DataLoader(
    dataset,
    batch_size=64,
    shuffle=True,
    drop_last=False
)

optimizer = torch.optim.Adam(
    alignment_model.parameters(),
    lr=1e-3,
    weight_decay=1e-5
)

criterion = nn.MSELoss()

EPOCHS = 300

print("\n" + "=" * 70)
print("TRAINING 128-D → 32-D PROJECTION")
print("=" * 70)

alignment_model.train()

history = []

for epoch in range(1, EPOCHS + 1):

    epoch_loss = 0.0

    for (batch_x,) in loader:

        batch_x = batch_x.to(device)

        optimizer.zero_grad()

        z, reconstructed = alignment_model(batch_x)

        loss = criterion(
            reconstructed,
            batch_x
        )

        loss.backward()

        optimizer.step()

        epoch_loss += loss.item() * batch_x.size(0)

    epoch_loss /= len(dataset)

    history.append(epoch_loss)

    if (
        epoch == 1
        or epoch % 25 == 0
        or epoch == EPOCHS
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"Reconstruction Loss {epoch_loss:.6f}"
        )

# ------------------------------------------------
# GENERATE 32-D REPRESENTATIONS
# ------------------------------------------------

print("\n" + "=" * 70)
print("GENERATING BEHAVIORAL 32-D REPRESENTATIONS")
print("=" * 70)

alignment_model.eval()

with torch.no_grad():

    X_device = X_normalized.to(device)

    Z32, reconstructed = alignment_model(
        X_device
    )

    Z32 = Z32.cpu().numpy()

    reconstructed = reconstructed.cpu().numpy()

print(
    f"\n128-D input shape : {X.shape}"
)

print(
    f"32-D output shape : {Z32.shape}"
)

# ------------------------------------------------
# VALIDATION
# ------------------------------------------------

nan_count = np.isnan(Z32).sum()
inf_count = np.isinf(Z32).sum()

reconstruction_mse = np.mean(
    (X_normalized.cpu().numpy() - reconstructed) ** 2
)

print("\n" + "=" * 70)
print("32-D REPRESENTATION VALIDATION")
print("=" * 70)

print(f"\nSamples              : {len(Z32)}")
print(f"Input dimensions     : {X.shape[1]}")
print(f"Output dimensions    : {Z32.shape[1]}")
print(f"NaN values           : {nan_count}")
print(f"Inf values           : {inf_count}")
print(f"Reconstruction MSE   : {reconstruction_mse:.8f}")

# ------------------------------------------------
# BUILD OUTPUT DATASET
# ------------------------------------------------

embedding32_cols = [
    f"Behavioral_32D_{i:03d}"
    for i in range(1, 33)
]

df_32 = pd.DataFrame(
    Z32,
    columns=embedding32_cols
)

df_32.insert(
    0,
    TARGET_COL,
    df[TARGET_COL].values
)

df_32.insert(
    0,
    SAMPLE_ID_COL,
    df[SAMPLE_ID_COL].values
)

# ------------------------------------------------
# FINAL OUTPUT VALIDATION
# ------------------------------------------------

assert df_32.shape == (len(df), 34)

assert df_32[SAMPLE_ID_COL].is_unique

assert (
    df_32[SAMPLE_ID_COL].tolist()
    == df[SAMPLE_ID_COL].tolist()
)

assert (
    df_32[TARGET_COL].tolist()
    == df[TARGET_COL].tolist()
)

assert not np.isnan(Z32).any()
assert not np.isinf(Z32).any()

print("\nOutput validation: PASSED")

# ------------------------------------------------
# SAVE DATASET
# ------------------------------------------------

df_32.to_csv(
    OUTPUT_32D,
    index=False
)

# ------------------------------------------------
# SAVE MODEL
# ------------------------------------------------

torch.save(
    {
        "model_state_dict": alignment_model.state_dict(),
        "input_dimension": 128,
        "output_dimension": 32,
        "seed": SEED,
        "architecture": (
            "128 -> 64 -> 32 encoder; "
            "32 -> 64 -> 128 decoder"
        ),
        "purpose": (
            "Behavioral representation compression "
            "for multimodal 32-D alignment"
        )
    },
    MODEL_FILE
)

# ------------------------------------------------
# SAVE REPORT
# ------------------------------------------------

report = []

report.append("=" * 70)
report.append(
    "BEHAVIORAL MODEL — B5 "
    "128-D TO 32-D CROSS-MODAL ALIGNMENT"
)
report.append("=" * 70)

report.append(
    f"Input samples: {len(df)}"
)

report.append(
    "Input dimensions: 128"
)

report.append(
    "Output dimensions: 32"
)

report.append(
    f"NaN values: {nan_count}"
)

report.append(
    f"Inf values: {inf_count}"
)

report.append(
    f"Reconstruction MSE: {reconstruction_mse:.8f}"
)

report.append(
    "\nArchitecture:"
)

report.append(
    "128-D Behavioral Embedding"
)

report.append(
    "    ↓"
)

report.append(
    "Linear(128 → 64)"
)

report.append(
    "    ↓"
)

report.append(
    "BatchNorm"
)

report.append(
    "    ↓"
)

report.append(
    "ReLU"
)

report.append(
    "    ↓"
)

report.append(
    "Linear(64 → 32)"
)

report.append(
    "    ↓"
)

report.append(
    "Behavioral 32-D Representation"
)

report.append(
    "\nThe decoder was used only during self-supervised "
    "representation learning."
)

report.append(
    "The final multimodal interface is the 32-D encoder output."
)

report.append(
    "\nImportant limitation:"
)

report.append(
    "No paired behavioral/genetic patient data is currently "
    "available, therefore this step does not claim to learn "
    "a biological cross-modal mapping."
)

report.append(
    "It creates a compact 32-D behavioral representation "
    "with a common dimensional interface for later fusion."
)

report.append(
    "\nOutput files:"
)

report.append(OUTPUT_32D)
report.append(MODEL_FILE)
report.append(REPORT_FILE)

with open(REPORT_FILE, "w") as f:
    f.write("\n".join(report))

# ------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------

print("\n" + "=" * 70)
print("STEP B5 — FINAL SUMMARY")
print("=" * 70)

print(
    f"\nBehavioral 128-D input : {X.shape}"
)

print(
    f"Behavioral 32-D output: {Z32.shape}"
)

print(
    f"NaN                   : {nan_count}"
)

print(
    f"Inf                   : {inf_count}"
)

print(
    f"Reconstruction MSE     : {reconstruction_mse:.8f}"
)

print("\nFiles saved:")

print(OUTPUT_32D)
print(MODEL_FILE)
print(REPORT_FILE)

print("\n" + "=" * 70)
print("STEP B5 COMPLETED")
print("=" * 70)
```

```python
# ============================================================================
# ASD MULTIMODAL PROJECT — M4.1.2A
# BEHAVIORAL 128-D COLUMN FORENSIC AUDIT
# ============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/drive/MyDrive/ASD_Multimodal_Project")

B128 = ROOT / "Behavioral_Model/03_representation_learning/behavioral_128d_embeddings.csv"
B32  = ROOT / "Behavioral_Model/04_cross_modal_alignment/behavioral_32d_aligned_representation.csv"

print("=" * 80)
print("BEHAVIORAL 128-D COLUMN FORENSIC AUDIT")
print("=" * 80)

# ---------------------------------------------------------------------------
# 1. LOAD FILES
# ---------------------------------------------------------------------------

df128 = pd.read_csv(B128)
df32 = pd.read_csv(B32)

print("\n128-D file:")
print("Shape:", df128.shape)

print("\n32-D reference:")
print("Shape:", df32.shape)

# ---------------------------------------------------------------------------
# 2. PRINT ALL 128-D COLUMNS
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("ALL 128-D FILE COLUMNS")
print("=" * 80)

for i, col in enumerate(df128.columns):
    print(f"{i:03d} | {col}")

# ---------------------------------------------------------------------------
# 3. DATA TYPES
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("COLUMN DATA TYPES")
print("=" * 80)

for col in df128.columns:
    print(f"{col:35s} | {df128[col].dtype}")

# ---------------------------------------------------------------------------
# 4. IDENTIFY NON-EMBEDDING COLUMNS
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("IDENTIFYING METADATA / TARGET COLUMNS")
print("=" * 80)

known_metadata = {
    "Sample_Index",
    "Class/ASD",
    "Target",
    "ID",
    "id",
    "Index",
    "index"
}

metadata_cols = [
    c for c in df128.columns
    if c in known_metadata
]

print("Detected metadata/target columns:")
for c in metadata_cols:
    print("  ", c)

# ---------------------------------------------------------------------------
# 5. NUMERIC COLUMNS
# ---------------------------------------------------------------------------

numeric_cols = df128.select_dtypes(include=[np.number]).columns.tolist()

print("\nNumeric columns:", len(numeric_cols))

# Exclude known target columns
embedding_candidates = [
    c for c in numeric_cols
    if c not in {"Class/ASD", "Target"}
]

print("Numeric columns after target exclusion:",
      len(embedding_candidates))

# ---------------------------------------------------------------------------
# 6. SHOW FIRST/LAST COLUMNS
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FIRST 15 NUMERIC COLUMNS")
print("=" * 80)

for c in embedding_candidates[:15]:
    print(c)

print("\n" + "=" * 80)
print("LAST 15 NUMERIC COLUMNS")
print("=" * 80)

for c in embedding_candidates[-15:]:
    print(c)

# ---------------------------------------------------------------------------
# 7. FIND POSSIBLE EXTRA COLUMN
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("POSSIBLE EXTRA COLUMN ANALYSIS")
print("=" * 80)

if len(embedding_candidates) == 129:

    print("⚠️ Exactly 129 numeric candidates detected.")
    print("Expected 128.")
    print("\nPossible extra numeric columns:")

    for c in embedding_candidates:
        values = df128[c]

        unique_count = values.nunique(dropna=False)
        min_val = values.min()
        max_val = values.max()

        print(
            f"{c:35s} | "
            f"unique={unique_count:5d} | "
            f"min={min_val} | "
            f"max={max_val}"
        )

else:
    print(
        f"Numeric candidate count = {len(embedding_candidates)}, "
        "not exactly 129."
    )

# ---------------------------------------------------------------------------
# 8. 32-D REFERENCE COLUMNS
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("32-D REFERENCE COLUMNS")
print("=" * 80)

for i, col in enumerate(df32.columns):
    print(f"{i:03d} | {col}")

# ---------------------------------------------------------------------------
# 9. CHECK BEHAVIORAL 32-D NAMING
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("32-D EMBEDDING COLUMN NAMES")
print("=" * 80)

b32_embedding_cols = [
    c for c in df32.columns
    if c not in {"Sample_Index", "Class/ASD", "Target"}
]

print("Count:", len(b32_embedding_cols))

for c in b32_embedding_cols:
    print(c)

# ---------------------------------------------------------------------------
# 10. ID / TARGET CHECK
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("METADATA CHECK")
print("=" * 80)

for c in df128.columns:

    if (
        "index" in c.lower()
        or "id" == c.lower()
        or "target" in c.lower()
        or "class" in c.lower()
        or "asd" in c.lower()
    ):
        print(
            f"{c:35s} | "
            f"dtype={df128[c].dtype} | "
            f"unique={df128[c].nunique(dropna=False)}"
        )

# ---------------------------------------------------------------------------
# 11. DUPLICATE / CONSTANT COLUMN CHECK
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("CONSTANT / DUPLICATE NUMERIC COLUMN CHECK")
print("=" * 80)

constant_cols = []
duplicate_cols = []

for c in embedding_candidates:

    if df128[c].nunique(dropna=False) <= 1:
        constant_cols.append(c)

# Compare columns pairwise using values
for i in range(len(embedding_candidates)):
    c1 = embedding_candidates[i]

    for j in range(i + 1, len(embedding_candidates)):
        c2 = embedding_candidates[j]

        if df128[c1].equals(df128[c2]):
            duplicate_cols.append((c1, c2))

print("Constant numeric columns:")
if constant_cols:
    for c in constant_cols:
        print("  ", c)
else:
    print("  NONE")

print("\nDuplicate numeric columns:")
if duplicate_cols:
    for a, b in duplicate_cols:
        print(f"  {a} == {b}")
else:
    print("  NONE")

# ---------------------------------------------------------------------------
# 12. FINAL DIAGNOSIS
# ---------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FORENSIC AUDIT STATUS")
print("=" * 80)

if len(embedding_candidates) == 128:
    print("✅ Behavioral file contains exactly 128 numeric embedding candidates.")
    print("The previous audit likely excluded/handled a column incorrectly.")

elif len(embedding_candidates) == 129:
    print("⚠️ Behavioral file contains 129 numeric candidates.")
    print("One column must be identified before reproducibility can be certified.")
    print("DO NOT retrain the behavioral model yet.")

else:
    print(
        f"❌ Unexpected number of numeric candidates: "
        f"{len(embedding_candidates)}"
    )
    print("Further investigation required.")

print("\n" + "=" * 80)
print("M4.1.2A COMPLETED")
print("=" * 80)
```

```python
# ============================================================================
# ASD MULTIMODAL PROJECT — M4.1.2B
# BEHAVIORAL 128-D FORENSIC COLUMN IDENTIFICATION
# ============================================================================
#
# PURPOSE:
#   Identify the extra numeric column detected in the Behavioral 128-D CSV.
#
# IMPORTANT:
#   This script DOES NOT retrain the model.
#   This script DOES NOT modify any existing artifact.
#   This script ONLY audits the existing Behavioral 128-D representation.
# ============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import re

print("=" * 80)
print("ASD MULTIMODAL PROJECT — STEP M4.1.2B")
print("BEHAVIORAL 128-D FORENSIC COLUMN IDENTIFICATION")
print("=" * 80)

# ----------------------------------------------------------------------------
# 1. PATHS
# ----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/ASD_Multimodal_Project"
)

BEHAVIORAL_ROOT = PROJECT_ROOT / "Behavioral_Model"

EMB128_PATH = (
    BEHAVIORAL_ROOT
    / "03_representation_learning"
    / "behavioral_128d_embeddings.csv"
)

REF32_PATH = (
    BEHAVIORAL_ROOT
    / "04_cross_modal_alignment"
    / "behavioral_32d_aligned_representation.csv"
)

RAW_PATH = (
    BEHAVIORAL_ROOT
    / "01_raw_data"
    / "behavioral_dataset.csv"
)

OUTPUT_DIR = (
    PROJECT_ROOT
    / "05_multimodal_fusion"
    / "M4_1_2B_behavioral_forensics"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ----------------------------------------------------------------------------
# 2. FILE CHECK
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("1. REQUIRED FILE CHECK")
print("=" * 80)

for name, path in [
    ("Behavioral 128-D embeddings", EMB128_PATH),
    ("Behavioral 32-D reference", REF32_PATH),
    ("Behavioral raw dataset", RAW_PATH),
]:
    if path.exists():
        print(f"[FOUND] {name}")
        print(f"        {path}")
        print(f"        Size: {path.stat().st_size / (1024**2):.4f} MB")
    else:
        print(f"[MISSING] {name}")
        print(f"          {path}")

if not EMB128_PATH.exists():
    raise FileNotFoundError(
        f"Behavioral 128-D embeddings not found:\n{EMB128_PATH}"
    )

if not REF32_PATH.exists():
    raise FileNotFoundError(
        f"Behavioral 32-D reference not found:\n{REF32_PATH}"
    )

# ----------------------------------------------------------------------------
# 3. LOAD 128-D FILE
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("2. LOADING BEHAVIORAL 128-D FILE")
print("=" * 80)

df128 = pd.read_csv(EMB128_PATH)

print("Shape:", df128.shape)
print("Number of columns:", len(df128.columns))

print("\nAll columns:")
for i, col in enumerate(df128.columns):
    print(f"{i:03d} | {col}")

# ----------------------------------------------------------------------------
# 4. BASIC METADATA IDENTIFICATION
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("3. METADATA / TARGET IDENTIFICATION")
print("=" * 80)

print("\nData types:")

for col in df128.columns:
    print(
        f"{col:<35} "
        f"dtype={str(df128[col].dtype):<12} "
        f"unique={df128[col].nunique(dropna=False):<6}"
    )

# Known metadata candidates
metadata_candidates = []

for col in df128.columns:

    name = str(col).lower()

    if (
        col == "Sample_Index"
        or col == "Class/ASD"
        or "target" in name
        or "class" in name
        or "label" in name
        or "index" in name
        or "id" in name
        or "prediction" in name
        or "prob" in name
        or "score" in name
    ):
        metadata_candidates.append(col)

print("\nPotential metadata / non-embedding columns:")

for col in metadata_candidates:
    print("  ", col)

# ----------------------------------------------------------------------------
# 5. NUMERIC COLUMN COUNT
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("4. NUMERIC COLUMN AUDIT")
print("=" * 80)

numeric_cols = df128.select_dtypes(include=[np.number]).columns.tolist()

print("Total numeric columns:", len(numeric_cols))

for i, col in enumerate(numeric_cols):
    print(f"{i:03d} | {col}")

# ----------------------------------------------------------------------------
# 6. EXPECTED BEHAVIORAL EMBEDDING NAME PATTERNS
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("5. SEARCHING FOR EMBEDDING COLUMN PATTERNS")
print("=" * 80)

patterns = [
    r"^Behavioral_Embedding_\d+$",
    r"^Behavioral_128D_\d+$",
    r"^Behavioral_32D_\d+$",
    r"^behavioral_embedding_\d+$",
    r"^embedding_\d+$",
    r"^Embedding_\d+$",
]

for pattern in patterns:

    matches = [
        col
        for col in df128.columns
        if re.match(pattern, str(col))
    ]

    print(f"\nPattern: {pattern}")
    print("Matches:", len(matches))

    if matches:
        print("First 10:")
        for col in matches[:10]:
            print("  ", col)

# ----------------------------------------------------------------------------
# 7. CHECK COMMON EXPECTED 128-D NAME SET
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("6. EXPECTED 128-D COLUMN SET CHECK")
print("=" * 80)

expected_sets = {

    "Behavioral_Embedding_001-128": [
        f"Behavioral_Embedding_{i:03d}"
        for i in range(1, 129)
    ],

    "Behavioral_128D_001-128": [
        f"Behavioral_128D_{i:03d}"
        for i in range(1, 129)
    ],

    "Embedding_001-128": [
        f"Embedding_{i:03d}"
        for i in range(1, 129)
    ],
}

for set_name, expected in expected_sets.items():

    present = [c for c in expected if c in df128.columns]
    missing = [c for c in expected if c not in df128.columns]

    print(f"\n{set_name}")
    print("Present:", len(present))
    print("Missing:", len(missing))

    if missing:
        print("Missing examples:", missing[:10])

# ----------------------------------------------------------------------------
# 8. COMPARE AGAINST 32-D REFERENCE METADATA
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("7. BEHAVIORAL 32-D REFERENCE CHECK")
print("=" * 80)

df32 = pd.read_csv(REF32_PATH)

print("32-D shape:", df32.shape)

print("\n32-D columns:")

for i, col in enumerate(df32.columns):
    print(f"{i:03d} | {col}")

embedding32_cols = [
    c for c in df32.columns
    if c not in ["Sample_Index", "Class/ASD"]
]

print("\n32-D embedding count:", len(embedding32_cols))

# ----------------------------------------------------------------------------
# 9. CHECK ROW ALIGNMENT
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("8. SAMPLE INDEX / ROW ALIGNMENT CHECK")
print("=" * 80)

if "Sample_Index" in df128.columns and "Sample_Index" in df32.columns:

    ids128 = df128["Sample_Index"].astype(str)
    ids32 = df32["Sample_Index"].astype(str)

    print("128-D unique Sample_Index:", ids128.nunique())
    print("32-D unique Sample_Index :", ids32.nunique())

    set128 = set(ids128)
    set32 = set(ids32)

    print("IDs in 128-D only:", len(set128 - set32))
    print("IDs in 32-D only :", len(set32 - set128))

    if set128 == set32:
        print("[PASSED] Same Sample_Index population")
    else:
        print("[WARNING] Sample populations differ")

# ----------------------------------------------------------------------------
# 10. CHECK TARGET
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("9. TARGET CHECK")
print("=" * 80)

if "Class/ASD" in df128.columns:

    print("Class/ASD present in 128-D file")

    print(
        df128["Class/ASD"]
        .value_counts(dropna=False)
        .sort_index()
    )

else:
    print("[INFO] Class/ASD not present")

# ----------------------------------------------------------------------------
# 11. DETECT POSSIBLE EXTRA COLUMN
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("10. EXTRA NUMERIC COLUMN DETECTION")
print("=" * 80)

# Remove obvious metadata
obvious_metadata = {
    "Sample_Index",
    "Class/ASD",
    "Target",
    "Label",
    "ID",
    "prediction",
    "probability",
    "score",
}

candidate_embeddings = [
    c for c in numeric_cols
    if c not in obvious_metadata
]

print("Numeric candidates after obvious metadata removal:",
      len(candidate_embeddings))

if len(candidate_embeddings) == 128:

    print("\n[PASSED]")
    print("Exactly 128 numeric embedding candidates remain.")

elif len(candidate_embeddings) == 129:

    print("\n[IMPORTANT]")
    print("Exactly ONE EXTRA numeric candidate detected.")

    print("\nCandidate columns:")

    for i, col in enumerate(candidate_embeddings):
        print(f"{i:03d} | {col}")

else:

    print("\n[WARNING]")
    print(
        f"Expected 128 embedding candidates but found "
        f"{len(candidate_embeddings)}."
    )

# ----------------------------------------------------------------------------
# 12. COLUMN STATISTICS
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("11. COLUMN STATISTICS")
print("=" * 80)

stats_rows = []

for col in numeric_cols:

    series = pd.to_numeric(df128[col], errors="coerce")

    stats_rows.append({
        "column": col,
        "dtype": str(df128[col].dtype),
        "unique": int(series.nunique(dropna=False)),
        "nan": int(series.isna().sum()),
        "mean": float(series.mean()),
        "std": float(series.std()),
        "min": float(series.min()),
        "max": float(series.max()),
    })

stats_df = pd.DataFrame(stats_rows)

print(stats_df.to_string(index=False))

stats_path = OUTPUT_DIR / "behavioral_128d_numeric_column_statistics.csv"

stats_df.to_csv(stats_path, index=False)

print("\n[SAVED]")
print(stats_path)

# ----------------------------------------------------------------------------
# 13. IDENTIFY POSSIBLE METADATA BY LOW CARDINALITY
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("12. LOW-CARDINALITY NUMERIC COLUMN CHECK")
print("=" * 80)

for col in numeric_cols:

    unique_count = df128[col].nunique(dropna=False)

    if unique_count <= 20:

        print(
            f"[LOW CARDINALITY] "
            f"{col} -> {unique_count} unique values"
        )

# ----------------------------------------------------------------------------
# 14. IDENTIFY BINARY COLUMNS
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("13. BINARY NUMERIC COLUMN CHECK")
print("=" * 80)

binary_cols = []

for col in numeric_cols:

    vals = df128[col].dropna().unique()

    if len(vals) <= 2:

        binary_cols.append(col)

        print(
            f"[BINARY] {col} -> values: "
            f"{sorted(vals.tolist())}"
        )

# ----------------------------------------------------------------------------
# 15. CHECK FOR PREDICTION-LIKE COLUMNS
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("14. PREDICTION / PROBABILITY-LIKE COLUMN CHECK")
print("=" * 80)

prediction_like = []

for col in numeric_cols:

    name = str(col).lower()

    if any(
        token in name
        for token in [
            "prediction",
            "probability",
            "prob",
            "pred",
            "score",
            "output",
            "logit",
        ]
    ):

        prediction_like.append(col)

        print(
            f"[POSSIBLE OUTPUT COLUMN] {col}"
        )

# ----------------------------------------------------------------------------
# 16. FINAL FORENSIC DECISION
# ----------------------------------------------------------------------------

print("\n" + "=" * 80)
print("15. FORENSIC DECISION")
print("=" * 80)

decision = ""

if len(candidate_embeddings) == 128:

    decision = (
        "EXACTLY 128 EMBEDDING CANDIDATES FOUND. "
        "The previous 129-column detection was caused by "
        "a metadata/output column that can now be excluded."
    )

elif len(candidate_embeddings) == 129:

    decision = (
        "ONE EXTRA NUMERIC COLUMN REMAINS. "
        "Behavioral 128-D reproducibility is NOT YET CERTIFIED. "
        "The extra column must be identified."
    )

else:

    decision = (
        f"UNEXPECTED EMBEDDING COUNT: {len(candidate_embeddings)}. "
        "Further investigation required."
    )

print(decision)

# ----------------------------------------------------------------------------
# 17. SAVE FORENSIC REPORT
# ----------------------------------------------------------------------------

report_path = OUTPUT_DIR / "M4_1_2B_behavioral_forensic_report.txt"

with open(report_path, "w") as f:

    f.write(
        "ASD MULTIMODAL PROJECT — M4.1.2B\n"
        "BEHAVIORAL 128-D FORENSIC COLUMN IDENTIFICATION\n"
        "=" * 80 + "\n\n"
    )

    f.write(f"128-D file: {EMB128_PATH}\n")
    f.write(f"Shape: {df128.shape}\n\n")

    f.write(
        f"Total numeric columns: {len(numeric_cols)}\n"
    )

    f.write(
        f"Candidate embedding columns after metadata removal: "
        f"{len(candidate_embeddings)}\n\n"
    )

    f.write("Potential metadata columns:\n")

    for col in metadata_candidates:
        f.write(f"  {col}\n")

    f.write("\nCandidate embedding columns:\n")

    for i, col in enumerate(candidate_embeddings):
        f.write(f"  {i+1:03d} | {col}\n")

    f.write("\nBinary numeric columns:\n")

    for col in binary_cols:
        f.write(f"  {col}\n")

    f.write("\nPrediction-like columns:\n")

    for col in prediction_like:
        f.write(f"  {col}\n")

    f.write("\nFINAL DECISION:\n")
    f.write(decision + "\n")

print("\n[SAVED]")
print(report_path)

print("\n" + "=" * 80)
print("M4.1.2B FORENSIC AUDIT COMPLETED")
print("=" * 80)

print("\nIMPORTANT:")
print("Do NOT retrain Behavioral.")
print("Do NOT modify the existing 128-D file.")
print("Do NOT modify the existing 32-D representation.")
print("Do NOT continue to M4.1.3 until the extra column is understood.")
```

```python
# =============================================================================
# ASD MULTIMODAL PROJECT — STEP M4.1.3
# EXACT TRAINING-PIPELINE RECONSTRUCTION AUDIT
# =============================================================================

from pathlib import Path
import os
import json

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/ASD_Multimodal_Project"
)

print("=" * 90)
print("ASD MULTIMODAL PROJECT — STEP M4.1.3")
print("EXACT TRAINING-PIPELINE RECONSTRUCTION AUDIT")
print("=" * 90)

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(PROJECT_ROOT)

print("\nProject root:")
print(PROJECT_ROOT)

# =============================================================================
# 1. SEARCH FOR ALL SOURCE CODE
# =============================================================================

print("\n" + "=" * 90)
print("1. SEARCHING FOR TRAINING / PREPROCESSING SOURCE CODE")
print("=" * 90)

SOURCE_EXTENSIONS = {
    ".py",
    ".ipynb",
    ".json",
    ".yaml",
    ".yml"
}

source_files = []

for path in PROJECT_ROOT.rglob("*"):
    if path.is_file() and path.suffix.lower() in SOURCE_EXTENSIONS:
        source_files.append(path)

print(f"\nSource/config files found: {len(source_files)}")

for path in sorted(source_files):
    print(f"[FOUND] {path}")

# =============================================================================
# 2. SEARCH FOR MODALITY-SPECIFIC SOURCE FILES
# =============================================================================

keywords = {
    "GENETIC": [
        "genetic",
        "gene",
        "transformer",
        "embedding",
        "alignment"
    ],
    "BEHAVIORAL": [
        "behavioral",
        "behavioural",
        "behavior",
        "encoder",
        "preprocess",
        "alignment"
    ],
    "ENVIRONMENTAL": [
        "environmental",
        "environment",
        "encoder",
        "preprocess",
        "alignment"
    ]
}

print("\n" + "=" * 90)
print("2. MODALITY-SPECIFIC SOURCE DISCOVERY")
print("=" * 90)

for modality, words in keywords.items():

    print("\n" + "-" * 80)
    print(modality)
    print("-" * 80)

    found = []

    for path in source_files:

        name = path.name.lower()

        if any(word in name for word in words):
            found.append(path)

    if found:
        for path in sorted(found):
            print("[FOUND]", path)
    else:
        print("[NONE FOUND]")

# =============================================================================
# 3. COMPLETE MODALITY ARTIFACT MAP
# =============================================================================

print("\n" + "=" * 90)
print("3. COMPLETE ARTIFACT MAP")
print("=" * 90)

artifact_patterns = {

    "GENETIC": [
        "Genetic_dataset.csv",
        "genetic_128d_embeddings.csv",
        "genetic_transformer_128d.pt",
        "genetic_128_to_32_alignment.pt",
        "genetic_32d_aligned_representation.csv",
        "genetic_32d_validation_report.txt",
    ],

    "BEHAVIORAL": [
        "behavioral_dataset.csv",
        "behavioral_preprocessor.joblib",
        "behavioral_encoder_128d.pt",
        "behavioral_128_to_32_alignment.pt",
        "behavioral_32d_aligned_representation.csv",
        "behavioral_32d_validation_report.txt",
    ],

    "ENVIRONMENTAL": [
        "environmental_dataset.csv",
        "environmental_preprocessor.joblib",
        "environmental_encoder_128d.pt",
        "environmental_128_to_32_alignment.pt",
        "environmental_32d_aligned_representation.csv",
        "environmental_32d_validation_report.txt",
    ],

    "FUSION": [
        "M3_1_genetic_expert.pt",
        "M3_1_behavioral_expert.pt",
        "M3_1_environmental_expert.pt",
        "M3_2_final_inference_config.json",
        "M4_production_inference_spec.json",
    ]
}

for modality, patterns in artifact_patterns.items():

    print("\n" + "-" * 80)
    print(modality)
    print("-" * 80)

    for pattern in patterns:

        matches = list(PROJECT_ROOT.rglob(pattern))

        if matches:
            for match in matches:
                print(f"[FOUND] {match}")
        else:
            print(f"[MISSING] {pattern}")

# =============================================================================
# 4. SEARCH ALL FILE CONTENT FOR PIPELINE TERMS
# =============================================================================

print("\n" + "=" * 90)
print("4. SEARCHING SOURCE FILES FOR PIPELINE OPERATIONS")
print("=" * 90)

search_terms = [
    "StandardScaler",
    "MinMaxScaler",
    "RobustScaler",
    "SimpleImputer",
    "KNNImputer",
    "OneHotEncoder",
    "LabelEncoder",
    "ColumnTransformer",
    "TransformerEncoder",
    "TransformerEncoderLayer",
    "torch.save",
    "joblib.dump",
    "128",
    "32",
    "alignment",
    "projection",
    "preprocessor",
    "embedding",
    "train_test_split",
    "StratifiedShuffleSplit"
]

for path in sorted(source_files):

    # Skip huge files
    try:
        if path.stat().st_size > 20 * 1024 * 1024:
            continue

        if path.suffix.lower() == ".ipynb":

            with open(path, "r", encoding="utf-8") as f:
                text = f.read()

        else:

            with open(
                path,
                "r",
                encoding="utf-8",
                errors="ignore"
            ) as f:
                text = f.read()

    except Exception:
        continue

    matched_terms = []

    lower_text = text.lower()

    for term in search_terms:

        if term.lower() in lower_text:
            matched_terms.append(term)

    if matched_terms:

        print("\n[PIPELINE SOURCE]")
        print(path)

        print("Matched terms:")
        print("  " + ", ".join(matched_terms))

# =============================================================================
# 5. CHECK GENETIC REPRODUCIBILITY GAP
# =============================================================================

print("\n" + "=" * 90)
print("5. GENETIC REPRODUCIBILITY GAP")
print("=" * 90)

genetic_required = [
    "Genetic_dataset.csv",
    "genetic_transformer_128d.pt",
    "genetic_128_to_32_alignment.pt",
    "genetic_128d_embeddings.csv",
    "genetic_32d_aligned_representation.csv"
]

for filename in genetic_required:

    matches = list(PROJECT_ROOT.rglob(filename))

    if matches:
        print(f"[FOUND] {filename}")
        print(f"        {matches[0]}")
    else:
        print(f"[MISSING] {filename}")

print("""
IMPORTANT GENETIC QUESTION:

The genetic Transformer checkpoint exists, but a checkpoint alone
does not tell us exactly how arbitrary raw genetic input was
preprocessed before entering the Transformer.

Therefore we must locate the original genetic preprocessing /
training implementation before declaring production inference
reproducible.
""")

# =============================================================================
# 6. CHECK BEHAVIORAL REPRODUCIBILITY
# =============================================================================

print("\n" + "=" * 90)
print("6. BEHAVIORAL REPRODUCIBILITY STATUS")
print("=" * 90)

behavioral_required = [
    "behavioral_dataset.csv",
    "behavioral_preprocessor.joblib",
    "behavioral_encoder_128d.pt",
    "behavioral_128_to_32_alignment.pt",
    "behavioral_32d_aligned_representation.csv"
]

for filename in behavioral_required:

    matches = list(PROJECT_ROOT.rglob(filename))

    if matches:
        print(f"[FOUND] {filename}")
        print(f"        {matches[0]}")
    else:
        print(f"[MISSING] {filename}")

# =============================================================================
# 7. CHECK ENVIRONMENTAL REPRODUCIBILITY
# =============================================================================

print("\n" + "=" * 90)
print("7. ENVIRONMENTAL REPRODUCIBILITY STATUS")
print("=" * 90)

environmental_required = [
    "environmental_dataset.csv",
    "environmental_preprocessor.joblib",
    "environmental_encoder_128d.pt",
    "environmental_128_to_32_alignment.pt",
    "environmental_32d_aligned_representation.csv"
]

for filename in environmental_required:

    matches = list(PROJECT_ROOT.rglob(filename))

    if matches:
        print(f"[FOUND] {filename}")
        print(f"        {matches[0]}")
    else:
        print(f"[MISSING] {filename}")

# =============================================================================
# 8. FINAL DIAGNOSIS
# =============================================================================

print("\n" + "=" * 90)
print("8. M4.1.3 FINAL DIAGNOSIS")
print("=" * 90)

print("""
The multimodal fusion architecture is already complete.

The remaining production requirement is NOT another fusion model.

The remaining requirement is:

RAW PATIENT DATA
      ↓
EXACT ORIGINAL PREPROCESSING
      ↓
EXACT ORIGINAL 128-D REPRESENTATION MODEL
      ↓
EXACT ORIGINAL 128→32 ALIGNMENT
      ↓
32-D KNOWLEDGE REPRESENTATION
      ↓
THREE TRAINED EXPERTS
      ↓
WEIGHTED EVIDENCE FUSION
      ↓
ONE ASD / NON-ASD OUTPUT

M4.1.3 therefore determines whether this complete path can be
reproduced exactly from the artifacts currently stored in Drive.
""")

print("=" * 90)
print("M4.1.3 AUDIT COMPLETED")
print("=" * 90)
```

```python
# ==========================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.2
# EXACT BEHAVIORAL ENCODER RECONSTRUCTION + FORWARD VALIDATION
# ==========================================================================================

import os
import torch
import torch.nn as nn

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.2")
print("EXACT BEHAVIORAL ENCODER RECONSTRUCTION + FORWARD VALIDATION")
print("=" * 100)

# ==========================================================================================
# 1. CONFIGURATION
# ==========================================================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

PROJECT_ROOT = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

BEHAVIORAL_ENCODER_PATH = os.path.join(
    PROJECT_ROOT,
    "Behavioral_Model",
    "03_representation_learning",
    "behavioral_encoder_128d.pt"
)

print("\nDevice :", DEVICE)
print("PyTorch:", torch.__version__)
print("Project:", PROJECT_ROOT)

if not os.path.exists(PROJECT_ROOT):
    raise FileNotFoundError("Project directory not found.")

if not os.path.exists(BEHAVIORAL_ENCODER_PATH):
    raise FileNotFoundError(
        f"Behavioral encoder checkpoint not found:\n"
        f"{BEHAVIORAL_ENCODER_PATH}"
    )

print("[PASSED] Project directory available.")
print("[PASSED] Behavioral encoder checkpoint available.")


# ==========================================================================================
# 2. SAFE CHECKPOINT LOADING
# ==========================================================================================

print("\n" + "=" * 100)
print("2. LOADING BEHAVIORAL CHECKPOINT")
print("-" * 100)

try:

    behavioral_checkpoint = torch.load(
        BEHAVIORAL_ENCODER_PATH,
        map_location="cpu",
        weights_only=True
    )

except Exception as e:

    print("[INFO] weights_only=True failed.")
    print("       Retrying trusted checkpoint with weights_only=False.")
    print("       Reason:", type(e).__name__)

    behavioral_checkpoint = torch.load(
        BEHAVIORAL_ENCODER_PATH,
        map_location="cpu",
        weights_only=False
    )

print("[PASSED] Behavioral checkpoint loaded.")


# ==========================================================================================
# 3. EXTRACT MODEL STATE
# ==========================================================================================

if "model_state_dict" not in behavioral_checkpoint:
    raise RuntimeError(
        "Expected 'model_state_dict' was not found in behavioral checkpoint."
    )

behavioral_state = behavioral_checkpoint["model_state_dict"]

saved_input_dim = behavioral_checkpoint.get("input_dim", None)
saved_embedding_dim = behavioral_checkpoint.get("embedding_dim", None)
saved_target_column = behavioral_checkpoint.get("target_column", None)

print("\nSaved input dimension    :", saved_input_dim)
print("Saved embedding dimension:", saved_embedding_dim)
print("Saved target column      :", saved_target_column)

if saved_input_dim != 87:
    raise RuntimeError(
        f"Unexpected saved input dimension: {saved_input_dim}. "
        f"Expected 87."
    )

if saved_embedding_dim != 128:
    raise RuntimeError(
        f"Unexpected saved embedding dimension: {saved_embedding_dim}. "
        f"Expected 128."
    )

print("[PASSED] Saved metadata matches expected architecture.")


# ==========================================================================================
# 4. EXACT ARCHITECTURE
# ==========================================================================================

print("\n" + "=" * 100)
print("4. RECONSTRUCTING EXACT BEHAVIORAL ENCODER")
print("-" * 100)

class BehavioralEncoderExact(nn.Module):

    def __init__(self):
        super().__init__()

        # Exact structure inferred from checkpoint keys:
        #
        # encoder.0 -> Linear(87, 128)
        # encoder.1 -> BatchNorm1d(128)
        # encoder.2 -> ReLU
        # encoder.3 -> Dropout / non-parameter layer
        # encoder.4 -> Linear(128, 128)
        # encoder.5 -> BatchNorm1d(128)
        # encoder.6 -> ReLU
        # classifier -> Linear(128, 1)

        self.encoder = nn.Sequential(
            nn.Linear(87, 128),       # encoder.0
            nn.BatchNorm1d(128),      # encoder.1
            nn.ReLU(),               # encoder.2
            nn.Dropout(),            # encoder.3
            nn.Linear(128, 128),      # encoder.4
            nn.BatchNorm1d(128),      # encoder.5
            nn.ReLU()                # encoder.6
        )

        self.classifier = nn.Linear(128, 1)

    def forward(self, x):

        embedding = self.encoder(x)
        logit = self.classifier(embedding)

        return embedding, logit


behavioral_encoder = BehavioralEncoderExact().to(DEVICE)


# ==========================================================================================
# 5. STRICT CHECKPOINT LOADING
# ==========================================================================================

print("\nLoading checkpoint with strict=True...")

try:

    behavioral_encoder.load_state_dict(
        behavioral_state,
        strict=True
    )

except RuntimeError as e:

    print("\n[FAILED] Exact architecture did not match checkpoint.")
    print("\nCheckpoint keys:")
    for key in behavioral_state.keys():
        print(" ", key)

    raise e

print("[PASSED] Behavioral encoder reconstructed with strict=True.")


# ==========================================================================================
# 6. EVALUATION MODE
# ==========================================================================================

behavioral_encoder.eval()

print("[PASSED] Behavioral encoder set to eval().")


# ==========================================================================================
# 7. CHECK MODEL PARAMETERS
# ==========================================================================================

print("\n" + "=" * 100)
print("7. PARAMETER SHAPE VALIDATION")
print("-" * 100)

expected_shapes = {

    "encoder.0.weight": (128, 87),
    "encoder.0.bias": (128,),

    "encoder.1.weight": (128,),
    "encoder.1.bias": (128,),
    "encoder.1.running_mean": (128,),
    "encoder.1.running_var": (128,),

    "encoder.4.weight": (128, 128),
    "encoder.4.bias": (128,),

    "encoder.5.weight": (128,),
    "encoder.5.bias": (128,),
    "encoder.5.running_mean": (128,),
    "encoder.5.running_var": (128,),

    "classifier.weight": (1, 128),
    "classifier.bias": (1,)
}

model_state = behavioral_encoder.state_dict()

for key, expected_shape in expected_shapes.items():

    if key not in model_state:
        raise RuntimeError(
            f"Missing reconstructed model key: {key}"
        )

    actual_shape = tuple(model_state[key].shape)

    if actual_shape != expected_shape:
        raise RuntimeError(
            f"Shape mismatch for {key}: "
            f"actual={actual_shape}, expected={expected_shape}"
        )

    print(
        f"[PASSED] {key:<35} "
        f"{actual_shape}"
    )


# ==========================================================================================
# 8. FORWARD-PASS TEST
# ==========================================================================================

print("\n" + "=" * 100)
print("8. FORWARD-PASS VALIDATION")
print("-" * 100)

# One sample with the exact saved input dimension.
#
# Batch size = 2 is deliberately used because BatchNorm requires
# more than one sample when a model is in training mode.
#
# The model is already in eval(), so batch size 1 would also work,
# but batch size 2 gives an additional robust sanity check.

test_input = torch.zeros(
    (2, 87),
    dtype=torch.float32,
    device=DEVICE
)

with torch.no_grad():

    test_embedding, test_logit = behavioral_encoder(
        test_input
    )

print(
    "Input shape       :",
    tuple(test_input.shape)
)

print(
    "Embedding shape   :",
    tuple(test_embedding.shape)
)

print(
    "Classifier output :",
    tuple(test_logit.shape)
)


# ==========================================================================================
# 9. OUTPUT VALIDATION
# ==========================================================================================

if tuple(test_embedding.shape) != (2, 128):

    raise RuntimeError(
        f"Unexpected embedding shape: "
        f"{tuple(test_embedding.shape)}"
    )

if tuple(test_logit.shape) != (2, 1):

    raise RuntimeError(
        f"Unexpected classifier shape: "
        f"{tuple(test_logit.shape)}"
    )

print("[PASSED] Input -> 87-D accepted.")
print("[PASSED] Encoder -> exact 128-D embedding produced.")
print("[PASSED] Classifier -> one output logit produced.")


# ==========================================================================================
# 10. PROBABILITY VALIDATION
# ==========================================================================================

print("\n" + "=" * 100)
print("10. CLASSIFIER PROBABILITY VALIDATION")
print("-" * 100)

test_probability = torch.sigmoid(
    test_logit
)

print(
    "Probability shape:",
    tuple(test_probability.shape)
)

print(
    "Probability values:",
    test_probability.detach().cpu().numpy().reshape(-1)
)

if torch.any(test_probability < 0) or torch.any(test_probability > 1):

    raise RuntimeError(
        "Invalid probability produced by sigmoid."
    )

print("[PASSED] Classifier logits converted to valid probabilities.")


# ==========================================================================================
# 11. EMBEDDING FINITE-VALUE CHECK
# ==========================================================================================

print("\n" + "=" * 100)
print("11. NUMERICAL SANITY CHECK")
print("-" * 100)

if not torch.isfinite(test_embedding).all():

    raise RuntimeError(
        "Behavioral embedding contains NaN or Inf."
    )

if not torch.isfinite(test_logit).all():

    raise RuntimeError(
        "Behavioral classifier output contains NaN or Inf."
    )

print("[PASSED] 128-D embedding contains only finite values.")
print("[PASSED] Classifier output contains only finite values.")


# ==========================================================================================
# 12. FINAL STATUS
# ==========================================================================================

print("\n" + "=" * 100)
print("M4.4.9.2 COMPLETED")
print("=" * 100)

print("""
[PASSED] Exact behavioral encoder architecture reconstructed.
[PASSED] Checkpoint loaded with strict=True.
[PASSED] Behavioral encoder is in eval() mode.
[PASSED] 87-D input accepted.
[PASSED] Exact 128-D behavioral representation produced.
[PASSED] Classifier produces one output logit.
[PASSED] Probability conversion validated.
[PASSED] Numerical sanity checks passed.

NO TRAINING PERFORMED.
NO CHECKPOINT MODIFIED.
NO PRODUCTION ARTIFACT OVERWRITTEN.

NEXT:
M4.4.9.3 — connect the EXACT behavioral preprocessing pipeline
to this reconstructed encoder and validate the real 5-patient
behavioral representations.

After that, we move directly toward the final P001–P005
end-to-end multimodal production inference.
""")
```

```python
# ================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3 FINAL
# ROBUST ARTIFACT DISCOVERY + EXACT BEHAVIORAL PRODUCTION INFERENCE
# ================================================================================================

import os
import glob
import json
import warnings
import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

warnings.filterwarnings("ignore")

# ------------------------------------------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------------------------------------------

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3 FINAL")
print("ROBUST ARTIFACT DISCOVERY + EXACT BEHAVIORAL PRODUCTION INFERENCE")
print("=" * 100)

print(f"Device  : {DEVICE}")
print(f"PyTorch : {torch.__version__}")
print(f"Project : {PROJECT_DIR}")

# ------------------------------------------------------------------------------------------------
# 1. PROJECT CHECK
# ------------------------------------------------------------------------------------------------

if not os.path.isdir(PROJECT_DIR):
    raise FileNotFoundError(f"Project directory not found:\n{PROJECT_DIR}")

print("[PASSED] Project directory available.")

# ------------------------------------------------------------------------------------------------
# 2. ROBUST RECURSIVE ARTIFACT DISCOVERY
# ------------------------------------------------------------------------------------------------

def find_file_recursive(root, filename):
    """
    Recursively search for an exact filename.
    Handles directories containing accidental leading/trailing spaces.
    """
    matches = []

    for current_root, dirs, files in os.walk(root):
        for f in files:
            if f == filename:
                matches.append(os.path.join(current_root, f))

    if not matches:
        # Secondary tolerant search
        target = filename.strip().lower()

        for current_root, dirs, files in os.walk(root):
            for f in files:
                if f.strip().lower() == target:
                    matches.append(os.path.join(current_root, f))

    if not matches:
        return None

    # Prefer the shortest path if duplicates exist
    matches.sort(key=len)

    return matches[0]


def require_artifact(label, filename):
    path = find_file_recursive(PROJECT_DIR, filename)

    if path is None:
        raise FileNotFoundError(
            f"\n[FAILED] Could not locate artifact:\n"
            f"Label    : {label}\n"
            f"Filename : {filename}\n"
            f"Project  : {PROJECT_DIR}\n"
        )

    size_mb = os.path.getsize(path) / (1024 * 1024)

    print(f"[FOUND] {label:<28} {size_mb:.4f} MB")
    print(f"        {path}")

    return path


print("\n" + "=" * 100)
print("1. ROBUST BEHAVIORAL ARTIFACT DISCOVERY")
print("=" * 100)

PATHS = {}

PATHS["raw"] = require_artifact(
    "behavioral_raw",
    "behavioral_dataset.csv"
)

PATHS["preprocessor"] = require_artifact(
    "behavioral_preprocessor",
    "behavioral_preprocessor.joblib"
)

PATHS["encoder"] = require_artifact(
    "behavioral_encoder",
    "behavioral_encoder_128d.pt"
)

PATHS["alignment"] = require_artifact(
    "behavioral_alignment",
    "behavioral_128_to_32_alignment.pt"
)

PATHS["scaler"] = require_artifact(
    "behavioral_scaler",
    "M3_1_behavioral_scaler.joblib"
)

PATHS["expert"] = require_artifact(
    "behavioral_expert",
    "M3_1_behavioral_expert.pt"
)

print("\n[PASSED] All behavioral production artifacts located.")

# ------------------------------------------------------------------------------------------------
# 3. DEMO DIRECTORY DISCOVERY
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("2. DEMO DATA DISCOVERY")
print("=" * 100)

demo_candidates = []

for current_root, dirs, files in os.walk(PROJECT_DIR):

    dirname = os.path.basename(current_root).strip().lower()

    if dirname == "demo":
        demo_candidates.append(current_root)

    # Also detect patient CSVs anywhere under the project
    if any(f.lower().endswith(".csv") and "p001" in f.lower() for f in files):
        demo_candidates.append(current_root)

if demo_candidates:
    DEMO_DIR = sorted(set(demo_candidates), key=len)[0]
    print(f"[FOUND] Demo directory:")
    print(f"        {DEMO_DIR}")
else:
    DEMO_DIR = None
    print("[INFO] Dedicated demo directory not found.")
    print("[INFO] Patient files will be searched recursively.")

# ------------------------------------------------------------------------------------------------
# 4. LOAD CHECKPOINTS
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("3. LOADING BEHAVIORAL CHECKPOINTS")
print("=" * 100)

def load_checkpoint(path):
    """
    First try safe weights-only loading.
    If the trusted project checkpoint contains metadata requiring full loading,
    retry with weights_only=False.
    """
    try:
        return torch.load(
            path,
            map_location=DEVICE,
            weights_only=True
        )
    except Exception as e:
        print(f"[INFO] weights_only=True failed for:")
        print(f"       {os.path.basename(path)}")
        print("[INFO] Retrying trusted checkpoint with weights_only=False.")

        return torch.load(
            path,
            map_location=DEVICE,
            weights_only=False
        )


encoder_checkpoint = load_checkpoint(PATHS["encoder"])
alignment_checkpoint = load_checkpoint(PATHS["alignment"])
expert_checkpoint = load_checkpoint(PATHS["expert"])

print("[PASSED] Behavioral encoder checkpoint loaded.")
print("[PASSED] Behavioral alignment checkpoint loaded.")
print("[PASSED] Behavioral expert checkpoint loaded.")

# ------------------------------------------------------------------------------------------------
# 5. LOAD PREPROCESSOR + SCALER
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("4. LOADING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)

behavioral_preprocessor = joblib.load(PATHS["preprocessor"])
behavioral_scaler = joblib.load(PATHS["scaler"])

print(f"Preprocessor: {type(behavioral_preprocessor)}")
print(f"Scaler      : {type(behavioral_scaler)}")

print(
    f"Preprocessor input features: "
    f"{getattr(behavioral_preprocessor, 'n_features_in_', 'unknown')}"
)

print(
    f"Scaler input features: "
    f"{getattr(behavioral_scaler, 'n_features_in_', 'unknown')}"
)

print("[PASSED] Behavioral preprocessing artifacts loaded.")

# ------------------------------------------------------------------------------------------------
# 6. RECONSTRUCT EXACT BEHAVIORAL ENCODER
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("5. RECONSTRUCTING EXACT BEHAVIORAL ENCODER")
print("=" * 100)

behavioral_state = encoder_checkpoint["model_state_dict"]

saved_input_dim = int(
    encoder_checkpoint.get("input_dim", 87)
)

saved_embedding_dim = int(
    encoder_checkpoint.get("embedding_dim", 128)
)

saved_target_column = encoder_checkpoint.get(
    "target_column",
    "Class/ASD"
)

print(f"Saved input dimension     : {saved_input_dim}")
print(f"Saved embedding dimension : {saved_embedding_dim}")
print(f"Saved target column       : {saved_target_column}")

class BehavioralEncoder(nn.Module):

    def __init__(self, input_dim=87, embedding_dim=128):

        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),

            nn.Dropout(0.2),

            nn.Linear(128, embedding_dim),
            nn.BatchNorm1d(embedding_dim)
        )

        self.classifier = nn.Linear(
            embedding_dim,
            1
        )

    def forward(self, x):

        embedding = self.encoder(x)

        logit = self.classifier(embedding)

        return embedding, logit


behavioral_encoder = BehavioralEncoder(
    input_dim=saved_input_dim,
    embedding_dim=saved_embedding_dim
).to(DEVICE)

behavioral_encoder.load_state_dict(
    behavioral_state,
    strict=True
)

behavioral_encoder.eval()

print("[PASSED] Behavioral encoder reconstructed with strict=True.")
print("[PASSED] Behavioral encoder set to eval().")

# ------------------------------------------------------------------------------------------------
# 7. RECONSTRUCT EXACT BEHAVIORAL ALIGNMENT
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("6. RECONSTRUCTING EXACT BEHAVIORAL ALIGNMENT")
print("=" * 100)

# Determine whether checkpoint contains encoder/decoder.
if "state_dict" in alignment_checkpoint:
    alignment_state = alignment_checkpoint["state_dict"]

elif "model_state_dict" in alignment_checkpoint:
    alignment_state = alignment_checkpoint["model_state_dict"]

elif "encoder" in alignment_checkpoint:
    alignment_state = alignment_checkpoint["encoder"]

else:
    alignment_state = alignment_checkpoint


class BehavioralAlignment(nn.Module):

    def __init__(self):

        super().__init__()

        # IMPORTANT:
        # The production checkpoint contains both encoder and decoder.
        # The inference path uses ONLY the encoder portion.

        self.encoder = nn.Sequential(
            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )

        self.decoder = nn.Sequential(
            nn.Linear(32, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 128)
        )

    def forward(self, x):

        return self.encoder(x)


behavioral_alignment = BehavioralAlignment().to(DEVICE)

behavioral_alignment.load_state_dict(
    alignment_state,
    strict=True
)

behavioral_alignment.eval()

print("[PASSED] Behavioral alignment reconstructed with strict=True.")
print("[PASSED] Behavioral alignment set to eval().")
print("Architecture: 128 -> 64 -> BatchNorm -> ReLU -> 32")

# ------------------------------------------------------------------------------------------------
# 8. EXACT EXPERT ARCHITECTURE RECOVERY
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("7. RECOVERING EXACT BEHAVIORAL EXPERT ARCHITECTURE")
print("=" * 100)

# Extract expert state dictionary.
if isinstance(expert_checkpoint, dict):

    if "model_state_dict" in expert_checkpoint:
        expert_state = expert_checkpoint["model_state_dict"]

    elif "state_dict" in expert_checkpoint:
        expert_state = expert_checkpoint["state_dict"]

    else:
        expert_state = expert_checkpoint

else:
    expert_state = expert_checkpoint.state_dict()


print("Expert checkpoint keys:")

for key, value in expert_state.items():

    if torch.is_tensor(value):
        print(
            f"  {key:<45} "
            f"{tuple(value.shape)}"
        )


# -----------------------------------------------------------------------------------------------
# The earlier errors showed that the expert contains:
#
# network.4.weight = (32, 64)
# network.4.bias   = (32,)
# network.7.weight = ...
# network.7.bias   = ...
#
# Therefore the expert is NOT a simple 64 -> 1 network.
#
# We reconstruct the architecture directly from the checkpoint.
# -----------------------------------------------------------------------------------------------

class BehavioralExpert(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            # network.0
            nn.Linear(32, 64),

            # network.1
            nn.BatchNorm1d(64),

            # network.2
            nn.ReLU(),

            # network.3
            nn.Dropout(0.2),

            # network.4
            nn.Linear(64, 32),

            # network.5
            nn.ReLU(),

            # network.6
            nn.Dropout(0.2),

            # network.7
            nn.Linear(32, 1)
        )

    def forward(self, x):

        return self.network(x)


behavioral_expert = BehavioralExpert().to(DEVICE)

behavioral_expert.load_state_dict(
    expert_state,
    strict=True
)

behavioral_expert.eval()

print("[PASSED] Behavioral expert reconstructed with strict=True.")
print("[PASSED] Behavioral expert set to eval().")
print("Architecture: 32 -> 64 -> BN -> ReLU -> Dropout -> 32 -> ReLU -> Dropout -> 1")

# ------------------------------------------------------------------------------------------------
# 9. FORWARD-PASS PIPELINE VALIDATION
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("8. FULL REPRESENTATION FORWARD-PASS VALIDATION")
print("=" * 100)

test_input = torch.zeros(
    (2, saved_input_dim),
    dtype=torch.float32,
    device=DEVICE
)

with torch.no_grad():

    test_embedding, test_logit = behavioral_encoder(
        test_input
    )

    test_aligned = behavioral_alignment(
        test_embedding
    )

    test_expert_logit = behavioral_expert(
        test_aligned
    )

print(f"Encoder input shape : {tuple(test_input.shape)}")
print(f"Embedding shape     : {tuple(test_embedding.shape)}")
print(f"Aligned shape       : {tuple(test_aligned.shape)}")
print(f"Expert output shape : {tuple(test_expert_logit.shape)}")

assert test_embedding.shape == (2, 128)
assert test_aligned.shape == (2, 32)
assert test_expert_logit.shape == (2, 1)

print("[PASSED] 87-D -> 128-D -> 32-D -> 1 pipeline validated.")

# ------------------------------------------------------------------------------------------------
# 10. PATIENT FILE DISCOVERY
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("9. SEARCHING FOR DEMO PATIENT INPUTS")
print("=" * 100)

patient_files = {}

for patient_id in ["P001", "P002", "P003", "P004", "P005"]:

    candidates = []

    for current_root, dirs, files in os.walk(PROJECT_DIR):

        for f in files:

            if not f.lower().endswith(".csv"):
                continue

            full_path = os.path.join(current_root, f)

            try:
                df_temp = pd.read_csv(full_path)

                # Patient ID can be represented in several ways.
                if "Patient_ID" in df_temp.columns:

                    if patient_id in df_temp["Patient_ID"].astype(str).values:
                        candidates.append(full_path)

                elif "patient_id" in df_temp.columns:

                    if patient_id in df_temp["patient_id"].astype(str).values:
                        candidates.append(full_path)

            except Exception:
                continue

    if candidates:
        # Prefer paths containing demo
        candidates.sort(
            key=lambda x: (
                "demo" not in x.lower(),
                len(x)
            )
        )

        patient_files[patient_id] = candidates[0]

        print(
            f"[FOUND] {patient_id}: "
            f"{candidates[0]}"
        )

    else:

        print(
            f"[WARNING] No dedicated patient CSV found for {patient_id}"
        )


# ------------------------------------------------------------------------------------------------
# 11. LOAD BEHAVIORAL RAW DATASET
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("10. LOADING BEHAVIORAL RAW DATASET")
print("=" * 100)

behavioral_raw = pd.read_csv(
    PATHS["raw"]
)

print(f"Raw dataset shape: {behavioral_raw.shape}")

print(
    f"Columns ({len(behavioral_raw.columns)}):"
)

print(
    list(behavioral_raw.columns)
)

# ------------------------------------------------------------------------------------------------
# 12. IDENTIFY PATIENT ID COLUMN
# ------------------------------------------------------------------------------------------------

possible_id_columns = [
    "Patient_ID",
    "patient_id",
    "PatientID",
    "patientID",
    "ID",
    "id"
]

patient_id_column = None

for col in possible_id_columns:

    if col in behavioral_raw.columns:

        patient_id_column = col
        break


if patient_id_column is None:

    raise RuntimeError(
        "Could not identify patient ID column in behavioral dataset."
    )

print(
    f"[FOUND] Patient ID column: {patient_id_column}"
)

# ------------------------------------------------------------------------------------------------
# 13. IDENTIFY PREPROCESSOR FEATURE COLUMNS
# ------------------------------------------------------------------------------------------------

if hasattr(
    behavioral_preprocessor,
    "feature_names_in_"
):

    preprocessor_columns = list(
        behavioral_preprocessor.feature_names_in_
    )

else:

    preprocessor_columns = []

    for name, transformer, columns in behavioral_preprocessor.transformers_:

        if columns == "drop":
            continue

        if isinstance(columns, (list, tuple, np.ndarray)):
            preprocessor_columns.extend(list(columns))


print("\nPreprocessor feature columns:")

for col in preprocessor_columns:
    print(f"  {col}")

print(
    f"\nNumber of raw features expected: "
    f"{len(preprocessor_columns)}"
)

# ------------------------------------------------------------------------------------------------
# 14. IDENTIFY DEMO PATIENTS
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("11. DEMO PATIENT DISCOVERY")
print("=" * 100)

demo_patients = {}

for patient_id in [
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
]:

    matches = behavioral_raw[
        behavioral_raw[patient_id_column].astype(str)
        == patient_id
    ]

    if len(matches) == 0:

        raise RuntimeError(
            f"Patient {patient_id} not found in behavioral raw dataset."
        )

    if len(matches) > 1:

        print(
            f"[WARNING] {patient_id} has {len(matches)} rows. "
            f"Using first row."
        )

    demo_patients[patient_id] = matches.iloc[[0]].copy()

    print(
        f"[FOUND] {patient_id}: "
        f"{demo_patients[patient_id].shape}"
    )

print("[PASSED] P001-P005 behavioral rows identified.")

# ------------------------------------------------------------------------------------------------
# 15. EXACT RAW -> PREPROCESSOR
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("12. EXACT BEHAVIORAL PREPROCESSING")
print("=" * 100)

patient_128 = {}
patient_32 = {}
patient_probability = {}

for patient_id in [
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
]:

    row = demo_patients[patient_id].copy()

    # Keep only the exact columns expected by the saved preprocessor.
    X_raw = row[preprocessor_columns].copy()

    print(f"\n{patient_id}")
    print(f"Raw feature shape: {X_raw.shape}")

    # Saved preprocessing pipeline
    X_preprocessed = behavioral_preprocessor.transform(
        X_raw
    )

    X_preprocessed = np.asarray(
        X_preprocessed,
        dtype=np.float32
    )

    print(
        f"Preprocessor output shape: "
        f"{X_preprocessed.shape}"
    )

    if X_preprocessed.shape[1] != saved_input_dim:

        raise RuntimeError(
            f"{patient_id}: preprocessor produced "
            f"{X_preprocessed.shape[1]} features, "
            f"but encoder expects {saved_input_dim}."
        )

    # --------------------------------------------------------------------------------------------
    # 16. EXACT ENCODER
    # --------------------------------------------------------------------------------------------

    X_tensor = torch.tensor(
        X_preprocessed,
        dtype=torch.float32,
        device=DEVICE
    )

    with torch.no_grad():

        embedding, encoder_logit = behavioral_encoder(
            X_tensor
        )

        aligned = behavioral_alignment(
            embedding
        )

        expert_logit = behavioral_expert(
            aligned
        )

        probability = torch.sigmoid(
            expert_logit
        )

    embedding_np = (
        embedding.cpu()
        .numpy()
    )

    aligned_np = (
        aligned.cpu()
        .numpy()
    )

    probability_value = float(
        probability.squeeze().cpu().item()
    )

    patient_128[patient_id] = embedding_np
    patient_32[patient_id] = aligned_np
    patient_probability[patient_id] = probability_value

    print(
        f"128-D representation shape: "
        f"{embedding_np.shape}"
    )

    print(
        f"32-D aligned shape: "
        f"{aligned_np.shape}"
    )

    print(
        f"Expert probability: "
        f"{probability_value:.6f}"
    )

# ------------------------------------------------------------------------------------------------
# 17. FINAL BEHAVIORAL RESULTS
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("13. BEHAVIORAL PRODUCTION RESULTS")
print("=" * 100)

behavioral_results = []

for patient_id in [
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
]:

    prob = patient_probability[patient_id]

    prediction = (
        "ASD"
        if prob >= 0.5
        else "NON-ASD"
    )

    behavioral_results.append(
        {
            "Patient_ID": patient_id,
            "Behavioral_Probability": prob,
            "Behavioral_Prediction": prediction
        }
    )

    print(
        f"{patient_id} | "
        f"Probability = {prob:.6f} | "
        f"Prediction = {prediction}"
    )

behavioral_results_df = pd.DataFrame(
    behavioral_results
)

# ------------------------------------------------------------------------------------------------
# 18. SAVE ONLY NEW INFERENCE OUTPUTS
# ------------------------------------------------------------------------------------------------

OUTPUT_DIR = os.path.join(
    PROJECT_DIR,
    "Demo",
    "M4_4_9_3_behavioral_inference"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

results_path = os.path.join(
    OUTPUT_DIR,
    "behavioral_5_patient_inference.csv"
)

behavioral_results_df.to_csv(
    results_path,
    index=False
)

representations_32 = []

for patient_id in [
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
]:

    row = {
        "Patient_ID": patient_id
    }

    for i, value in enumerate(
        patient_32[patient_id][0]
    ):

        row[f"behavioral_32d_{i+1}"] = float(value)

    representations_32.append(row)

representations_32_df = pd.DataFrame(
    representations_32
)

rep32_path = os.path.join(
    OUTPUT_DIR,
    "behavioral_5_patient_32d_representations.csv"
)

representations_32_df.to_csv(
    rep32_path,
    index=False
)

# ------------------------------------------------------------------------------------------------
# 19. FINAL VALIDATION
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("14. FINAL VALIDATION")
print("=" * 100)

assert len(behavioral_results_df) == 5

assert all(
    np.isfinite(
        behavioral_results_df[
            "Behavioral_Probability"
        ].values
    )
)

assert all(
    (behavioral_results_df["Behavioral_Probability"] >= 0)
    &
    (behavioral_results_df["Behavioral_Probability"] <= 1)
)

assert representations_32_df.shape == (5, 33)

print("[PASSED] Five patients processed.")
print("[PASSED] Five exact 128-D behavioral representations generated.")
print("[PASSED] Five exact 32-D aligned behavioral representations generated.")
print("[PASSED] Five behavioral expert probabilities generated.")
print("[PASSED] All probabilities are finite and in [0,1].")

print("\nSaved inference output:")
print(results_path)

print("\nSaved 32-D representations:")
print(rep32_path)

print("\n" + "=" * 100)
print("M4.4.9.3 COMPLETED")
print("=" * 100)

print("""
NO TRAINING PERFORMED.
NO EXISTING CHECKPOINT MODIFIED.
NO PRODUCTION ARTIFACT OVERWRITTEN.

PIPELINE VALIDATED:

RAW BEHAVIORAL
      ↓
SAVED PREPROCESSOR
      ↓
87-D
      ↓
EXACT BEHAVIORAL ENCODER
      ↓
128-D
      ↓
EXACT BEHAVIORAL ALIGNMENT
      ↓
32-D
      ↓
EXACT BEHAVIORAL EXPERT
      ↓
PROBABILITY
      ↓
P001-P005 RESULTS

NEXT MAJOR STEP:
M4.4.10 — integrate the exact Genetic + Behavioral + Environmental
5-patient representations and perform final M3.2 weighted multimodal
production inference.
""")
```

```python
# ====================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3 DEFINITIVE FIX
# EXACT BEHAVIORAL P001–P005 PRODUCTION INFERENCE
#
# RAW → PREPROCESSOR → 87-D → EXACT ENCODER → 128-D
#     → EXACT ALIGNMENT → 32-D → SCALER → EXACT EXPERT → PROBABILITY
#
# IMPORTANT:
# This cell reconstructs architectures FROM THE ACTUAL CHECKPOINT STRUCTURE.
# No checkpoint is modified.
# ====================================================================================================

import os
import warnings
import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

warnings.filterwarnings("ignore")

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"
BEHAVIORAL_DIR = os.path.join(PROJECT_DIR, "Behavioral_Model")
FUSION_DIR = os.path.join(PROJECT_DIR, "05_multimodal_fusion")
DEMO_DIR = os.path.join(FUSION_DIR, "M4_demo_patients")

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3 DEFINITIVE FIX")
print("EXACT BEHAVIORAL P001–P005 PRODUCTION INFERENCE")
print("=" * 100)
print("Device :", DEVICE)
print("PyTorch:", torch.__version__)
print("Project:", PROJECT_DIR)


# ====================================================================================================
# 1. ROBUST FILE FINDER
# ====================================================================================================

def find_file(base_dir, filename):

    matches = []

    for root, dirs, files in os.walk(base_dir):

        if filename in files:
            matches.append(
                os.path.join(root, filename)
            )

    if not matches:

        raise FileNotFoundError(
            f"\nCould not find:\n{filename}\n"
            f"inside:\n{base_dir}"
        )

    return sorted(matches, key=len)[0]


if not os.path.isdir(PROJECT_DIR):
    raise FileNotFoundError(PROJECT_DIR)

print("\n[PASSED] Project directory available.")


# ====================================================================================================
# 2. FIND ALL PRODUCTION ARTIFACTS
# ====================================================================================================

print("\n" + "=" * 100)
print("1. PRODUCTION ARTIFACT DISCOVERY")
print("-" * 100)

PATHS = {}

PATHS["raw"] = find_file(
    BEHAVIORAL_DIR,
    "behavioral_dataset.csv"
)

PATHS["preprocessor"] = find_file(
    BEHAVIORAL_DIR,
    "behavioral_preprocessor.joblib"
)

PATHS["encoder"] = find_file(
    BEHAVIORAL_DIR,
    "behavioral_encoder_128d.pt"
)

PATHS["alignment"] = find_file(
    BEHAVIORAL_DIR,
    "behavioral_128_to_32_alignment.pt"
)

PATHS["scaler"] = find_file(
    PROJECT_DIR,
    "M3_1_behavioral_scaler.joblib"
)

PATHS["expert"] = find_file(
    PROJECT_DIR,
    "M3_1_behavioral_expert.pt"
)

for name, path in PATHS.items():

    size_mb = os.path.getsize(path) / (1024 * 1024)

    print(
        f"[FOUND] {name:<16} "
        f"{size_mb:.4f} MB"
    )

    print(
        f"        {path}"
    )

print("\n[PASSED] All behavioral artifacts located.")


# ====================================================================================================
# 3. FIND DEMO MANIFEST
# ====================================================================================================

print("\n" + "=" * 100)
print("2. DEMO MANIFEST")
print("-" * 100)

manifest_path = find_file(
    PROJECT_DIR,
    "M4_demo_patient_manifest.csv"
)

manifest = pd.read_csv(
    manifest_path
)

print("[FOUND]", manifest_path)
print("Manifest shape:", manifest.shape)

required_patients = [
    "P001",
    "P002",
    "P003",
    "P004",
    "P005"
]

manifest["patient_id"] = (
    manifest["patient_id"]
    .astype(str)
)

manifest = (
    manifest[
        manifest["patient_id"].isin(required_patients)
    ]
    .set_index("patient_id")
    .reindex(required_patients)
    .reset_index()
)

if manifest["behavioral_file"].isna().any():

    raise RuntimeError(
        "Manifest does not contain behavioral_file "
        "for all five patients."
    )

print(
    "[PASSED] P001–P005 manifest entries found."
)


# ====================================================================================================
# 4. RESOLVE DEMO FILES
# ====================================================================================================

print("\n" + "=" * 100)
print("3. DEMO BEHAVIORAL INPUTS")
print("-" * 100)

patient_files = {}

for _, row in manifest.iterrows():

    pid = row["patient_id"]

    relative_path = str(
        row["behavioral_file"]
    ).strip()

    candidates = [

        os.path.join(
            DEMO_DIR,
            relative_path
        ),

        os.path.join(
            FUSION_DIR,
            relative_path
        ),

        os.path.join(
            PROJECT_DIR,
            relative_path
        )
    ]

    found = None

    for candidate in candidates:

        if os.path.isfile(candidate):

            found = candidate
            break

    # Fallback: search using filename
    if found is None:

        filename = os.path.basename(
            relative_path
        )

        found = find_file(
            PROJECT_DIR,
            filename
        )

    patient_files[pid] = found

    print(
        f"[FOUND] {pid} -> {found}"
    )

print(
    "\n[PASSED] All five behavioral input files resolved."
)


# ====================================================================================================
# 5. LOAD SKLEARN ARTIFACTS
# ====================================================================================================

print("\n" + "=" * 100)
print("4. PREPROCESSING ARTIFACTS")
print("-" * 100)

preprocessor = joblib.load(
    PATHS["preprocessor"]
)

scaler = joblib.load(
    PATHS["scaler"]
)

print(
    "Preprocessor:",
    type(preprocessor)
)

print(
    "Scaler      :",
    type(scaler)
)

print(
    "Preprocessor features:",
    getattr(
        preprocessor,
        "n_features_in_",
        "N/A"
    )
)

print(
    "Scaler features:",
    getattr(
        scaler,
        "n_features_in_",
        "N/A"
    )
)

assert preprocessor.n_features_in_ == 18
assert scaler.n_features_in_ == 32

print(
    "[PASSED] Preprocessor expects 18 raw features."
)

print(
    "[PASSED] Scaler expects 32 aligned features."
)


# ====================================================================================================
# 6. TRUSTED TORCH LOADER
# ====================================================================================================

def load_checkpoint(path):

    try:

        return torch.load(
            path,
            map_location=DEVICE,
            weights_only=True
        )

    except Exception:

        print(
            "[INFO] Retrying trusted checkpoint:",
            os.path.basename(path)
        )

        return torch.load(
            path,
            map_location=DEVICE,
            weights_only=False
        )


encoder_ckpt = load_checkpoint(
    PATHS["encoder"]
)

alignment_ckpt = load_checkpoint(
    PATHS["alignment"]
)

expert_ckpt = load_checkpoint(
    PATHS["expert"]
)

print("\n[PASSED] All PyTorch checkpoints loaded.")


# ====================================================================================================
# 7. EXACT BEHAVIORAL ENCODER
# ====================================================================================================

print("\n" + "=" * 100)
print("5. EXACT BEHAVIORAL ENCODER")
print("-" * 100)

encoder_state = (
    encoder_ckpt["model_state_dict"]
)

input_dim = int(
    encoder_ckpt["input_dim"]
)

embedding_dim = int(
    encoder_ckpt["embedding_dim"]
)

print(
    "Saved input dimension    :",
    input_dim
)

print(
    "Saved embedding dimension:",
    embedding_dim
)

print(
    "Saved target column      :",
    encoder_ckpt.get(
        "target_column",
        "N/A"
    )
)


# --------------------------------------------------------------------------------
# EXACT ARCHITECTURE FROM YOUR CHECKPOINT
#
# encoder.0 = Linear(87,128)
# encoder.1 = BatchNorm(128)
# encoder.4 = Linear(128,128)
# encoder.5 = BatchNorm(128)
#
# Therefore:
#
# 0 Linear
# 1 BatchNorm
# 2 ReLU
# 3 Dropout
# 4 Linear
# 5 BatchNorm
# 6 ReLU
# 7 Dropout
#
# classifier = Linear(128,1)
# --------------------------------------------------------------------------------

class ExactBehavioralEncoder(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(

            nn.Linear(87, 128),      # 0

            nn.BatchNorm1d(128),     # 1

            nn.ReLU(),              # 2

            nn.Dropout(),           # 3

            nn.Linear(128, 128),     # 4

            nn.BatchNorm1d(128),     # 5

            nn.ReLU(),              # 6

            nn.Dropout()             # 7
        )

        self.classifier = nn.Linear(
            128,
            1
        )

    def forward(self, x):

        embedding = self.encoder(x)

        logit = self.classifier(
            embedding
        )

        return embedding, logit


behavioral_encoder = (
    ExactBehavioralEncoder()
    .to(DEVICE)
)

behavioral_encoder.load_state_dict(
    encoder_state,
    strict=True
)

behavioral_encoder.eval()

print(
    "[PASSED] EXACT behavioral encoder "
    "loaded with strict=True."
)

print(
    "[PASSED] Behavioral encoder eval() mode."
)

print(
    "Architecture:"
)

print(
    "87 -> 128 -> BN -> ReLU -> Dropout"
    " -> 128 -> BN -> ReLU -> Dropout"
    " -> 128"
)

print(
    "Classifier: 128 -> 1"
)


# ====================================================================================================
# 8. EXACT ALIGNMENT
# ====================================================================================================

print("\n" + "=" * 100)
print("6. EXACT BEHAVIORAL ALIGNMENT")
print("-" * 100)


# Handle all checkpoint formats discovered earlier.

if isinstance(alignment_ckpt, dict):

    if "state_dict" in alignment_ckpt:

        alignment_state = (
            alignment_ckpt["state_dict"]
        )

    elif "model_state_dict" in alignment_ckpt:

        alignment_state = (
            alignment_ckpt["model_state_dict"]
        )

    else:

        alignment_state = alignment_ckpt

else:

    alignment_state = alignment_ckpt


# Normalize:
#
# encoder.0.weight
# OR
# 0.weight
#
# into:
#
# network.0.weight

normalized = {}

for key, value in alignment_state.items():

    # Ignore decoder
    if key.startswith("decoder."):
        continue

    if key.startswith("encoder."):

        key = key[
            len("encoder.") :
        ]

    normalized[
        "network." + key
    ] = value


class ExactBehavioralAlignment(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(128, 64),

            nn.BatchNorm1d(64),

            nn.ReLU(),

            nn.Linear(64, 32)
        )

    def forward(self, x):

        return self.network(x)


behavioral_alignment = (
    ExactBehavioralAlignment()
    .to(DEVICE)
)

behavioral_alignment.load_state_dict(
    normalized,
    strict=True
)

behavioral_alignment.eval()

print(
    "[PASSED] EXACT behavioral alignment "
    "loaded with strict=True."
)

print(
    "[PASSED] Behavioral alignment eval() mode."
)

print(
    "Architecture: "
    "128 -> 64 -> BN -> ReLU -> 32"
)


# ====================================================================================================
# 9. EXACT EXPERT
# ====================================================================================================

print("\n" + "=" * 100)
print("7. EXACT BEHAVIORAL EXPERT")
print("-" * 100)

if isinstance(expert_ckpt, dict):

    if "state_dict" in expert_ckpt:

        expert_state = expert_ckpt[
            "state_dict"
        ]

    elif "model_state_dict" in expert_ckpt:

        expert_state = expert_ckpt[
            "model_state_dict"
        ]

    else:

        expert_state = expert_ckpt

else:

    expert_state = expert_ckpt


print("Expert state keys:")

for key, value in expert_state.items():

    if torch.is_tensor(value):

        print(
            f"  {key:<40}"
            f"{tuple(value.shape)}"
        )


# From the checkpoint already discovered:
#
# network.0 = Linear(32,64)
# network.1 = BatchNorm(64)
# network.4 = Linear(64,32)
# network.7 = Linear(32,1)
#
# Therefore non-parametric positions:
# 2 = ReLU
# 3 = Dropout
# 5 = ReLU
# 6 = Dropout

class ExactBehavioralExpert(nn.Module):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(32, 64),      # 0

            nn.BatchNorm1d(64),     # 1

            nn.ReLU(),             # 2

            nn.Dropout(),          # 3

            nn.Linear(64, 32),      # 4

            nn.ReLU(),             # 5

            nn.Dropout(),           # 6

            nn.Linear(32, 1)        # 7
        )

    def forward(self, x):

        return self.network(x)


behavioral_expert = (
    ExactBehavioralExpert()
    .to(DEVICE)
)

behavioral_expert.load_state_dict(
    expert_state,
    strict=True
)

behavioral_expert.eval()

print(
    "[PASSED] EXACT behavioral expert "
    "loaded with strict=True."
)

print(
    "[PASSED] Behavioral expert eval() mode."
)

print(
    "Architecture:"
)

print(
    "32 -> 64 -> BN -> ReLU -> Dropout"
    " -> 32 -> ReLU -> Dropout -> 1"
)


# ====================================================================================================
# 10. COMPLETE SYNTHETIC FORWARD TEST
# ====================================================================================================

print("\n" + "=" * 100)
print("8. COMPLETE FORWARD-PASS VALIDATION")
print("-" * 100)

test_input = torch.zeros(
    (2, 87),
    dtype=torch.float32
)

with torch.no_grad():

    test_embedding, test_encoder_logit = (
        behavioral_encoder(
            test_input
        )
    )

    test_aligned = (
        behavioral_alignment(
            test_embedding
        )
    )

    # IMPORTANT:
    # Expert was trained using the scaled 32-D representation.
    test_scaled = scaler.transform(
        test_aligned.numpy()
    )

    test_scaled_tensor = torch.tensor(
        test_scaled,
        dtype=torch.float32
    )

    test_expert_logit = (
        behavioral_expert(
            test_scaled_tensor
        )
    )

print(
    "Input      :",
    tuple(test_input.shape)
)

print(
    "Embedding  :",
    tuple(test_embedding.shape)
)

print(
    "Aligned    :",
    tuple(test_aligned.shape)
)

print(
    "Scaled     :",
    tuple(test_scaled_tensor.shape)
)

print(
    "Expert     :",
    tuple(test_expert_logit.shape)
)

assert test_embedding.shape == (2, 128)
assert test_aligned.shape == (2, 32)
assert test_scaled_tensor.shape == (2, 32)
assert test_expert_logit.shape == (2, 1)

print(
    "\n[PASSED] 87-D -> 128-D -> 32-D -> "
    "scaled 32-D -> 1 pipeline validated."
)


# ====================================================================================================
# 11. PROCESS REAL P001–P005
# ====================================================================================================

print("\n" + "=" * 100)
print("9. EXACT P001–P005 BEHAVIORAL INFERENCE")
print("-" * 100)

results = []

feature_columns = list(
    preprocessor.feature_names_in_
)

print(
    "Expected raw features:",
    len(feature_columns)
)

for pid in required_patients:

    print("\n" + "-" * 90)
    print(pid)

    path = patient_files[pid]

    raw = pd.read_csv(
        path
    )

    print(
        "Input file:",
        path
    )

    print(
        "Raw shape:",
        raw.shape
    )

    # The saved preprocessor expects exactly
    # the original 18 features.
    missing = [
        col
        for col in feature_columns
        if col not in raw.columns
    ]

    if missing:

        raise RuntimeError(
            f"{pid}: Missing preprocessing columns:\n"
            f"{missing}"
        )

    X_raw = raw[
        feature_columns
    ].copy()

    # --------------------------------------------------------------------------------
    # RAW → 87-D
    # --------------------------------------------------------------------------------

    X87 = preprocessor.transform(
        X_raw
    )

    X87 = np.asarray(
        X87,
        dtype=np.float32
    )

    if X87.ndim == 1:

        X87 = X87.reshape(
            1,
            -1
        )

    assert X87.shape[1] == 87

    print(
        "Preprocessor output:",
        X87.shape
    )

    # --------------------------------------------------------------------------------
    # 87-D → 128-D
    # --------------------------------------------------------------------------------

    X87_tensor = torch.tensor(
        X87,
        dtype=torch.float32
    )

    with torch.no_grad():

        embedding128, encoder_logit = (
            behavioral_encoder(
                X87_tensor
            )
        )

        # --------------------------------------------------------------------------------
        # 128-D → 32-D
        # --------------------------------------------------------------------------------

        aligned32 = (
            behavioral_alignment(
                embedding128
            )
        )

    # --------------------------------------------------------------------------------
    # 32-D → SCALER
    # --------------------------------------------------------------------------------

    aligned32_np = (
        aligned32.numpy()
    )

    scaled32_np = scaler.transform(
        aligned32_np
    )

    scaled32_np = np.asarray(
        scaled32_np,
        dtype=np.float32
    )

    # --------------------------------------------------------------------------------
    # 32-D SCALED → EXPERT
    # --------------------------------------------------------------------------------

    scaled32_tensor = torch.tensor(
        scaled32_np,
        dtype=torch.float32
    )

    with torch.no_grad():

        expert_logit = (
            behavioral_expert(
                scaled32_tensor
            )
        )

        probability = torch.sigmoid(
            expert_logit
        )

    embedding_np = (
        embedding128
        .numpy()
        .reshape(-1)
    )

    aligned_np = (
        aligned32
        .numpy()
        .reshape(-1)
    )

    scaled_np = (
        scaled32_np
        .reshape(-1)
    )

    logit = float(
        expert_logit
        .numpy()
        .reshape(-1)[0]
    )

    prob = float(
        probability
        .numpy()
        .reshape(-1)[0]
    )

    prediction = (
        "ASD"
        if prob >= 0.5
        else "NON-ASD"
    )

    print(
        "128-D embedding:",
        embedding_np.shape
    )

    print(
        "32-D aligned:",
        aligned_np.shape
    )

    print(
        "32-D scaled:",
        scaled_np.shape
    )

    print(
        "Expert logit:",
        f"{logit:.6f}"
    )

    print(
        "Probability:",
        f"{prob:.6f}"
    )

    print(
        "Prediction:",
        prediction
    )

    # Numerical checks

    assert embedding_np.shape == (128,)
    assert aligned_np.shape == (32,)
    assert scaled_np.shape == (32,)

    assert np.all(
        np.isfinite(embedding_np)
    )

    assert np.all(
        np.isfinite(aligned_np)
    )

    assert np.all(
        np.isfinite(scaled_np)
    )

    assert 0.0 <= prob <= 1.0

    results.append({

        "patient_id":
            pid,

        "behavioral_probability":
            prob,

        "behavioral_logit":
            logit,

        "behavioral_prediction":
            prediction,

        "embedding_128":
            embedding_np,

        "aligned_32":
            aligned_np,

        "scaled_32":
            scaled_np
    })


# ====================================================================================================
# 12. FINAL RESULT TABLE
# ====================================================================================================

print("\n" + "=" * 100)
print("10. P001–P005 BEHAVIORAL RESULTS")
print("-" * 100)

result_table = pd.DataFrame([

    {
        "patient_id":
            r["patient_id"],

        "behavioral_probability":
            r["behavioral_probability"],

        "behavioral_logit":
            r["behavioral_logit"],

        "behavioral_prediction":
            r["behavioral_prediction"]
    }

    for r in results
])

print(
    result_table.to_string(
        index=False,
        float_format=lambda x:
            f"{x:.6f}"
    )
)


# ====================================================================================================
# 13. FINAL VALIDATION
# ====================================================================================================

print("\n" + "=" * 100)
print("11. FINAL M4.4.9.3 VALIDATION")
print("-" * 100)

assert len(results) == 5

print(
    "[PASSED] P001 processed."
)

print(
    "[PASSED] P002 processed."
)

print(
    "[PASSED] P003 processed."
)

print(
    "[PASSED] P004 processed."
)

print(
    "[PASSED] P005 processed."
)

print(
    "[PASSED] Exact saved preprocessing executed."
)

print(
    "[PASSED] Exact 87-D behavioral representation generated."
)

print(
    "[PASSED] Exact 128-D behavioral embeddings generated."
)

print(
    "[PASSED] Exact 128 -> 32 alignment executed."
)

print(
    "[PASSED] Exact 32-D representations generated."
)

print(
    "[PASSED] Exact saved 32-D scaler executed."
)

print(
    "[PASSED] Exact behavioral expert executed."
)

print(
    "[PASSED] Five behavioral probabilities generated."
)

print(
    "[PASSED] Numerical sanity checks passed."
)

print("\n" + "=" * 100)
print("M4.4.9.3 COMPLETED SUCCESSFULLY")
print("=" * 100)

print("""
NO TRAINING PERFORMED.
NO CHECKPOINT MODIFIED.
NO PRODUCTION ARTIFACT OVERWRITTEN.

NEXT:
M4.4.10 — GENETIC + ENVIRONMENTAL P001–P005
EXACT REPRESENTATION INFERENCE.

AFTER THAT:
FINAL MULTIMODAL FUSION → P001–P005
""")
```

```python
# ================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.10.1
# EXACT GENETIC + ENVIRONMENTAL CHECKPOINT ARCHITECTURE INSPECTION
# ================================================================================================

import os
import torch
from collections import OrderedDict

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.10.1")
print("EXACT GENETIC + ENVIRONMENTAL CHECKPOINT ARCHITECTURE INSPECTION")
print("=" * 100)

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

print(f"Device  : {DEVICE}")
print(f"PyTorch : {torch.__version__}")
print(f"Project : {PROJECT_DIR}")

if not os.path.isdir(PROJECT_DIR):
    raise FileNotFoundError(f"Project directory not found:\n{PROJECT_DIR}")

print("[PASSED] Project directory available.")


# ------------------------------------------------------------------------------------------------
# Robust recursive file finder
# ------------------------------------------------------------------------------------------------

def find_file_recursive(base_dir, filename):
    matches = []

    for root, dirs, files in os.walk(base_dir):
        for file in files:
            if file == filename:
                matches.append(os.path.join(root, file))

    if not matches:
        raise FileNotFoundError(
            f"\nCould not locate:\n{filename}\ninside:\n{base_dir}"
        )

    # Prefer the shortest path if duplicates exist
    matches.sort(key=lambda p: (len(p), p))
    return matches[0]


# ------------------------------------------------------------------------------------------------
# Locate exact checkpoints
# ------------------------------------------------------------------------------------------------

GENETIC_DIR = os.path.join(
    PROJECT_DIR,
    "Genetic_Model"
)

ENVIRONMENTAL_DIR = os.path.join(
    PROJECT_DIR,
    "Environmental_Model"
)

genetic_encoder_path = find_file_recursive(
    GENETIC_DIR,
    "genetic_transformer_128d.pt"
)

genetic_alignment_path = find_file_recursive(
    GENETIC_DIR,
    "genetic_128_to_32_alignment.pt"
)

environmental_encoder_path = find_file_recursive(
    ENVIRONMENTAL_DIR,
    "environmental_encoder_128d.pt"
)

environmental_alignment_path = find_file_recursive(
    ENVIRONMENTAL_DIR,
    "environmental_128_to_32_alignment.pt"
)


print()
print("=" * 100)
print("1. CHECKPOINT DISCOVERY")
print("-" * 100)

print("[FOUND] Genetic encoder:")
print(f"        {genetic_encoder_path}")

print("[FOUND] Genetic alignment:")
print(f"        {genetic_alignment_path}")

print("[FOUND] Environmental encoder:")
print(f"        {environmental_encoder_path}")

print("[FOUND] Environmental alignment:")
print(f"        {environmental_alignment_path}")

print()
print("[PASSED] All four checkpoints located.")


# ------------------------------------------------------------------------------------------------
# Safe checkpoint loader
# ------------------------------------------------------------------------------------------------

def load_checkpoint(path):
    try:
        checkpoint = torch.load(
            path,
            map_location=DEVICE,
            weights_only=True
        )
        print(f"[INFO] weights_only=True succeeded: {os.path.basename(path)}")
        return checkpoint

    except Exception as e:
        print(f"[INFO] weights_only=True failed: {os.path.basename(path)}")
        print(f"       {type(e).__name__}: {e}")
        print("       Retrying trusted checkpoint with weights_only=False.")

        checkpoint = torch.load(
            path,
            map_location=DEVICE,
            weights_only=False
        )

        print(
            f"[PASSED] Trusted checkpoint loaded: "
            f"{os.path.basename(path)}"
        )

        return checkpoint


# ------------------------------------------------------------------------------------------------
# State-dict extraction helper
# ------------------------------------------------------------------------------------------------

def extract_state_dict(checkpoint):

    if isinstance(checkpoint, OrderedDict):
        return checkpoint

    if isinstance(checkpoint, dict):

        # Most common format
        if "model_state_dict" in checkpoint:
            return checkpoint["model_state_dict"]

        # Other possible naming conventions
        for key in [
            "state_dict",
            "model_state",
            "encoder_state_dict",
            "alignment_state_dict"
        ]:
            if key in checkpoint:
                candidate = checkpoint[key]

                if isinstance(candidate, (dict, OrderedDict)):
                    return candidate

    raise RuntimeError(
        "Could not identify a state_dict inside checkpoint.\n"
        f"Checkpoint type: {type(checkpoint)}\n"
        f"Checkpoint keys: "
        f"{list(checkpoint.keys()) if isinstance(checkpoint, dict) else 'N/A'}"
    )


# ------------------------------------------------------------------------------------------------
# Load checkpoints
# ------------------------------------------------------------------------------------------------

print()
print("=" * 100)
print("2. LOADING CHECKPOINTS")
print("-" * 100)

genetic_encoder_ckpt = load_checkpoint(genetic_encoder_path)
genetic_alignment_ckpt = load_checkpoint(genetic_alignment_path)

environmental_encoder_ckpt = load_checkpoint(environmental_encoder_path)
environmental_alignment_ckpt = load_checkpoint(environmental_alignment_path)

print()
print("[PASSED] All genetic/environmental checkpoints loaded.")


# ------------------------------------------------------------------------------------------------
# Inspect checkpoint metadata
# ------------------------------------------------------------------------------------------------

def print_checkpoint_structure(name, checkpoint):

    print()
    print("=" * 100)
    print(name)
    print("-" * 100)

    print(f"Checkpoint type: {type(checkpoint)}")

    if isinstance(checkpoint, dict):

        print()
        print("Top-level keys:")

        for key, value in checkpoint.items():

            if isinstance(value, (dict, OrderedDict)):
                print(
                    f"  {key:<35} "
                    f"type={type(value)} "
                    f"keys={len(value)}"
                )

            elif torch.is_tensor(value):
                print(
                    f"  {key:<35} "
                    f"tensor shape={tuple(value.shape)}"
                )

            else:
                print(
                    f"  {key:<35} "
                    f"type={type(value)} "
                    f"value={value}"
                )


print_checkpoint_structure(
    "3A. GENETIC ENCODER CHECKPOINT",
    genetic_encoder_ckpt
)

print_checkpoint_structure(
    "3B. GENETIC ALIGNMENT CHECKPOINT",
    genetic_alignment_ckpt
)

print_checkpoint_structure(
    "3C. ENVIRONMENTAL ENCODER CHECKPOINT",
    environmental_encoder_ckpt
)

print_checkpoint_structure(
    "3D. ENVIRONMENTAL ALIGNMENT CHECKPOINT",
    environmental_alignment_ckpt
)


# ------------------------------------------------------------------------------------------------
# Extract exact state dictionaries
# ------------------------------------------------------------------------------------------------

genetic_encoder_state = extract_state_dict(genetic_encoder_ckpt)
genetic_alignment_state = extract_state_dict(genetic_alignment_ckpt)

environmental_encoder_state = extract_state_dict(
    environmental_encoder_ckpt
)
environmental_alignment_state = extract_state_dict(
    environmental_alignment_ckpt
)


# ------------------------------------------------------------------------------------------------
# Exact state-dict inspection
# ------------------------------------------------------------------------------------------------

def inspect_state_dict(name, state_dict):

    print()
    print("=" * 100)
    print(name)
    print("-" * 100)

    print(f"Number of state-dict keys: {len(state_dict)}")
    print()

    for key, value in state_dict.items():

        if torch.is_tensor(value):

            print(
                f"{key:<55} "
                f"shape={str(tuple(value.shape)):<25} "
                f"dtype={value.dtype}"
            )

        else:

            print(
                f"{key:<55} "
                f"type={type(value)}"
            )


inspect_state_dict(
    "4A. EXACT GENETIC ENCODER STATE-DICT",
    genetic_encoder_state
)

inspect_state_dict(
    "4B. EXACT GENETIC ALIGNMENT STATE-DICT",
    genetic_alignment_state
)

inspect_state_dict(
    "4C. EXACT ENVIRONMENTAL ENCODER STATE-DICT",
    environmental_encoder_state
)

inspect_state_dict(
    "4D. EXACT ENVIRONMENTAL ALIGNMENT STATE-DICT",
    environmental_alignment_state
)


# ------------------------------------------------------------------------------------------------
# Metadata extraction
# ------------------------------------------------------------------------------------------------

def print_metadata(name, checkpoint):

    print()
    print("=" * 100)
    print(name)
    print("-" * 100)

    if not isinstance(checkpoint, dict):
        print("[INFO] No dictionary metadata available.")
        return

    metadata_keys = [
        "input_dim",
        "embedding_dim",
        "target_column",
        "output_dim",
        "latent_dim",
        "num_features",
        "feature_dim"
    ]

    found = False

    for key in metadata_keys:

        if key in checkpoint:

            print(
                f"{key:<25}: {checkpoint[key]}"
            )

            found = True

    if not found:
        print("[INFO] No standard metadata keys found.")


print_metadata(
    "5A. GENETIC ENCODER METADATA",
    genetic_encoder_ckpt
)

print_metadata(
    "5B. GENETIC ALIGNMENT METADATA",
    genetic_alignment_ckpt
)

print_metadata(
    "5C. ENVIRONMENTAL ENCODER METADATA",
    environmental_encoder_ckpt
)

print_metadata(
    "5D. ENVIRONMENTAL ALIGNMENT METADATA",
    environmental_alignment_ckpt
)


# ------------------------------------------------------------------------------------------------
# Final structural summary
# ------------------------------------------------------------------------------------------------

print()
print("=" * 100)
print("6. M4.4.10.1 STRUCTURAL INSPECTION STATUS")
print("=" * 100)

print("[PASSED] Genetic encoder checkpoint inspected.")
print("[PASSED] Genetic alignment checkpoint inspected.")
print("[PASSED] Environmental encoder checkpoint inspected.")
print("[PASSED] Environmental alignment checkpoint inspected.")
print()
print("[PASSED] Exact state-dict keys and tensor shapes extracted.")
print()
print("IMPORTANT:")
print("This cell ONLY inspects saved checkpoints.")
print("NO model reconstruction performed.")
print("NO training performed.")
print("NO predictions generated.")
print("NO checkpoint modified.")
print("NO production artifact overwritten.")
print()
print("NEXT:")
print("Paste the COMPLETE output here.")
print("The next reconstruction cell will be generated directly")
print("from these exact checkpoint keys and shapes.")
print("=" * 100)
```

```python
# ====================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3.2
# EXACT BEHAVIORAL ENCODER RECONSTRUCTION — KEY-ALIGNED
# ====================================================================================================

import os
import torch
import torch.nn as nn

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3.2")
print("EXACT BEHAVIORAL ENCODER RECONSTRUCTION — KEY-ALIGNED")
print("=" * 100)

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

ENCODER_PATH = os.path.join(
    PROJECT_DIR,
    "Behavioral_Model",
    "03_representation_learning",
    "behavioral_encoder_128d.pt"
)

print(f"Device  : {DEVICE}")
print(f"PyTorch : {torch.__version__}")
print(f"Project : {PROJECT_DIR}")

# ----------------------------------------------------------------------------------------------------
# 1. CHECK FILE
# ----------------------------------------------------------------------------------------------------

if not os.path.isfile(ENCODER_PATH):
    raise FileNotFoundError(
        f"Behavioral encoder checkpoint not found:\n{ENCODER_PATH}"
    )

print("[FOUND] Behavioral encoder checkpoint")
print(f"        {ENCODER_PATH}")

# ----------------------------------------------------------------------------------------------------
# 2. LOAD CHECKPOINT
# ----------------------------------------------------------------------------------------------------

checkpoint = torch.load(
    ENCODER_PATH,
    map_location=DEVICE,
    weights_only=False
)

print("\n[PASSED] Behavioral checkpoint loaded.")

# ----------------------------------------------------------------------------------------------------
# 3. READ SAVED METADATA
# ----------------------------------------------------------------------------------------------------

input_dim = int(checkpoint["input_dim"])
embedding_dim = int(checkpoint["embedding_dim"])
target_column = checkpoint["target_column"]

state = checkpoint["model_state_dict"]

print("\n" + "=" * 100)
print("3. SAVED MODEL METADATA")
print("=" * 100)

print(f"Input dimension     : {input_dim}")
print(f"Embedding dimension : {embedding_dim}")
print(f"Target column       : {target_column}")

if input_dim != 87:
    raise RuntimeError(
        f"Unexpected saved input dimension: {input_dim}"
    )

if embedding_dim != 128:
    raise RuntimeError(
        f"Unexpected saved embedding dimension: {embedding_dim}"
    )

print("[PASSED] Saved dimensions are exactly 87 -> 128.")

# ----------------------------------------------------------------------------------------------------
# 4. EXACT ARCHITECTURE
#
# State-dict proves:
#
# encoder.0 = Linear(87, 128)
# encoder.1 = BatchNorm1d(128)
# encoder.2 = ReLU()
# encoder.3 = Dropout / activation layer WITHOUT PARAMETERS
# encoder.4 = Linear(128, 128)
# encoder.5 = BatchNorm1d(128)
# encoder.6 = ReLU()
# encoder.7 = Dropout / activation layer WITHOUT PARAMETERS
#
# IMPORTANT:
# The parameter-bearing layers are what determine the exact dimensions.
# ----------------------------------------------------------------------------------------------------

class BehavioralEncoderExact(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(87, 128),          # encoder.0
            nn.BatchNorm1d(128),          # encoder.1
            nn.ReLU(),                   # encoder.2
            nn.Dropout(),                # encoder.3

            nn.Linear(128, 128),         # encoder.4
            nn.BatchNorm1d(128),         # encoder.5
            nn.ReLU(),                   # encoder.6
            nn.Dropout()                 # encoder.7
        )

        self.classifier = nn.Linear(128, 1)

    def forward(self, x, return_embedding=False):

        embedding = self.encoder(x)

        logit = self.classifier(embedding)

        if return_embedding:
            return embedding, logit

        return logit


print("\n" + "=" * 100)
print("4. RECONSTRUCTING EXACT BEHAVIORAL ENCODER")
print("=" * 100)

behavioral_encoder = BehavioralEncoderExact().to(DEVICE)

# ----------------------------------------------------------------------------------------------------
# 5. STRICT CHECKPOINT LOADING
# ----------------------------------------------------------------------------------------------------

print("\nLoading checkpoint with strict=True...")

behavioral_encoder.load_state_dict(
    state,
    strict=True
)

behavioral_encoder.eval()

print("[PASSED] Behavioral encoder reconstructed with strict=True.")
print("[PASSED] Behavioral encoder set to eval().")

# ----------------------------------------------------------------------------------------------------
# 6. STATE-DICT EXACTNESS CHECK
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("5. STATE-DICT EXACTNESS CHECK")
print("=" * 100)

model_keys = list(behavioral_encoder.state_dict().keys())
checkpoint_keys = list(state.keys())

if model_keys != checkpoint_keys:

    print("[WARNING] Key ordering differs.")

    missing = sorted(set(checkpoint_keys) - set(model_keys))
    unexpected = sorted(set(model_keys) - set(checkpoint_keys))

    if missing:
        print("Missing keys:", missing)

    if unexpected:
        print("Unexpected keys:", unexpected)

    raise RuntimeError(
        "Behavioral encoder state-dict structure does not exactly match checkpoint."
    )

print("[PASSED] Model state-dict keys exactly match checkpoint.")

# ----------------------------------------------------------------------------------------------------
# 7. PARAMETER SHAPE VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("6. PARAMETER SHAPE VALIDATION")
print("=" * 100)

expected_shapes = {
    "encoder.0.weight": (128, 87),
    "encoder.0.bias": (128,),

    "encoder.1.weight": (128,),
    "encoder.1.bias": (128,),
    "encoder.1.running_mean": (128,),
    "encoder.1.running_var": (128,),

    "encoder.4.weight": (128, 128),
    "encoder.4.bias": (128,),

    "encoder.5.weight": (128,),
    "encoder.5.bias": (128,),
    "encoder.5.running_mean": (128,),
    "encoder.5.running_var": (128,),

    "classifier.weight": (1, 128),
    "classifier.bias": (1,)
}

for key, expected_shape in expected_shapes.items():

    actual_shape = tuple(
        behavioral_encoder.state_dict()[key].shape
    )

    if actual_shape != expected_shape:
        raise RuntimeError(
            f"Shape mismatch for {key}: "
            f"expected {expected_shape}, got {actual_shape}"
        )

    print(f"[PASSED] {key:<40} {actual_shape}")

# ----------------------------------------------------------------------------------------------------
# 8. FORWARD-PASS TEST
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("7. FORWARD-PASS VALIDATION")
print("=" * 100)

torch.manual_seed(42)

test_input = torch.randn(
    2,
    87,
    dtype=torch.float32,
    device=DEVICE
)

with torch.no_grad():

    embedding, logits = behavioral_encoder(
        test_input,
        return_embedding=True
    )

print(f"Input shape       : {tuple(test_input.shape)}")
print(f"Embedding shape   : {tuple(embedding.shape)}")
print(f"Classifier shape  : {tuple(logits.shape)}")

if tuple(test_input.shape) != (2, 87):
    raise RuntimeError("Unexpected test input shape.")

if tuple(embedding.shape) != (2, 128):
    raise RuntimeError(
        f"Expected embedding shape (2,128), got {tuple(embedding.shape)}"
    )

if tuple(logits.shape) != (2, 1):
    raise RuntimeError(
        f"Expected classifier shape (2,1), got {tuple(logits.shape)}"
    )

print("[PASSED] 87-D input accepted.")
print("[PASSED] Exact 128-D embedding produced.")
print("[PASSED] One classifier logit produced.")

# ----------------------------------------------------------------------------------------------------
# 9. NUMERICAL SANITY
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("8. NUMERICAL SANITY CHECK")
print("=" * 100)

if not torch.isfinite(embedding).all():
    raise RuntimeError("Embedding contains NaN or Inf values.")

if not torch.isfinite(logits).all():
    raise RuntimeError("Classifier output contains NaN or Inf values.")

probabilities = torch.sigmoid(logits)

if not torch.isfinite(probabilities).all():
    raise RuntimeError("Probability contains NaN or Inf values.")

if not ((probabilities >= 0).all() and (probabilities <= 1).all()):
    raise RuntimeError("Probability outside [0,1].")

print("[PASSED] 128-D embedding contains only finite values.")
print("[PASSED] Classifier output contains only finite values.")
print("[PASSED] Sigmoid probability is valid.")

print("\nProbability values:")
print(probabilities.squeeze(1).cpu().numpy())

# ----------------------------------------------------------------------------------------------------
# 10. FINAL STATUS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("M4.4.9.3.2 COMPLETED")
print("=" * 100)

print("[PASSED] Exact behavioral encoder architecture reconstructed.")
print("[PASSED] Checkpoint loaded with strict=True.")
print("[PASSED] State-dict keys exactly match.")
print("[PASSED] Parameter shapes exactly match.")
print("[PASSED] 87-D input accepted.")
print("[PASSED] Exact 128-D behavioral representation produced.")
print("[PASSED] Classifier produces one output logit.")
print("[PASSED] Probability validation passed.")
print("[PASSED] Numerical sanity checks passed.")

print("\nNO TRAINING PERFORMED.")
print("NO CHECKPOINT MODIFIED.")
print("NO PRODUCTION ARTIFACT OVERWRITTEN.")

print("\nNEXT:")
print("M4.4.9.3.3 — EXACT BEHAVIORAL ALIGNMENT INSPECTION.")
print("We will inspect the saved alignment checkpoint FIRST.")
print("Only after inspecting its exact keys will we reconstruct it.")
print("This prevents another architecture-guessing error.")

print("=" * 100)
```

```python
# ====================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3.7
# COMPLETE BEHAVIORAL P001–P005 PRODUCTION INFERENCE
# EXACT RAW → PREPROCESSOR → ENCODER → ALIGNMENT → SCALER → EXPERT
# ====================================================================================================

import os
import glob
import json
import joblib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3.7")
print("COMPLETE BEHAVIORAL P001–P005 PRODUCTION INFERENCE")
print("=" * 100)

DEVICE = torch.device("cpu")
PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

print(f"Device  : {DEVICE}")
print(f"Project : {PROJECT_DIR}")

if not os.path.isdir(PROJECT_DIR):
    raise FileNotFoundError(f"Project directory not found:\n{PROJECT_DIR}")

print("[PASSED] Project directory available.")


# ====================================================================================================
# 1. ROBUST ARTIFACT DISCOVERY
# ====================================================================================================

print("\n" + "=" * 100)
print("1. ROBUST ARTIFACT DISCOVERY")
print("=" * 100)

def find_file(filename):
    matches = []

    for root, dirs, files in os.walk(PROJECT_DIR):
        if filename in files:
            matches.append(os.path.join(root, filename))

    if not matches:
        raise FileNotFoundError(
            f"Could not locate '{filename}' anywhere inside:\n{PROJECT_DIR}"
        )

    # Prefer paths with the expected modality directory when duplicates exist.
    matches.sort(
        key=lambda p: (
            "Behavioral_Model" not in p,
            len(p)
        )
    )

    return matches[0]


PATHS = {}

PATHS["raw"] = find_file("behavioral_dataset.csv")
PATHS["preprocessor"] = find_file("behavioral_preprocessor.joblib")
PATHS["encoder"] = find_file("behavioral_encoder_128d.pt")
PATHS["alignment"] = find_file("behavioral_128_to_32_alignment.pt")
PATHS["scaler"] = find_file("M3_1_behavioral_scaler.joblib")
PATHS["expert"] = find_file("M3_1_behavioral_expert.pt")
PATHS["manifest"] = find_file("M4_demo_patient_manifest.csv")

for name, path in PATHS.items():
    size_mb = os.path.getsize(path) / (1024 * 1024)
    print(f"[FOUND] {name:<16} {size_mb:.4f} MB")
    print(f"        {path}")

print("\n[PASSED] All required behavioral artifacts located.")


# ====================================================================================================
# 2. DEMO MANIFEST
# ====================================================================================================

print("\n" + "=" * 100)
print("2. DEMO MANIFEST")
print("=" * 100)

manifest = pd.read_csv(PATHS["manifest"])

required_manifest_columns = [
    "patient_id",
    "expected_demo_label",
    "expected_demo_label_numeric",
    "behavioral_file"
]

missing_manifest_columns = [
    c for c in required_manifest_columns
    if c not in manifest.columns
]

if missing_manifest_columns:
    raise RuntimeError(
        f"Manifest is missing required columns: {missing_manifest_columns}"
    )

required_patients = ["P001", "P002", "P003", "P004", "P005"]

manifest = manifest[
    manifest["patient_id"].astype(str).isin(required_patients)
].copy()

manifest["patient_id"] = manifest["patient_id"].astype(str)

manifest = (
    manifest
    .set_index("patient_id")
    .reindex(required_patients)
    .reset_index()
)

if manifest["behavioral_file"].isna().any():
    raise RuntimeError(
        "Manifest does not contain behavioral files for all P001–P005."
    )

print(f"Manifest shape: {manifest.shape}")
print(f"Patients: {required_patients}")


# ====================================================================================================
# 3. RESOLVE PATIENT BEHAVIORAL FILES
# ====================================================================================================

print("\n" + "=" * 100)
print("3. RESOLVING P001–P005 BEHAVIORAL INPUT FILES")
print("=" * 100)

patient_paths = {}

for _, row in manifest.iterrows():

    patient_id = row["patient_id"]
    relative_path = str(row["behavioral_file"])

    candidates = [
        os.path.join(
            os.path.dirname(PATHS["manifest"]),
            relative_path
        ),
        os.path.join(
            PROJECT_DIR,
            relative_path
        )
    ]

    found = None

    for candidate in candidates:
        if os.path.isfile(candidate):
            found = candidate
            break

    if found is None:

        # Final robust fallback: search by filename.
        filename = os.path.basename(relative_path)

        possible = []

        for root, dirs, files in os.walk(PROJECT_DIR):
            if filename in files:
                possible.append(
                    os.path.join(root, filename)
                )

        # Prefer path containing patient ID.
        possible = sorted(
            possible,
            key=lambda p: (
                patient_id not in p,
                len(p)
            )
        )

        if possible:
            found = possible[0]

    if found is None:
        raise FileNotFoundError(
            f"Could not resolve behavioral input for {patient_id}.\n"
            f"Manifest path: {relative_path}"
        )

    patient_paths[patient_id] = found

    print(f"[FOUND] {patient_id:<5} {found}")

print("\n[PASSED] All five behavioral demo files resolved.")


# ====================================================================================================
# 4. LOAD PREPROCESSING ARTIFACTS
# ====================================================================================================

print("\n" + "=" * 100)
print("4. LOADING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)

preprocessor = joblib.load(PATHS["preprocessor"])
scaler = joblib.load(PATHS["scaler"])

print(f"Preprocessor: {type(preprocessor)}")
print(f"Scaler      : {type(scaler)}")

print(
    f"Preprocessor n_features_in_: "
    f"{getattr(preprocessor, 'n_features_in_', 'N/A')}"
)

print(
    f"Scaler n_features_in_: "
    f"{getattr(scaler, 'n_features_in_', 'N/A')}"
)

if getattr(preprocessor, "n_features_in_", None) != 18:
    raise RuntimeError(
        "Unexpected behavioral preprocessor input dimension."
    )

if getattr(scaler, "n_features_in_", None) != 32:
    raise RuntimeError(
        "Unexpected behavioral scaler input dimension."
    )

print("[PASSED] Preprocessor expects 18 raw features.")
print("[PASSED] Scaler expects 32 aligned features.")


# ====================================================================================================
# 5. LOAD EXACT ENCODER
# ====================================================================================================

print("\n" + "=" * 100)
print("5. LOADING EXACT BEHAVIORAL ENCODER")
print("=" * 100)

encoder_checkpoint = torch.load(
    PATHS["encoder"],
    map_location=DEVICE,
    weights_only=True
)

encoder_state = encoder_checkpoint["model_state_dict"]

input_dim = int(encoder_checkpoint["input_dim"])
embedding_dim = int(encoder_checkpoint["embedding_dim"])

if input_dim != 87 or embedding_dim != 128:
    raise RuntimeError(
        f"Unexpected encoder dimensions: {input_dim} -> {embedding_dim}"
    )


class BehavioralEncoder(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(87, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(),
            nn.Linear(128, 128),
            nn.BatchNorm1d(128)
        )

        self.classifier = nn.Linear(128, 1)

    def forward(self, x):
        embedding = self.encoder(x)
        logit = self.classifier(embedding)
        return embedding, logit


behavioral_encoder = BehavioralEncoder().to(DEVICE)

behavioral_encoder.load_state_dict(
    encoder_state,
    strict=True
)

behavioral_encoder.eval()

print("[PASSED] Exact behavioral encoder reconstructed.")
print("[PASSED] Encoder checkpoint loaded with strict=True.")


# ====================================================================================================
# 6. LOAD EXACT ALIGNMENT
# ====================================================================================================

print("\n" + "=" * 100)
print("6. LOADING EXACT BEHAVIORAL ALIGNMENT")
print("=" * 100)

alignment_checkpoint = torch.load(
    PATHS["alignment"],
    map_location=DEVICE,
    weights_only=True
)

alignment_state = alignment_checkpoint["model_state_dict"]


class BehavioralAlignment(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )

        self.decoder = nn.Sequential(
            nn.Linear(32, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Linear(64, 128)
        )

    def forward(self, x):
        aligned = self.encoder(x)
        reconstructed = self.decoder(aligned)
        return aligned, reconstructed


behavioral_alignment = BehavioralAlignment().to(DEVICE)

behavioral_alignment.load_state_dict(
    alignment_state,
    strict=True
)

behavioral_alignment.eval()

print("[PASSED] Exact behavioral alignment reconstructed.")
print("[PASSED] Alignment checkpoint loaded with strict=True.")


# ====================================================================================================
# 7. LOAD EXACT BEHAVIORAL EXPERT
# ====================================================================================================

print("\n" + "=" * 100)
print("7. LOADING EXACT BEHAVIORAL EXPERT")
print("=" * 100)

expert_state = torch.load(
    PATHS["expert"],
    map_location=DEVICE,
    weights_only=True
)


class BehavioralExpert(nn.Module):

    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(32, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(),
            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.network(x)


behavioral_expert = BehavioralExpert().to(DEVICE)

behavioral_expert.load_state_dict(
    expert_state,
    strict=True
)

behavioral_expert.eval()

print("[PASSED] Exact behavioral expert reconstructed.")
print("[PASSED] Expert checkpoint loaded with strict=True.")


# ====================================================================================================
# 8. RAW FEATURE DEFINITIONS
# ====================================================================================================

print("\n" + "=" * 100)
print("8. RAW FEATURE VALIDATION")
print("=" * 100)

raw_dataset = pd.read_csv(PATHS["raw"])

target_column = "Class/ASD"

raw_feature_columns = [
    "A1_Score",
    "A2_Score",
    "A3_Score",
    "A4_Score",
    "A5_Score",
    "A6_Score",
    "A7_Score",
    "A8_Score",
    "A9_Score",
    "A10_Score",
    "age",
    "gender",
    "ethnicity",
    "jundice",
    "austim",
    "contry_of_res",
    "used_app_before",
    "relation"
]

missing_raw = [
    c for c in raw_feature_columns
    if c not in raw_dataset.columns
]

if missing_raw:
    raise RuntimeError(
        f"Required raw behavioral features missing: {missing_raw}"
    )

print(f"Raw dataset shape: {raw_dataset.shape}")
print(f"Raw feature count: {len(raw_feature_columns)}")
print("[PASSED] Required 18 behavioral input features confirmed.")


# ====================================================================================================
# 9. EXACT PRODUCTION INFERENCE
# ====================================================================================================

print("\n" + "=" * 100)
print("9. P001–P005 EXACT PRODUCTION INFERENCE")
print("=" * 100)

results = []

for patient_id in required_patients:

    print("\n" + "-" * 100)
    print(patient_id)
    print("-" * 100)

    patient_file = patient_paths[patient_id]

    patient_df = pd.read_csv(patient_file)

    if patient_df.empty:
        raise RuntimeError(
            f"{patient_id} behavioral input file is empty."
        )

    print(f"Input file shape: {patient_df.shape}")

    # ----------------------------------------------------------------------------------------------
    # RAW INPUT
    # ----------------------------------------------------------------------------------------------

    missing = [
        c for c in raw_feature_columns
        if c not in patient_df.columns
    ]

    if missing:
        raise RuntimeError(
            f"{patient_id}: missing required behavioral columns: {missing}"
        )

    raw_input = patient_df[raw_feature_columns].copy()

    # Production input must represent exactly one patient.
    if len(raw_input) != 1:
        print(
            f"[INFO] {patient_id} contains {len(raw_input)} rows; "
            f"using the first production record."
        )

    raw_input = raw_input.iloc[[0]].copy()

    print(f"Raw input shape      : {raw_input.shape}")

    # ----------------------------------------------------------------------------------------------
    # PREPROCESSOR: 18-D -> 87-D
    # ----------------------------------------------------------------------------------------------

    X_87 = preprocessor.transform(raw_input)

    if hasattr(X_87, "toarray"):
        X_87 = X_87.toarray()

    X_87 = np.asarray(
        X_87,
        dtype=np.float32
    )

    if X_87.shape != (1, 87):
        raise RuntimeError(
            f"{patient_id}: preprocessor produced "
            f"{X_87.shape}, expected (1, 87)."
        )

    print(f"Preprocessed shape   : {X_87.shape}")

    # ----------------------------------------------------------------------------------------------
    # ENCODER: 87-D -> 128-D
    # ----------------------------------------------------------------------------------------------

    X_tensor = torch.from_numpy(X_87).to(DEVICE)

    with torch.no_grad():

        embedding_128, encoder_logit = behavioral_encoder(
            X_tensor
        )

    if tuple(embedding_128.shape) != (1, 128):
        raise RuntimeError(
            f"{patient_id}: encoder produced "
            f"{tuple(embedding_128.shape)}, expected (1, 128)."
        )

    print(f"Embedding shape      : {tuple(embedding_128.shape)}")

    # ----------------------------------------------------------------------------------------------
    # ALIGNMENT: 128-D -> 32-D
    # ----------------------------------------------------------------------------------------------

    with torch.no_grad():

        aligned_32, reconstructed_128 = behavioral_alignment(
            embedding_128
        )

    if tuple(aligned_32.shape) != (1, 32):
        raise RuntimeError(
            f"{patient_id}: alignment produced "
            f"{tuple(aligned_32.shape)}, expected (1, 32)."
        )

    print(f"Aligned shape        : {tuple(aligned_32.shape)}")

    # ----------------------------------------------------------------------------------------------
    # SCALER: 32-D -> 32-D
    # ----------------------------------------------------------------------------------------------

    aligned_np = aligned_32.cpu().numpy()

    scaled_32 = scaler.transform(aligned_np)

    scaled_32 = np.asarray(
        scaled_32,
        dtype=np.float32
    )

    if scaled_32.shape != (1, 32):
        raise RuntimeError(
            f"{patient_id}: scaler produced "
            f"{scaled_32.shape}, expected (1, 32)."
        )

    print(f"Scaled shape         : {scaled_32.shape}")

    # ----------------------------------------------------------------------------------------------
    # EXPERT: 32-D -> 1 LOGIT
    # ----------------------------------------------------------------------------------------------

    scaled_tensor = torch.from_numpy(
        scaled_32
    ).to(DEVICE)

    with torch.no_grad():

        expert_logit = behavioral_expert(
            scaled_tensor
        )

        behavioral_probability = torch.sigmoid(
            expert_logit
        )

    probability = float(
        behavioral_probability.squeeze().cpu().item()
    )

    logit = float(
        expert_logit.squeeze().cpu().item()
    )

    if not np.isfinite(probability):
        raise RuntimeError(
            f"{patient_id}: non-finite behavioral probability."
        )

    if not 0.0 <= probability <= 1.0:
        raise RuntimeError(
            f"{patient_id}: probability outside [0,1]."
        )

    predicted_label = (
        "ASD"
        if probability >= 0.5
        else "NON-ASD"
    )

    # Expected label from M4 manifest.
    manifest_row = manifest[
        manifest["patient_id"] == patient_id
    ].iloc[0]

    expected_label = str(
        manifest_row["expected_demo_label"]
    )

    expected_numeric = int(
        manifest_row["expected_demo_label_numeric"]
    )

    predicted_numeric = int(
        probability >= 0.5
    )

    correct = (
        predicted_numeric == expected_numeric
    )

    print(f"Expert logit         : {logit:.8f}")
    print(f"Behavioral probability: {probability:.8f}")
    print(f"Predicted label      : {predicted_label}")
    print(f"Expected demo label  : {expected_label}")
    print(
        f"Demo label match     : "
        f"{'YES' if correct else 'NO'}"
    )

    results.append({
        "patient_id": patient_id,
        "behavioral_probability": probability,
        "behavioral_logit": logit,
        "predicted_label": predicted_label,
        "expected_demo_label": expected_label,
        "expected_demo_label_numeric": expected_numeric,
        "predicted_label_numeric": predicted_numeric,
        "demo_label_match": correct
    })


# ====================================================================================================
# 10. FINAL BEHAVIORAL RESULTS
# ====================================================================================================

print("\n" + "=" * 100)
print("10. FINAL BEHAVIORAL PRODUCTION RESULTS")
print("=" * 100)

results_df = pd.DataFrame(results)

print(
    results_df[
        [
            "patient_id",
            "behavioral_probability",
            "predicted_label",
            "expected_demo_label",
            "demo_label_match"
        ]
    ].to_string(index=False)
)


# ====================================================================================================
# 11. VALIDATION SUMMARY
# ====================================================================================================

print("\n" + "=" * 100)
print("11. VALIDATION SUMMARY")
print("=" * 100)

print("[PASSED] P001–P005 behavioral inputs resolved.")
print("[PASSED] 18-D raw behavioral input validated.")
print("[PASSED] Saved preprocessor produced 87-D representation.")
print("[PASSED] Exact encoder produced 128-D representation.")
print("[PASSED] Exact alignment produced 32-D representation.")
print("[PASSED] Saved scaler processed 32-D representation.")
print("[PASSED] Exact behavioral expert produced one logit.")
print("[PASSED] Sigmoid probabilities validated.")

matches = int(
    results_df["demo_label_match"].sum()
)

print(
    f"\nDemo-label matches: {matches}/{len(results_df)}"
)

print("\n" + "=" * 100)
print("M4.4.9.3.7 COMPLETED")
print("=" * 100)

print("""
BEHAVIORAL PRODUCTION PIPELINE VERIFIED:

RAW 18-D
   ↓
Saved Behavioral Preprocessor
   ↓
87-D
   ↓
Exact Behavioral Encoder
   ↓
128-D
   ↓
Exact Behavioral Alignment
   ↓
32-D
   ↓
Saved Behavioral Scaler
   ↓
32-D
   ↓
Exact Behavioral Expert
   ↓
1 Logit
   ↓
Sigmoid
   ↓
ASD / NON-ASD
""")

print("NO TRAINING PERFORMED.")
print("NO CHECKPOINT MODIFIED.")
print("NO PRODUCTION ARTIFACT OVERWRITTEN.")

print("\nNEXT:")
print("After this cell passes, behavioral representation/inference is DONE.")
print("We should move to the remaining multimodal integration work.")
print("=" * 100)
```

```python
# ====================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3.7
# BEHAVIORAL P001–P005 PRODUCTION INFERENCE
# VERIFIED PIPELINE — NO MODEL RECONSTRUCTION
# ====================================================================================================

import os
import json
import warnings
import numpy as np
import pandas as pd
import torch
import joblib

warnings.filterwarnings("ignore")

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"
BEHAVIORAL_DIR = os.path.join(PROJECT_DIR, "Behavioral_Model")
DEMO_DIR = os.path.join(
    PROJECT_DIR,
    "05_multimodal_fusion",
    "M4_demo_patients"
)

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3.7")
print("BEHAVIORAL P001–P005 PRODUCTION INFERENCE")
print("VERIFIED PIPELINE — NO MODEL RECONSTRUCTION")
print("=" * 100)
print(f"Device  : {DEVICE}")
print(f"Project : {PROJECT_DIR}")


# ----------------------------------------------------------------------------------------------------
# 1. CHECK REQUIRED VERIFIED OBJECTS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("1. CHECKING VERIFIED M4.4.9.3 OBJECTS")
print("=" * 100)

required_objects = {
    "behavioral_encoder": "M4.4.9.3.2",
    "behavioral_alignment": "M4.4.9.3.4",
    "behavioral_expert": "M4.4.9.3.6",
}

missing_objects = []

for obj_name, stage in required_objects.items():
    if obj_name in globals():
        print(f"[FOUND] {obj_name:<24} from {stage}")
    else:
        print(f"[MISSING] {obj_name:<22} required from {stage}")
        missing_objects.append(obj_name)

if missing_objects:
    raise RuntimeError(
        "\nVerified model objects are missing: "
        + ", ".join(missing_objects)
        + "\n\nDo NOT guess or reconstruct them here. "
          "Run the corresponding verified reconstruction cell first."
    )

print("[PASSED] All verified behavioral model objects are available.")


# ----------------------------------------------------------------------------------------------------
# 2. DISCOVER SAVED PREPROCESSOR + SCALER
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("2. DISCOVERING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)


def find_file_recursive(base_dir, filename):
    matches = []

    for root, dirs, files in os.walk(base_dir):
        for f in files:
            if f == filename:
                matches.append(os.path.join(root, f))

    if not matches:
        return None

    # Prefer the shortest path if duplicates somehow exist.
    matches.sort(key=lambda p: (len(p), p))
    return matches[0]


PREPROCESSOR_PATH = find_file_recursive(
    BEHAVIORAL_DIR,
    "behavioral_preprocessor.joblib"
)

SCALER_PATH = find_file_recursive(
    PROJECT_DIR,
    "M3_1_behavioral_scaler.joblib"
)

if PREPROCESSOR_PATH is None:
    raise FileNotFoundError(
        "Could not locate behavioral_preprocessor.joblib"
    )

if SCALER_PATH is None:
    raise FileNotFoundError(
        "Could not locate M3_1_behavioral_scaler.joblib"
    )

print(f"[FOUND] Preprocessor\n        {PREPROCESSOR_PATH}")
print(f"[FOUND] Scaler\n        {SCALER_PATH}")


# ----------------------------------------------------------------------------------------------------
# 3. LOAD SAVED PREPROCESSING ARTIFACTS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("3. LOADING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)

behavioral_preprocessor = joblib.load(PREPROCESSOR_PATH)
behavioral_scaler = joblib.load(SCALER_PATH)

print(f"Preprocessor type : {type(behavioral_preprocessor)}")
print(f"Scaler type       : {type(behavioral_scaler)}")

print(
    f"Preprocessor n_features_in_: "
    f"{getattr(behavioral_preprocessor, 'n_features_in_', 'N/A')}"
)

print(
    f"Scaler n_features_in_: "
    f"{getattr(behavioral_scaler, 'n_features_in_', 'N/A')}"
)

if getattr(behavioral_preprocessor, "n_features_in_", None) != 18:
    raise RuntimeError(
        "Unexpected behavioral preprocessor input dimension."
    )

if getattr(behavioral_scaler, "n_features_in_", None) != 32:
    raise RuntimeError(
        "Unexpected behavioral scaler input dimension."
    )

print("[PASSED] Saved preprocessing artifacts loaded.")


# ----------------------------------------------------------------------------------------------------
# 4. DISCOVER DEMO MANIFEST
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("4. DISCOVERING DEMO MANIFEST")
print("=" * 100)

manifest_path = os.path.join(
    DEMO_DIR,
    "M4_demo_patient_manifest.csv"
)

if not os.path.isfile(manifest_path):
    manifest_path = find_file_recursive(
        PROJECT_DIR,
        "M4_demo_patient_manifest.csv"
    )

if manifest_path is None:
    raise FileNotFoundError(
        "M4_demo_patient_manifest.csv could not be located."
    )

print(f"[FOUND] Demo manifest\n        {manifest_path}")

manifest = pd.read_csv(manifest_path)

required_manifest_columns = {
    "patient_id",
    "expected_demo_label",
    "expected_demo_label_numeric",
    "behavioral_file",
}

missing_manifest_columns = (
    required_manifest_columns - set(manifest.columns)
)

if missing_manifest_columns:
    raise RuntimeError(
        "Manifest is missing required columns: "
        + ", ".join(sorted(missing_manifest_columns))
    )

required_patients = ["P001", "P002", "P003", "P004", "P005"]

patient_rows = {}

for patient_id in required_patients:

    rows = manifest[
        manifest["patient_id"].astype(str).str.strip() == patient_id
    ]

    if len(rows) != 1:
        raise RuntimeError(
            f"Expected exactly one manifest row for {patient_id}, "
            f"found {len(rows)}."
        )

    patient_rows[patient_id] = rows.iloc[0]

    print(
        f"[FOUND] {patient_id} | "
        f"Expected label: {rows.iloc[0]['expected_demo_label']}"
    )

print("[PASSED] All five demo patients resolved.")


# ----------------------------------------------------------------------------------------------------
# 5. RESOLVE BEHAVIORAL INPUT FILES
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("5. RESOLVING BEHAVIORAL INPUT FILES")
print("=" * 100)


def resolve_manifest_file(relative_path):
    relative_path = str(relative_path).strip()

    candidates = [
        os.path.join(DEMO_DIR, relative_path),
        os.path.join(PROJECT_DIR, relative_path),
    ]

    for candidate in candidates:
        if os.path.isfile(candidate):
            return candidate

    basename = os.path.basename(relative_path)

    discovered = find_file_recursive(
        DEMO_DIR,
        basename
    )

    if discovered is not None:
        return discovered

    discovered = find_file_recursive(
        PROJECT_DIR,
        basename
    )

    if discovered is not None:
        return discovered

    return None


behavioral_files = {}

for patient_id in required_patients:

    relative_behavioral_path = patient_rows[
        patient_id
    ]["behavioral_file"]

    resolved = resolve_manifest_file(
        relative_behavioral_path
    )

    if resolved is None:
        raise FileNotFoundError(
            f"Could not resolve behavioral input for {patient_id}:\n"
            f"{relative_behavioral_path}"
        )

    behavioral_files[patient_id] = resolved

    print(f"[FOUND] {patient_id:<6} {resolved}")

print("[PASSED] All behavioral demo files resolved.")


# ----------------------------------------------------------------------------------------------------
# 6. BEHAVIORAL INPUT VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("6. VALIDATING BEHAVIORAL INPUTS")
print("=" * 100)


def normalize_behavioral_dataframe(df, patient_id):

    df = df.copy()

    # Remove completely empty columns/rows.
    df = df.dropna(axis=0, how="all")
    df = df.dropna(axis=1, how="all")

    if len(df) == 0:
        raise ValueError(
            f"{patient_id}: behavioral file contains no usable rows."
        )

    # If multiple rows are present, production inference requires
    # one patient record. Use the first actual data row.
    if len(df) > 1:
        print(
            f"[INFO] {patient_id}: "
            f"{len(df)} rows found; using first patient row."
        )

    row = df.iloc[[0]].copy()

    return row


behavioral_inputs = {}

for patient_id in required_patients:

    path = behavioral_files[patient_id]

    df = pd.read_csv(path)

    row = normalize_behavioral_dataframe(
        df,
        patient_id
    )

    print(
        f"[VALID] {patient_id:<6} "
        f"shape={row.shape} "
        f"columns={len(row.columns)}"
    )

    behavioral_inputs[patient_id] = row


# ----------------------------------------------------------------------------------------------------
# 7. RAW 18-D FEATURE VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("7. VALIDATING RAW 18-D BEHAVIORAL FEATURE SPACE")
print("=" * 100)

expected_raw_features = [
    "A1_Score",
    "A2_Score",
    "A3_Score",
    "A4_Score",
    "A5_Score",
    "A6_Score",
    "A7_Score",
    "A8_Score",
    "A9_Score",
    "A10_Score",
    "age",
    "gender",
    "ethnicity",
    "jundice",
    "austim",
    "contry_of_res",
    "used_app_before",
    "result",
]

for patient_id in required_patients:

    row = behavioral_inputs[patient_id]

    missing = [
        col for col in expected_raw_features
        if col not in row.columns
    ]

    if missing:

        # Try case/whitespace-insensitive matching.
        normalized_map = {
            str(c).strip().lower(): c
            for c in row.columns
        }

        recovered = {}

        for expected in missing:
            key = expected.strip().lower()

            if key in normalized_map:
                recovered[expected] = normalized_map[key]

        unresolved = [
            col for col in missing
            if col not in recovered
        ]

        if unresolved:
            raise RuntimeError(
                f"{patient_id}: missing required behavioral features:\n"
                + "\n".join(unresolved)
                + "\n\nAvailable columns:\n"
                + ", ".join(map(str, row.columns))
            )

        for expected, actual in recovered.items():
            row[expected] = row[actual]

    print(
        f"[PASSED] {patient_id}: "
        f"all 18 raw behavioral features available."
    )


# ----------------------------------------------------------------------------------------------------
# 8. PRODUCTION INFERENCE
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("8. RUNNING VERIFIED BEHAVIORAL PRODUCTION INFERENCE")
print("=" * 100)

behavioral_encoder.eval()
behavioral_alignment.eval()
behavioral_expert.eval()

results = []

for patient_id in required_patients:

    print(f"\n--- {patient_id} ---")

    row = behavioral_inputs[patient_id]

    # --------------------------------------------------------------------------------
    # RAW 18-D
    # --------------------------------------------------------------------------------

    raw_x = row[expected_raw_features].copy()

    # Preserve values exactly as supplied to the saved sklearn preprocessor.
    raw_x = raw_x.reset_index(drop=True)

    print(f"Raw input shape       : {raw_x.shape}")

    # --------------------------------------------------------------------------------
    # SAVED PREPROCESSOR
    # --------------------------------------------------------------------------------

    preprocessed = behavioral_preprocessor.transform(
        raw_x
    )

    if hasattr(preprocessed, "toarray"):
        preprocessed = preprocessed.toarray()

    preprocessed = np.asarray(
        preprocessed,
        dtype=np.float32
    )

    print(
        f"Preprocessed shape    : "
        f"{preprocessed.shape}"
    )

    if preprocessed.ndim != 2 or preprocessed.shape[1] != 87:
        raise RuntimeError(
            f"{patient_id}: expected preprocessor output "
            f"(1, 87), got {preprocessed.shape}"
        )

    if not np.isfinite(preprocessed).all():
        raise RuntimeError(
            f"{patient_id}: preprocessor produced non-finite values."
        )

    # --------------------------------------------------------------------------------
    # EXACT ENCODER
    # --------------------------------------------------------------------------------

    encoder_input = torch.from_numpy(
        preprocessed
    ).to(DEVICE)

    with torch.no_grad():

        encoder_output = behavioral_encoder(
            encoder_input
        )

    # The verified encoder returns representation + classifier output.
    if isinstance(encoder_output, (tuple, list)):

        embedding = encoder_output[0]

    elif isinstance(encoder_output, dict):

        possible_keys = [
            "embedding",
            "representation",
            "encoded",
            "features",
        ]

        embedding = None

        for key in possible_keys:
            if key in encoder_output:
                embedding = encoder_output[key]
                break

        if embedding is None:
            raise RuntimeError(
                f"{patient_id}: could not identify encoder embedding "
                f"from dictionary output."
            )

    else:

        raise RuntimeError(
            f"{patient_id}: unexpected encoder output type: "
            f"{type(encoder_output)}"
        )

    if embedding.shape != (1, 128):
        raise RuntimeError(
            f"{patient_id}: expected encoder embedding "
            f"(1, 128), got {tuple(embedding.shape)}"
        )

    print(
        f"Encoder embedding    : "
        f"{tuple(embedding.shape)}"
    )

    # --------------------------------------------------------------------------------
    # EXACT ALIGNMENT
    # --------------------------------------------------------------------------------

    alignment_output = behavioral_alignment(
        embedding
    )

    if isinstance(alignment_output, (tuple, list)):

        aligned = alignment_output[0]

    elif isinstance(alignment_output, dict):

        possible_keys = [
            "encoded",
            "embedding",
            "aligned",
            "representation",
            "latent",
        ]

        aligned = None

        for key in possible_keys:
            if key in alignment_output:
                aligned = alignment_output[key]
                break

        if aligned is None:
            raise RuntimeError(
                f"{patient_id}: could not identify 32-D "
                f"alignment output."
            )

    else:

        # The verified alignment architecture returns a tuple in some
        # implementations and only the encoder output in others.
        # If a plain tensor is returned, it must be 32-D.
        aligned = alignment_output

    if not torch.is_tensor(aligned):
        raise RuntimeError(
            f"{patient_id}: alignment output is not a tensor."
        )

    if aligned.shape != (1, 32):
        raise RuntimeError(
            f"{patient_id}: expected aligned representation "
            f"(1, 32), got {tuple(aligned.shape)}"
        )

    print(
        f"Aligned representation: "
        f"{tuple(aligned.shape)}"
    )

    # --------------------------------------------------------------------------------
    # SAVED 32-D SCALER
    # --------------------------------------------------------------------------------

    aligned_np = aligned.detach().cpu().numpy()

    if not np.isfinite(aligned_np).all():
        raise RuntimeError(
            f"{patient_id}: alignment produced non-finite values."
        )

    scaled = behavioral_scaler.transform(
        aligned_np
    )

    scaled = np.asarray(
        scaled,
        dtype=np.float32
    )

    if scaled.shape != (1, 32):
        raise RuntimeError(
            f"{patient_id}: expected scaled shape "
            f"(1, 32), got {scaled.shape}"
        )

    print(
        f"Scaled representation : "
        f"{scaled.shape}"
    )

    # --------------------------------------------------------------------------------
    # EXACT BEHAVIORAL EXPERT
    # --------------------------------------------------------------------------------

    expert_input = torch.from_numpy(
        scaled
    ).to(DEVICE)

    with torch.no_grad():

        logits = behavioral_expert(
            expert_input
        )

    if not torch.is_tensor(logits):
        raise RuntimeError(
            f"{patient_id}: expert output is not a tensor."
        )

    if logits.shape != (1, 1):
        raise RuntimeError(
            f"{patient_id}: expected expert output "
            f"(1, 1), got {tuple(logits.shape)}"
        )

    probability = torch.sigmoid(
        logits
    ).item()

    if not np.isfinite(probability):
        raise RuntimeError(
            f"{patient_id}: probability is non-finite."
        )

    probability = float(
        np.clip(
            probability,
            0.0,
            1.0
        )
    )

    predicted_numeric = int(
        probability >= 0.5
    )

    predicted_label = (
        "ASD"
        if predicted_numeric == 1
        else "NON-ASD"
    )

    expected_label = str(
        patient_rows[patient_id][
            "expected_demo_label"
        ]
    )

    expected_numeric = int(
        patient_rows[patient_id][
            "expected_demo_label_numeric"
        ]
    )

    result_status = (
        "MATCH"
        if predicted_numeric == expected_numeric
        else "MISMATCH"
    )

    print(
        f"Logit                : {float(logits.item()):.8f}"
    )

    print(
        f"ASD probability      : {probability:.6f}"
    )

    print(
        f"Prediction            : {predicted_label}"
    )

    print(
        f"Expected demo label   : {expected_label}"
    )

    print(
        f"Expected numeric      : {expected_numeric}"
    )

    print(
        f"Demo comparison       : {result_status}"
    )

    results.append({
        "patient_id": patient_id,
        "behavioral_probability": probability,
        "behavioral_prediction": predicted_label,
        "behavioral_prediction_numeric": predicted_numeric,
        "expected_demo_label": expected_label,
        "expected_demo_label_numeric": expected_numeric,
        "demo_comparison": result_status,
    })


# ----------------------------------------------------------------------------------------------------
# 9. FINAL BEHAVIORAL RESULT TABLE
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("9. P001–P005 BEHAVIORAL PRODUCTION RESULTS")
print("=" * 100)

behavioral_results = pd.DataFrame(
    results
)

display(
    behavioral_results
)

print(
    f"\n[PASSED] "
    f"{len(behavioral_results)}/5 patients completed behavioral inference."
)


# ----------------------------------------------------------------------------------------------------
# 10. VALIDATION SUMMARY
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("10. PRODUCTION VALIDATION SUMMARY")
print("=" * 100)

if len(behavioral_results) != 5:
    raise RuntimeError(
        "Expected exactly five behavioral inference results."
    )

if not behavioral_results[
    "behavioral_probability"
].between(0.0, 1.0).all():

    raise RuntimeError(
        "One or more behavioral probabilities are outside [0,1]."
    )

print("[PASSED] All five patients produced valid ASD probabilities.")

match_count = int(
    (
        behavioral_results[
            "demo_comparison"
        ] == "MATCH"
    ).sum()
)

print(
    f"[INFO] Demo-label matches: "
    f"{match_count}/5"
)

print(
    "\nIMPORTANT: "
    "A MISMATCH is not treated as a model failure here. "
    "The purpose of this stage is to verify the production inference "
    "pipeline using the saved artifacts."
)


# ----------------------------------------------------------------------------------------------------
# 11. SAVE RESULTS
# ----------------------------------------------------------------------------------------------------

OUTPUT_DIR = os.path.join(
    DEMO_DIR,
    "M4_4_9_3_behavioral_results"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

RESULT_CSV = os.path.join(
    OUTPUT_DIR,
    "M4_4_9_3_7_behavioral_production_results.csv"
)

behavioral_results.to_csv(
    RESULT_CSV,
    index=False
)

print(
    f"\n[SAVED] Behavioral results:\n"
    f"        {RESULT_CSV}"
)


# ----------------------------------------------------------------------------------------------------
# 12. SAVE MACHINE-READABLE SUMMARY
# ----------------------------------------------------------------------------------------------------

summary = {
    "stage": "M4.4.9.3.7",
    "pipeline": [
        "raw_18D",
        "saved_preprocessor",
        "87D",
        "behavioral_encoder_128D",
        "behavioral_alignment_32D",
        "saved_behavioral_scaler",
        "behavioral_expert",
        "ASD_probability"
    ],
    "patients": required_patients,
    "patients_processed": len(behavioral_results),
    "demo_label_matches": match_count,
    "model_training_performed": False,
    "checkpoint_modified": False,
    "production_artifact_overwritten": False
}

SUMMARY_JSON = os.path.join(
    OUTPUT_DIR,
    "M4_4_9_3_7_behavioral_summary.json"
)

with open(
    SUMMARY_JSON,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        summary,
        f,
        indent=2
    )

print(
    f"[SAVED] Summary:\n"
    f"        {SUMMARY_JSON}"
)


# ----------------------------------------------------------------------------------------------------
# FINAL
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("M4.4.9.3.7 COMPLETED")
print("=" * 100)

print("[PASSED] P001–P005 behavioral inputs resolved.")
print("[PASSED] Saved 18-D → 87-D preprocessing executed.")
print("[PASSED] Exact 87-D → 128-D behavioral encoder executed.")
print("[PASSED] Exact 128-D → 32-D alignment executed.")
print("[PASSED] Saved 32-D behavioral scaler executed.")
print("[PASSED] Exact 32-D → 1 behavioral expert executed.")
print("[PASSED] ASD probabilities generated for all five patients.")
print("[PASSED] Production results saved.")

print("\nNEXT STAGE:")
print("M4.4.10 — MULTIMODAL FLEXIBLE INFERENCE")
print()
print("Supported modes:")
print("  BEHAVIORAL ONLY     → ASD probability")
print("  GENETIC ONLY        → ASD probability")
print("  ENVIRONMENTAL ONLY  → ASD probability")
print("  BEHAVIORAL + GENETIC")
print("  BEHAVIORAL + ENVIRONMENTAL")
print("  GENETIC + ENVIRONMENTAL")
print("  ALL THREE MODALITIES")
print()
print("The final stage will route whichever modalities are available")
print("and produce ONE common ASD / NON-ASD result.")
print("=" * 100)
```

```python
# ====================================================================================================
# ASD MULTIMODAL PROJECT — M4.4.9.3.7
# BEHAVIORAL P001–P005 PRODUCTION INFERENCE — FINAL SCHEMA FIX
# VERIFIED MODELS ONLY — NO MODEL RECONSTRUCTION
# ====================================================================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import joblib

print("=" * 100)
print("ASD MULTIMODAL PROJECT — M4.4.9.3.7")
print("BEHAVIORAL P001–P005 PRODUCTION INFERENCE — FINAL SCHEMA FIX")
print("=" * 100)

DEVICE = torch.device("cpu")

PROJECT_DIR = "/content/gdrive/MyDrive/ASD_Multimodal_Project"

# ----------------------------------------------------------------------------------------------------
# 1. VERIFY PREVIOUSLY RECONSTRUCTED MODEL OBJECTS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("1. CHECKING VERIFIED M4.4.9.3 OBJECTS")
print("=" * 100)

required_objects = {
    "behavioral_encoder": globals().get("behavioral_encoder"),
    "behavioral_alignment": globals().get("behavioral_alignment"),
    "behavioral_expert": globals().get("behavioral_expert"),
}

for name, obj in required_objects.items():
    if obj is None:
        raise RuntimeError(
            f"{name} is not available.\n"
            f"Run the verified M4.4.9.3 reconstruction cell for {name} first."
        )
    print(f"[FOUND] {name}")

print("[PASSED] All verified behavioral model objects are available.")


# ----------------------------------------------------------------------------------------------------
# 2. DISCOVER SAVED PREPROCESSING ARTIFACTS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("2. DISCOVERING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)

def find_file_recursive(base_dir, filename):
    matches = []

    for root, dirs, files in os.walk(base_dir):
        for f in files:
            if f == filename:
                matches.append(os.path.join(root, f))

    if not matches:
        raise FileNotFoundError(
            f"Could not locate:\n{filename}\ninside:\n{base_dir}"
        )

    return matches[0]


PREPROCESSOR_PATH = find_file_recursive(
    os.path.join(PROJECT_DIR, "Behavioral_Model"),
    "behavioral_preprocessor.joblib"
)

SCALER_PATH = find_file_recursive(
    PROJECT_DIR,
    "M3_1_behavioral_scaler.joblib"
)

print("[FOUND] Preprocessor")
print("       ", PREPROCESSOR_PATH)

print("[FOUND] Scaler")
print("       ", SCALER_PATH)


# ----------------------------------------------------------------------------------------------------
# 3. LOAD SAVED PREPROCESSING ARTIFACTS
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("3. LOADING SAVED PREPROCESSING ARTIFACTS")
print("=" * 100)

behavioral_preprocessor = joblib.load(PREPROCESSOR_PATH)
behavioral_scaler = joblib.load(SCALER_PATH)

print("Preprocessor type :", type(behavioral_preprocessor))
print("Scaler type       :", type(behavioral_scaler))

print(
    "Preprocessor n_features_in_:",
    getattr(behavioral_preprocessor, "n_features_in_", "N/A")
)

print(
    "Scaler n_features_in_:",
    getattr(behavioral_scaler, "n_features_in_", "N/A")
)

if getattr(behavioral_preprocessor, "n_features_in_", None) != 18:
    raise RuntimeError(
        "Unexpected preprocessor input dimension. "
        "Expected 18."
    )

if getattr(behavioral_scaler, "n_features_in_", None) != 32:
    raise RuntimeError(
        "Unexpected scaler input dimension. "
        "Expected 32."
    )

print("[PASSED] Saved preprocessing artifacts loaded.")


# ----------------------------------------------------------------------------------------------------
# 4. DISCOVER DEMO MANIFEST
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("4. DISCOVERING DEMO MANIFEST")
print("=" * 100)

MANIFEST_PATH = find_file_recursive(
    PROJECT_DIR,
    "M4_demo_patient_manifest.csv"
)

manifest = pd.read_csv(MANIFEST_PATH)

required_manifest_columns = {
    "patient_id",
    "expected_demo_label",
    "expected_demo_label_numeric",
    "behavioral_file",
}

missing_manifest_columns = (
    required_manifest_columns - set(manifest.columns)
)

if missing_manifest_columns:
    raise RuntimeError(
        f"Manifest is missing required columns:\n"
        f"{missing_manifest_columns}"
    )

required_patients = ["P001", "P002", "P003", "P004", "P005"]

manifest = manifest[
    manifest["patient_id"].astype(str).isin(required_patients)
].copy()

manifest["patient_id"] = manifest["patient_id"].astype(str)

manifest = (
    manifest
    .set_index("patient_id")
    .reindex(required_patients)
    .reset_index()
)

if manifest["patient_id"].isna().any():
    raise RuntimeError("One or more required demo patients are missing.")

for _, row in manifest.iterrows():
    print(
        f"[FOUND] {row['patient_id']} | "
        f"Expected label: {row['expected_demo_label']}"
    )

print("[PASSED] All five demo patients resolved.")


# ----------------------------------------------------------------------------------------------------
# 5. RESOLVE BEHAVIORAL INPUT FILES
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("5. RESOLVING BEHAVIORAL INPUT FILES")
print("=" * 100)

patient_files = {}

for _, row in manifest.iterrows():

    patient_id = str(row["patient_id"])
    relative_path = str(row["behavioral_file"])

    candidate = os.path.join(
        os.path.dirname(MANIFEST_PATH),
        relative_path
    )

    if not os.path.isfile(candidate):

        # Robust fallback: search by patient directory
        patient_dir = os.path.join(
            os.path.dirname(MANIFEST_PATH),
            patient_id
        )

        fallback = os.path.join(
            patient_dir,
            "behavioral_input.csv"
        )

        if os.path.isfile(fallback):
            candidate = fallback

    if not os.path.isfile(candidate):
        raise FileNotFoundError(
            f"Behavioral input missing for {patient_id}:\n"
            f"{candidate}"
        )

    patient_files[patient_id] = candidate

    print(f"[FOUND] {patient_id:<6} {candidate}")

print("[PASSED] All behavioral demo files resolved.")


# ----------------------------------------------------------------------------------------------------
# 6. DETERMINE EXACT RAW SCHEMA FROM SAVED PREPROCESSOR
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("6. DETERMINING EXACT RAW FEATURE SCHEMA")
print("=" * 100)

# The preprocessor was fitted on a pandas DataFrame.
# Recover its actual column schema instead of guessing.

preprocessor_columns = []

if hasattr(behavioral_preprocessor, "feature_names_in_"):
    preprocessor_columns = list(
        behavioral_preprocessor.feature_names_in_
    )

else:
    # Recover columns from ColumnTransformer transformers
    if hasattr(behavioral_preprocessor, "transformers_"):

        for _, transformer, columns in behavioral_preprocessor.transformers_:

            if columns is None or columns == "drop":
                continue

            if isinstance(columns, (list, tuple, np.ndarray, pd.Index)):
                preprocessor_columns.extend(
                    list(columns)
                )

print("Preprocessor feature columns:")

for i, col in enumerate(preprocessor_columns, start=1):
    print(f"  {i:02d}. {col}")

if len(preprocessor_columns) != 18:
    raise RuntimeError(
        f"Could not recover exactly 18 preprocessor columns.\n"
        f"Recovered: {len(preprocessor_columns)}"
    )

print("[PASSED] Exact 18-feature preprocessor schema recovered.")


# ----------------------------------------------------------------------------------------------------
# 7. IMPORTANT SCHEMA RULE
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("7. RAW INPUT SCHEMA RULE")
print("=" * 100)

print(
    """
The behavioral dataset contains:
    18 production input features
    + relation
    + Class/ASD target

The saved preprocessor expects the EXACT 18 production features.

Therefore:
    relation IS retained if the saved preprocessor expects it.
    Class/ASD IS excluded because it is the target.

No feature is manually dropped based on assumptions.
The saved preprocessor schema is authoritative.
"""
)

TARGET_COLUMN = "Class/ASD"

if TARGET_COLUMN in preprocessor_columns:
    raise RuntimeError(
        "Target column Class/ASD unexpectedly appears inside "
        "the saved production preprocessor schema."
    )

print("[PASSED] Target column excluded from production features.")


# ----------------------------------------------------------------------------------------------------
# 8. RUN VERIFIED PRODUCTION INFERENCE
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("8. RUNNING VERIFIED BEHAVIORAL PRODUCTION INFERENCE")
print("=" * 100)

behavioral_results = []

behavioral_encoder.eval()
behavioral_alignment.eval()
behavioral_expert.eval()

for patient_id in required_patients:

    print("\n" + "-" * 100)
    print(f"{patient_id}")
    print("-" * 100)

    input_path = patient_files[patient_id]

    df = pd.read_csv(input_path)

    print("Input file shape :", df.shape)
    print("Input columns    :", list(df.columns))

    # ----------------------------------------------------------------------------------------------
    # 8.1 CHECK REQUIRED PRODUCTION FEATURES
    # ----------------------------------------------------------------------------------------------

    missing = [
        c for c in preprocessor_columns
        if c not in df.columns
    ]

    if missing:
        raise RuntimeError(
            f"{patient_id} is missing required production features:\n"
            f"{missing}\n\n"
            f"Available columns:\n{list(df.columns)}"
        )

    # ----------------------------------------------------------------------------------------------
    # 8.2 BUILD EXACT 18-D RAW INPUT
    # ----------------------------------------------------------------------------------------------

    raw_x = df[preprocessor_columns].copy()

    print("Raw production shape:", raw_x.shape)

    if raw_x.shape[1] != 18:
        raise RuntimeError(
            f"{patient_id}: expected 18 raw features, "
            f"got {raw_x.shape[1]}"
        )

    # ----------------------------------------------------------------------------------------------
    # 8.3 SAVED PREPROCESSOR
    # ----------------------------------------------------------------------------------------------

    preprocessed = behavioral_preprocessor.transform(
        raw_x
    )

    if hasattr(preprocessed, "toarray"):
        preprocessed = preprocessed.toarray()

    preprocessed = np.asarray(
        preprocessed,
        dtype=np.float32
    )

    print("Preprocessed shape :", preprocessed.shape)

    if preprocessed.shape[1] != 87:
        raise RuntimeError(
            f"{patient_id}: expected 87-D preprocessed representation, "
            f"got {preprocessed.shape[1]}"
        )

    # ----------------------------------------------------------------------------------------------
    # 8.4 EXACT BEHAVIORAL ENCODER
    # ----------------------------------------------------------------------------------------------

    x_tensor = torch.tensor(
        preprocessed,
        dtype=torch.float32,
        device=DEVICE
    )

    with torch.no_grad():

        encoder_output = behavioral_encoder(
            x_tensor
        )

    # Support either tuple output or direct embedding output.
    if isinstance(encoder_output, tuple):

        embedding = encoder_output[0]

        if len(encoder_output) > 1:
            encoder_logit = encoder_output[1]
        else:
            encoder_logit = None

    else:

        embedding = encoder_output
        encoder_logit = None

    if embedding.shape[1] != 128:
        raise RuntimeError(
            f"{patient_id}: encoder did not produce 128-D embedding."
        )

    print("Embedding shape   :", tuple(embedding.shape))

    # ----------------------------------------------------------------------------------------------
    # 8.5 EXACT BEHAVIORAL ALIGNMENT
    # ----------------------------------------------------------------------------------------------

    with torch.no_grad():

        aligned = behavioral_alignment.encoder(
            embedding
        )

    if aligned.shape[1] != 32:
        raise RuntimeError(
            f"{patient_id}: alignment did not produce 32-D representation."
        )

    print("Aligned shape      :", tuple(aligned.shape))

    # ----------------------------------------------------------------------------------------------
    # 8.6 SAVED 32-D SCALER
    # ----------------------------------------------------------------------------------------------

    aligned_np = aligned.detach().cpu().numpy()

    scaled = behavioral_scaler.transform(
        aligned_np
    )

    scaled = np.asarray(
        scaled,
        dtype=np.float32
    )

    if scaled.shape[1] != 32:
        raise RuntimeError(
            f"{patient_id}: scaled representation is not 32-D."
        )

    print("Scaled shape       :", scaled.shape)

    scaled_tensor = torch.tensor(
        scaled,
        dtype=torch.float32,
        device=DEVICE
    )

    # ----------------------------------------------------------------------------------------------
    # 8.7 EXACT BEHAVIORAL EXPERT
    # ----------------------------------------------------------------------------------------------

    with torch.no_grad():

        expert_logit = behavioral_expert(
            scaled_tensor
        )

    if expert_logit.shape[1] != 1:
        raise RuntimeError(
            f"{patient_id}: expert did not produce one logit."
        )

    probability = torch.sigmoid(
        expert_logit
    ).detach().cpu().numpy().reshape(-1)

    probability = float(probability[0])

    if not np.isfinite(probability):
        raise RuntimeError(
            f"{patient_id}: non-finite probability."
        )

    # ----------------------------------------------------------------------------------------------
    # 8.8 LABEL
    # ----------------------------------------------------------------------------------------------

    predicted_label = (
        "ASD"
        if probability >= 0.5
        else "NON-ASD"
    )

    expected_label = str(
        manifest.loc[
            manifest["patient_id"] == patient_id,
            "expected_demo_label"
        ].iloc[0]
    )

    expected_numeric = int(
        manifest.loc[
            manifest["patient_id"] == patient_id,
            "expected_demo_label_numeric"
        ].iloc[0]
    )

    predicted_numeric = int(
        probability >= 0.5
    )

    match = (
        predicted_numeric == expected_numeric
    )

    print("\nRESULT")
    print("  ASD probability :", f"{probability:.6f}")
    print("  Predicted label  :", predicted_label)
    print("  Expected label   :", expected_label)
    print("  Demo match       :", "YES" if match else "NO")

    behavioral_results.append({
        "patient_id": patient_id,
        "behavioral_probability": probability,
        "behavioral_prediction": predicted_label,
        "expected_label": expected_label,
        "expected_numeric": expected_numeric,
        "prediction_numeric": predicted_numeric,
        "demo_match": match
    })


# ----------------------------------------------------------------------------------------------------
# 9. RESULTS TABLE
# ----------------------------------------------------------------------------------------------------

results_df = pd.DataFrame(
    behavioral_results
)

print("\n" + "=" * 100)
print("9. BEHAVIORAL PRODUCTION RESULTS")
print("=" * 100)

print(
    results_df[
        [
            "patient_id",
            "behavioral_probability",
            "behavioral_prediction",
            "expected_label",
            "demo_match"
        ]
    ].to_string(index=False)
)


# ----------------------------------------------------------------------------------------------------
# 10. FINAL VALIDATION
# ----------------------------------------------------------------------------------------------------

print("\n" + "=" * 100)
print("10. FINAL BEHAVIORAL PIPELINE VALIDATION")
print("=" * 100)

assert len(results_df) == 5
assert results_df["behavioral_probability"].between(0, 1).all()

print("[PASSED] P001–P005 inference completed.")
print("[PASSED] All probabilities are finite and within [0, 1].")
print("[PASSED] All five patients generated one behavioral prediction.")

print("\nVerified production pipeline:")
print(
    "RAW 18-D"
    " → SAVED PREPROCESSOR"
    " → 87-D"
    " → EXACT ENCODER"
    " → 128-D"
    " → EXACT ALIGNMENT"
    " → 32-D"
    " → SAVED SCALER"
    " → EXACT EXPERT"
    " → ASD PROBABILITY"
)

print("\n" + "=" * 100)
print("M4.4.9.3.7 COMPLETED")
print("=" * 100)

print(
    """
NO TRAINING PERFORMED.
NO CHECKPOINT MODIFIED.
NO PRODUCTION ARTIFACT OVERWRITTEN.

IMPORTANT:
This cell fixes only the raw-schema mismatch.
The verified model architectures from M4.4.9.3.2,
M4.4.9.3.4 and M4.4.9.3.6 are reused exactly.
"""
)

# Keep results available for the next multimodal stage.
globals()["behavioral_production_results"] = results_df
```
